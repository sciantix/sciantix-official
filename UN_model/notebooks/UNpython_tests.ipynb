````python
import math
from dataclasses import dataclass, replace
from typing import Optional, Sequence

try:
    import matplotlib.pyplot as plt
except ModuleNotFoundError:
    plt = None


@dataclass
class UNParameters:
    # Operating conditions: stationary T and F
    temperature: float = 1600.0              # K
    fission_rate: float = 5.0e19             # fiss / (m3 s), for Rizk validation experiments (DN1, LHR ~100 kW/m)
    grain_radius: float = 5.0e-6             # m

    # If target_burnup_percent_fima is None, final_time is used directly.
    # If target_burnup_percent_fima is not None, final_time is overwritten in __post_init__.
    final_time: float = 24.0 * 3600.0        # s
    target_burnup_percent_fima: Optional[float] = None

    dt: float = 60.0                         # s
    n_modes: int = 40                        # Sciantix default

    # Gas production
    xe_yield: float = 0.24                   # atoms / fission
    precursor_factor: float = 1.0            # stable Xe

    # Xe diffusivity in UN, Rizk / SCIANTIX-UN
    D10: float = 1.56e-3                     # m2/s
    Q1: float = 4.94                         # eV
    A20: float = 1.21e-67
    B21: float = 25.87
    B22: float = -1.49
    B23: float = 0.0
    A30: float = 1.85e-39                    # m5

    # Boltzmann constants
    kB_eV: float = 8.617333262e-5            # eV/K, for Arrhenius
    kB_J: float = 1.380649e-23               # J/K, for pressure/vacancy equations

    use_sciantix_D2_zero: bool = True

    # Vacancy diffusivity in UN
    D10_vU: float = 1.35e-2                  # m2/s
    Q1_vU: float = 5.66                      # eV
    A20_vU: float = 1.32e-19
    B21_vU: float = -0.62                    # eV
    B22_vU: float = -0.04                    # eV2

    # Bubble / material parameters
    radius_in_lattice: float = 0.21e-9       # m
    omega_fg: float = 8.5e-29                # m3/atom
    lattice_parameter: float = 4.889e-10     # m
    gamma_b: float = 1.11                    # J/m2
    hydrostatic_stress: float = 0.0          # Pa

    # Numerical floor only to avoid singular p_eq = 2 gamma / R at R = 0.
    min_radius_for_pressure: float = 1.0e-15 # m

    # Nucleation
    f_n: float = 1.0e-6

    # Dislocations
    rho_d: float = 3.0e13                    # m-2
    K_d: float = 5.0e5                       # bub/m, Rizk value for UN
    r_d: float = 3.46e-10                    # m
    Z_d: float = 5.0

    # Initial microstructure
    # Fresh-fuel-like bulk population: no initial bulk bubbles.
    R_b: float = 0.0                         # m
    N_b: float = 0.0                         # bub/m3

    # Dislocation bubble population:
    # N_d is initialized from K_d * rho_d if N_d is None.
    # R_d = 0 means no initial bubble volume; it grows when gas is trapped.
    R_d: float = 0.0                         # m
    N_d: Optional[float] = None              # bub/m3; if None, N_d = K_d * rho_d

    # Initial gas concentrations
    c0: float = 0.0                          # at/m3
    mb0: float = 0.0                         # at/m3
    md0: float = 0.0                         # at/m3

    # Initial vacancy concentrations in bubbles.
    # If None, they are initialized from initial R and N.
    nvb0: Optional[float] = None             # vac/m3
    nvd0: Optional[float] = None             # vac/m3

    # Scaling factors
    diffusivity_sf: float = 1.0
    trapping_sf: float = 1.0
    resolution_sf: float = 1.0

    # Numerical switches
    vacancy_absorption_only: bool = True
    update_bulk_vacancies: bool = True
    min_number_density: float = 0.0
    min_volume: float = 0.0

    def __post_init__(self):
        if self.N_d is None:
            self.N_d = self.K_d * self.rho_d

        if self.target_burnup_percent_fima is not None:
            self.final_time = burnup_percent_to_time(
                self.target_burnup_percent_fima,
                self.fission_rate,
                self.lattice_parameter,
            )


def omega_matrix(p):
    return p.lattice_parameter**3 / 4.0


def uranium_atom_density_from_lattice(lattice_parameter):
    # Rock-salt UN has 4 uranium atoms per unit cell.
    return 4.0 / lattice_parameter**3


def burnup_percent_to_time(burnup_percent_fima, fission_rate, lattice_parameter):
    if fission_rate <= 0.0:
        raise ValueError("fission_rate must be positive to compute time from burnup.")

    burnup_fraction_fima = burnup_percent_fima / 100.0
    uranium_density = uranium_atom_density_from_lattice(lattice_parameter)

    return burnup_fraction_fima * uranium_density / fission_rate


def time_to_burnup_percent(time, fission_rate, lattice_parameter):
    uranium_density = uranium_atom_density_from_lattice(lattice_parameter)
    return 100.0 * fission_rate * time / uranium_density


def sphere_volume(R):
    if R <= 0.0:
        return 0.0
    return (4.0 / 3.0) * math.pi * R**3


def radius_from_volume(V):
    if V <= 0.0:
        return 0.0
    return (3.0 * V / (4.0 * math.pi)) ** (1.0 / 3.0)


def xe_diffusivity_UN(p):
    T = p.temperature
    F = p.fission_rate
    kBT = p.kB_eV * T

    D1 = p.D10 * math.exp(-p.Q1 / kBT)

    if p.use_sciantix_D2_zero:
        D2 = 0.0
    else:
        D2 = p.A20 * math.sqrt(F) * math.exp(
            -p.B21 / kBT
            - p.B22 / kBT**2
            - p.B23 / kBT**3
        )

    D3 = p.A30 * F
    Dg = (D1 + D2 + D3) * p.diffusivity_sf * p.precursor_factor

    return Dg, {"D1": D1, "D2": D2, "D3": D3, "Dg": Dg}


def vacancy_diffusivity_UN(p):
    T = p.temperature
    F = p.fission_rate
    kBT = p.kB_eV * T

    D1 = p.D10_vU * math.exp(-p.Q1_vU / kBT)

    # A20_vU_fig4_refit is a refit of the 'Vacancies, bulk' curve in Rizk Fig. 4;
    # the tabulated A20_vU = 1.32e-19 does not reproduce that curve.
    A20_vU_fig4_refit = 4.6304523933553033e-29
    D2 = math.sqrt(F) * A20_vU_fig4_refit * math.exp(
        p.B21_vU / kBT
        + p.B22_vU / kBT**2
    )

    Dv = D1 + D2

    return Dv, {"Dv1": D1, "Dv2": D2, "Dv": Dv}


def b0_resolution(R):
    R = max(R, 1.0e-15)
    return 1.0e-25 * (
        2.64
        - 2.02 * math.exp(-2.61e-9 / R)
    )


def resolution_rates_UN(p, R_b, R_d):
    Rb_eff = R_b + p.radius_in_lattice
    Rd_eff = R_d + p.radius_in_lattice

    b_b = p.fission_rate * b0_resolution(Rb_eff) * p.resolution_sf
    b_d = p.fission_rate * b0_resolution(Rd_eff) * p.resolution_sf

    return b_b, b_d


def trapping_rates_UN(p, Dg, R_b, N_b, R_d, N_d):
    Rb_eff = R_b + p.radius_in_lattice
    Rd_eff = R_d + p.radius_in_lattice

    if N_b <= 0.0:
        g_b = 0.0
    else:
        g_b = 4.0 * math.pi * Dg * Rb_eff * N_b

    Gamma_d = 1.0 / math.sqrt(math.pi * p.rho_d)

    den = math.log(Gamma_d / (p.Z_d * p.r_d)) - 3.0 / 5.0
    if den <= 0.0:
        raise ValueError(f"Invalid dislocation sink denominator: {den:g}")

    free_dislocation = max(p.rho_d - 2.0 * R_d * N_d, 0.0)

    term_bubbles = 4.0 * math.pi * Dg * Rd_eff * N_d
    term_dislocation = (2.0 * math.pi * Dg / den) * free_dislocation

    g_d = term_bubbles + term_dislocation

    return p.trapping_sf * g_b, p.trapping_sf * g_d, {
        "Gamma_d": Gamma_d,
        "den": den,
        "free_dislocation": free_dislocation,
        "term_bubbles": term_bubbles,
        "term_dislocation": term_dislocation,
    }


def beta_production(p):
    return p.xe_yield * p.fission_rate


def nucleation_rate_bulk(p, Dg, c):
    return (
        8.0
        * math.pi
        * p.f_n
        * Dg
        * p.omega_fg ** (1.0 / 3.0)
        * max(c, 0.0) ** 2
    )


def phi_bulk(mb, Nb):
    if Nb <= 0.0:
        return 0.0

    atoms_per_bubble = mb / Nb

    if atoms_per_bubble <= 1.0:
        return 0.0

    return 1.0 / (atoms_per_bubble - 1.0)


def coalescence_lambda(Vd, Nd):
    xi = Vd * Nd
    xi = max(0.0, min(xi, 0.999999))

    return (2.0 - xi) / (2.0 * (1.0 - xi) ** 3)


def pressure_internal(p, m_gas, n_vac):
    if m_gas <= 0.0:
        return 0.0

    if n_vac <= 0.0:
        return math.inf

    denominator = n_vac * omega_matrix(p)

    if denominator <= 0.0:
        return math.inf

    return (
        p.kB_J
        * p.temperature
        * m_gas
        / denominator
    )


def pressure_equilibrium(p, R):
    R_eff = max(R, p.min_radius_for_pressure)
    return 2.0 * p.gamma_b / R_eff - p.hydrostatic_stress


def gas_only_radius_for_population(p, m_gas, N):
    """
    Provisional radius used only inside the vacancy absorption update.

    m_gas: gas concentration in the population [at/m3]
    N: bubble number density [bub/m3]

    Returns the radius obtained from the van der Waals gas volume only.
    """
    if m_gas <= 0.0 or N <= 0.0:
        return 0.0

    V_gas_per_bubble = p.omega_fg * m_gas / N

    if V_gas_per_bubble <= 0.0:
        return 0.0

    return radius_from_volume(V_gas_per_bubble)


def radius_for_vacancy_update(p, R_old, N, m_gas):
    """
    Radius used in p_eq and zeta during the vacancy update.

    Priority:
    1. use the current geometrical radius if positive;
    2. if R_old = 0 but gas exists, use gas-only provisional radius;
    3. if no gas exists, return 0 and skip vacancy absorption.
    """
    if R_old > 0.0:
        return R_old

    R_gas = gas_only_radius_for_population(p, m_gas, N)

    if R_gas > 0.0:
        return R_gas

    return 0.0


def pressure_equilibrium_for_vacancy_update(p, R_update):
    """
    Young-Laplace pressure for the vacancy update.

    If R_update <= 0, return inf. The caller should usually skip
    vacancy absorption in that case.
    """
    if R_update <= 0.0:
        return math.inf

    return 2.0 * p.gamma_b / R_update - p.hydrostatic_stress


def wigner_seitz_delta(N):
    return (3.0 / (4.0 * math.pi * max(N, 1.0))) ** (1.0 / 3.0)


def zeta_geometry(R, N):
    delta = wigner_seitz_delta(N)
    psi = max(R / delta, 1.0e-12)

    den = -psi**6 + 5.0 * psi**2 - 9.0 * psi + 5.0

    if den <= 1.0e-30:
        den = 1.0e-30

    return max(
        10.0 * psi * (1.0 + psi**3) / den,
        1.0e-30
    )


def vacancy_rate_explicit_diagnostic(p, Dv, R, N, m_gas, n_vac):
    """
    Explicit diagnostic rate.

    Same physics as before, but avoids evaluating zeta and p_eq at R=0
    when gas already exists.
    """
    if N <= 0.0 or m_gas <= 0.0:
        return 0.0

    R_update = radius_for_vacancy_update(p, R, N, m_gas)

    if R_update <= 0.0:
        return 0.0

    p_int = pressure_internal(p, m_gas, n_vac)
    p_eq = pressure_equilibrium_for_vacancy_update(p, R_update)

    driving = p_int - p_eq

    if p.vacancy_absorption_only and driving <= 0.0:
        return 0.0

    if not math.isfinite(driving):
        # Pressure is infinite because n_vac = 0.
        # Do not return inf here, because this is only a diagnostic.
        # The implicit solver below handles this case.
        return math.inf

    delta = wigner_seitz_delta(N)
    zeta = zeta_geometry(R_update, N)

    return (
        2.0
        * math.pi
        * Dv
        * delta
        * N
        / (p.kB_J * p.temperature * zeta)
        * driving
    )


def vacancy_concentration_implicit_step(p, Dv, R, N, m_gas, n_old, dt):
    """
    Implicit vacancy update.

    Same equation as before:

        n_new = n_old + dt*A*(C/n_new - p_eq)

    but p_eq and zeta are evaluated at a non-zero effective radius
    when gas has already entered a newly nucleated population.
    """
    if N <= 0.0 or m_gas <= 0.0:
        return n_old, 0.0

    R_update = radius_for_vacancy_update(p, R, N, m_gas)

    if R_update <= 0.0:
        return n_old, 0.0

    p_eq = pressure_equilibrium_for_vacancy_update(p, R_update)
    p_int_old = pressure_internal(p, m_gas, n_old)

    if p.vacancy_absorption_only and p_int_old <= p_eq:
        return n_old, 0.0

    delta = wigner_seitz_delta(N)
    zeta = zeta_geometry(R_update, N)

    A = (
        2.0
        * math.pi
        * Dv
        * delta
        * N
        / (p.kB_J * p.temperature * zeta)
    )

    C = p.kB_J * p.temperature * m_gas / omega_matrix(p)

    # Backward Euler:
    # n_new = n_old + dt*A*(C/n_new - p_eq)
    # n_new^2 - (n_old - dt*A*p_eq)*n_new - dt*A*C = 0
    B = n_old - dt * A * p_eq
    discriminant = B * B + 4.0 * dt * A * C

    if discriminant < 0.0:
        raise ValueError(f"Negative discriminant in vacancy implicit step: {discriminant:g}")

    sqrt_disc = math.sqrt(discriminant)

    # Numerically stable positive root.
    if B >= 0.0:
        n_new = 0.5 * (B + sqrt_disc)
    else:
        denominator = sqrt_disc - B
        if denominator <= 0.0:
            n_new = 0.0
        else:
            n_new = (2.0 * dt * A * C) / denominator

    if p.vacancy_absorption_only:
        n_new = max(n_new, n_old)

    dn_dt = (n_new - n_old) / dt

    return n_new, dn_dt


def initialize_vacancy_concentration(p, N, R, m_gas):
    if N <= 0.0 or R <= 0.0:
        return 0.0

    total_bubble_volume = N * sphere_volume(R)
    gas_volume = p.omega_fg * m_gas
    vacancy_volume = max(total_bubble_volume - gas_volume, 0.0)

    return vacancy_volume / omega_matrix(p)


def initialize_modes_from_average(average, n_modes, n_iter=20):
    modes = [0.0 for _ in range(n_modes)]

    projection_coeff = -math.sqrt(8.0 / math.pi)
    remainder = average

    for _ in range(n_iter):
        reconstructed = 0.0

        for i in range(n_modes):
            n = i + 1
            n_coeff = (-1.0) ** n / n

            modes[i] += projection_coeff * n_coeff * remainder

            reconstructed += (
                projection_coeff
                * n_coeff
                * modes[i]
                * 3.0
                / (4.0 * math.pi)
            )

        remainder = average - reconstructed

    return modes


def reconstruct_average(modes):
    projection_coeff = -2.0 * math.sqrt(2.0 / math.pi)

    average = 0.0

    for i, value in enumerate(modes):
        n = i + 1
        n_coeff = (-1.0) ** n / n

        average += (
            projection_coeff
            * n_coeff
            * value
            / ((4.0 / 3.0) * math.pi)
        )

    return average


def det3(A):
    return (
        A[0][0] * (A[1][1] * A[2][2] - A[1][2] * A[2][1])
        - A[0][1] * (A[1][0] * A[2][2] - A[1][2] * A[2][0])
        + A[0][2] * (A[1][0] * A[2][1] - A[1][1] * A[2][0])
    )


def solve3x3_cramer(A, b):
    detA = det3(A)

    if abs(detA) < 1.0e-300:
        raise ZeroDivisionError("Singular 3x3 system in spectral mode solve")

    Ax = [[b[i], A[i][1], A[i][2]] for i in range(3)]
    Ay = [[A[i][0], b[i], A[i][2]] for i in range(3)]
    Az = [[A[i][0], A[i][1], b[i]] for i in range(3)]

    return [
        det3(Ax) / detA,
        det3(Ay) / detA,
        det3(Az) / detA,
    ]


def sciantix_3x3_exchange_step(
    modes_c,
    modes_mb,
    modes_md,
    Dg,
    R,
    beta,
    g_b,
    g_d,
    b_b,
    b_d,
    dt,
):
    projection_coeff = -2.0 * math.sqrt(2.0 / math.pi)
    diffusion_rate_coeff = math.pi**2 * Dg / R**2

    for i in range(len(modes_c)):
        n = i + 1
        n_coeff = (-1.0) ** n / n

        diffusion_rate = diffusion_rate_coeff * n**2
        source_rate = projection_coeff * beta * n_coeff

        A = [
            [
                1.0 + (diffusion_rate + g_b + g_d) * dt,
                -b_b * dt,
                -b_d * dt,
            ],
            [
                -g_b * dt,
                1.0 + b_b * dt,
                0.0,
            ],
            [
                -g_d * dt,
                0.0,
                1.0 + b_d * dt,
            ],
        ]

        rhs = [
            modes_c[i] + source_rate * dt,
            modes_mb[i],
            modes_md[i],
        ]

        modes_c[i], modes_mb[i], modes_md[i] = solve3x3_cramer(A, rhs)

    return (
        reconstruct_average(modes_c),
        reconstruct_average(modes_mb),
        reconstruct_average(modes_md),
    )


def solve_UN_sciantix_intragranular_extended(p):
    modes_c = initialize_modes_from_average(p.c0, p.n_modes)
    modes_mb = initialize_modes_from_average(p.mb0, p.n_modes)
    modes_md = initialize_modes_from_average(p.md0, p.n_modes)

    R_b = p.R_b
    R_d = p.R_d

    N_b = p.N_b
    N_d = p.N_d

    V_b = sphere_volume(R_b)
    V_d = sphere_volume(R_d)

    if p.nvb0 is None:
        nvb = initialize_vacancy_concentration(p, N_b, R_b, p.mb0)
    else:
        nvb = p.nvb0

    if p.nvd0 is None:
        nvd = initialize_vacancy_concentration(p, N_d, R_d, p.md0)
    else:
        nvd = p.nvd0

    beta = beta_production(p)

    initial_gas = p.c0 + p.mb0 + p.md0
    generated = 0.0
    q_gb = 0.0
    retained = initial_gas

    hist = {
        "time": [0.0],
        "burnup_percent_fima": [
            time_to_burnup_percent(0.0, p.fission_rate, p.lattice_parameter)
        ],
        "c": [reconstruct_average(modes_c)],
        "mb": [reconstruct_average(modes_mb)],
        "md": [reconstruct_average(modes_md)],
        "Nb": [N_b],
        "Nd": [N_d],
        "Vb": [V_b],
        "Vd": [V_d],
        "Rb": [R_b],
        "Rd": [R_d],
        "nvb": [nvb],
        "nvd": [nvd],
        "generated": [generated],
        "retained": [retained],
        "q_gb": [q_gb],
        "qdot_gb": [0.0],
        "swelling_b": [N_b * V_b],
        "swelling_d": [N_d * V_d],
        "swelling_ig": [N_b * V_b + N_d * V_d],
        "p_b": [pressure_internal(p, p.mb0, nvb)],
        "p_d": [pressure_internal(p, p.md0, nvd)],
        "p_b_eq": [pressure_equilibrium(p, R_b)],
        "p_d_eq": [pressure_equilibrium(p, R_d)],
        "lambda_d": [coalescence_lambda(V_d, N_d)],
        "nu_b": [0.0],
        "phi_b": [0.0],
        "dnvb_dt": [0.0],
        "dnvd_dt": [0.0],
        "dnvb_dt_explicit_diagnostic": [0.0],
        "dnvd_dt_explicit_diagnostic": [0.0],
        "dVd_growth_dt": [0.0],
    }

    last_rates = {}
    n_steps = int(math.ceil(p.final_time / p.dt))

    for _ in range(n_steps):
        t_old = hist["time"][-1]
        dt = min(p.dt, p.final_time - t_old)

        if dt <= 0.0:
            break

        c_old = hist["c"][-1]
        mb_old = hist["mb"][-1]
        md_old = hist["md"][-1]

        Nb_old = N_b
        Nd_old = N_d
        Vd_old = V_d

        Dg, D_parts = xe_diffusivity_UN(p)
        Dv, Dv_parts = vacancy_diffusivity_UN(p)

        b_b, b_d = resolution_rates_UN(p, R_b, R_d)

        # Coefficienti old-step per diagnostica e per g_d.
        g_b_old, g_d, trapping_parts_old = trapping_rates_UN(
            p,
            Dg,
            R_b,
            Nb_old,
            R_d,
            Nd_old,
        )

        # --------------------------------------------------------------
        # Update N_b prima del gas solver:
        # dN_b/dt = nu_b - b_b phi_b N_b
        # --------------------------------------------------------------
        nu_b = nucleation_rate_bulk(p, Dg, c_old)
        phi_b = phi_bulk(mb_old, Nb_old)

        N_b = (Nb_old + dt * nu_b) / (1.0 + dt * b_b * phi_b)
        N_b = max(N_b, p.min_number_density)

        Nb_for_microstructure = N_b

        # --------------------------------------------------------------
        # Correzione importante:
        # dopo aver aggiornato N_b, ricalcolo g_b per il gas solver.
        # g_d resta calcolato con Nd_old, perché N_d cambia dopo.
        # --------------------------------------------------------------
        g_b, _, trapping_parts_updated_bulk = trapping_rates_UN(
            p,
            Dg,
            R_b,
            Nb_for_microstructure,
            R_d,
            Nd_old,
        )

        trapping_parts = dict(trapping_parts_old)
        trapping_parts["g_b_old_before_Nb_update"] = g_b_old
        trapping_parts["g_b_used_after_Nb_update"] = g_b
        trapping_parts["Nb_old"] = Nb_old
        trapping_parts["Nb_used_for_gb"] = Nb_for_microstructure

        # --------------------------------------------------------------
        # Gas diffusion / trapping / re-solution.
        # --------------------------------------------------------------
        c_new, mb_new, md_new = sciantix_3x3_exchange_step(
            modes_c,
            modes_mb,
            modes_md,
            Dg,
            p.grain_radius,
            beta,
            g_b,
            g_d,
            b_b,
            b_d,
            dt,
        )

        # Protezione contro piccoli valori negativi numerici.
        c_new = max(c_new, 0.0)
        mb_new = max(mb_new, 0.0)
        md_new = max(md_new, 0.0)

        dmb_dt = (mb_new - mb_old) / dt
        dmd_dt = (md_new - md_old) / dt

        # --------------------------------------------------------------
        # Vacancy equations.
        # --------------------------------------------------------------
        if p.update_bulk_vacancies:
            dnvb_dt_explicit = vacancy_rate_explicit_diagnostic(
                p,
                Dv,
                R_b,
                Nb_for_microstructure,
                mb_new,
                nvb,
            )

            nvb, dnvb_dt = vacancy_concentration_implicit_step(
                p,
                Dv,
                R_b,
                Nb_for_microstructure,
                mb_new,
                nvb,
                dt,
            )
        else:
            dnvb_dt_explicit = 0.0
            dnvb_dt = 0.0

        dnvd_dt_explicit = vacancy_rate_explicit_diagnostic(
            p,
            Dv,
            R_d,
            Nd_old,
            md_new,
            nvd,
        )

        nvd, dnvd_dt = vacancy_concentration_implicit_step(
            p,
            Dv,
            R_d,
            Nd_old,
            md_new,
            nvd,
            dt,
        )

        # --------------------------------------------------------------
        # Volume growth before dislocation bubble coalescence.
        # --------------------------------------------------------------
        if Nb_for_microstructure > 0.0:
            V_b_growth = V_b + dt * (
                p.omega_fg / Nb_for_microstructure * dmb_dt
                + omega_matrix(p) / Nb_for_microstructure * dnvb_dt
            )
            V_b_growth = max(V_b_growth, p.min_volume)
        else:
            V_b_growth = V_b

        if Nd_old > 0.0:
            dVd_growth_dt = (
                p.omega_fg / Nd_old * dmd_dt
                + omega_matrix(p) / Nd_old * dnvd_dt
            )
            V_d_growth = V_d + dt * dVd_growth_dt
            V_d_growth = max(V_d_growth, p.min_volume)
        else:
            dVd_growth_dt = 0.0
            V_d_growth = V_d

        # --------------------------------------------------------------
        # Dislocation bubble coalescence, rho_d costante:
        # dN_d/dt = -4 lambda N_d^2 dV_d/dt
        # --------------------------------------------------------------
        lambda_d = coalescence_lambda(Vd_old, Nd_old)
        dVd_positive = max(V_d_growth - Vd_old, 0.0)

        if dVd_positive > 0.0 and Nd_old > 0.0:
            N_d = Nd_old / (1.0 + 4.0 * lambda_d * Nd_old * dVd_positive)
        else:
            N_d = Nd_old

        N_d = max(N_d, p.min_number_density)

        # --------------------------------------------------------------
        # Reconstruct volume per bubble from gas + vacancy contents.
        # --------------------------------------------------------------
        if N_b > 0.0:
            V_b = (
                p.omega_fg * max(mb_new, 0.0)
                + omega_matrix(p) * nvb
            ) / N_b
        else:
            V_b = 0.0

        if N_d > 0.0:
            V_d = (
                p.omega_fg * max(md_new, 0.0)
                + omega_matrix(p) * nvd
            ) / N_d
        else:
            V_d = 0.0

        V_b = max(V_b, p.min_volume)
        V_d = max(V_d, p.min_volume)

        R_b = radius_from_volume(V_b)
        R_d = radius_from_volume(V_d)

        # --------------------------------------------------------------
        # Gas transported to grain boundary.
        # This is a cumulative balance term, not full intergranular model.
        # --------------------------------------------------------------
        generated += beta * dt
        retained = c_new + mb_new + md_new

        q_gb_old = q_gb
        q_gb = max(initial_gas + generated - retained, 0.0)
        qdot_gb = max((q_gb - q_gb_old) / dt, 0.0)

        hist["time"].append(t_old + dt)
        hist["burnup_percent_fima"].append(
            time_to_burnup_percent(t_old + dt, p.fission_rate, p.lattice_parameter)
        )
        hist["c"].append(c_new)
        hist["mb"].append(mb_new)
        hist["md"].append(md_new)
        hist["Nb"].append(N_b)
        hist["Nd"].append(N_d)
        hist["Vb"].append(V_b)
        hist["Vd"].append(V_d)
        hist["Rb"].append(R_b)
        hist["Rd"].append(R_d)
        hist["nvb"].append(nvb)
        hist["nvd"].append(nvd)
        hist["generated"].append(generated)
        hist["retained"].append(retained)
        hist["q_gb"].append(q_gb)
        hist["qdot_gb"].append(qdot_gb)
        hist["swelling_b"].append(N_b * V_b)
        hist["swelling_d"].append(N_d * V_d)
        hist["swelling_ig"].append(N_b * V_b + N_d * V_d)
        hist["p_b"].append(pressure_internal(p, mb_new, nvb))
        hist["p_d"].append(pressure_internal(p, md_new, nvd))
        hist["p_b_eq"].append(pressure_equilibrium(p, R_b))
        hist["p_d_eq"].append(pressure_equilibrium(p, R_d))
        hist["lambda_d"].append(lambda_d)
        hist["nu_b"].append(nu_b)
        hist["phi_b"].append(phi_b)
        hist["dnvb_dt"].append(dnvb_dt)
        hist["dnvd_dt"].append(dnvd_dt)
        hist["dnvb_dt_explicit_diagnostic"].append(dnvb_dt_explicit)
        hist["dnvd_dt_explicit_diagnostic"].append(dnvd_dt_explicit)
        hist["dVd_growth_dt"].append(dVd_growth_dt)

        last_rates = {
            "Dg": Dg,
            "Dv": Dv,
            "beta": beta,
            "g_b": g_b,
            "g_b_old_before_Nb_update": g_b_old,
            "g_d": g_d,
            "b_b": b_b,
            "b_d": b_d,
            "nu_b": nu_b,
            "phi_b": phi_b,
            "lambda_d": lambda_d,
            "dVd_growth_dt": dVd_growth_dt,
            "dnvb_dt": dnvb_dt,
            "dnvd_dt": dnvd_dt,
            "dnvb_dt_explicit_diagnostic": dnvb_dt_explicit,
            "dnvd_dt_explicit_diagnostic": dnvd_dt_explicit,
            "p_b": pressure_internal(p, mb_new, nvb),
            "p_d": pressure_internal(p, md_new, nvd),
            "p_b_eq": pressure_equilibrium(p, R_b),
            "p_d_eq": pressure_equilibrium(p, R_d),
            "Nb_old": Nb_old,
            "Nb_updated_before_gas": N_b,
            "Nd_old": Nd_old,
            "Nd_after_coalescence": N_d,
            **D_parts,
            **Dv_parts,
            **trapping_parts,
        }

    return hist, last_rates

def case_label(p):
    burnup = time_to_burnup_percent(p.final_time, p.fission_rate, p.lattice_parameter)
    return (
        f"T = {p.temperature:.0f} K, "
        f"F = {p.fission_rate:.3e} fiss m$^{{-3}}$ s$^{{-1}}$, "
        f"BU = {burnup:.3e}% FIMA"
    )


def safe_print_value(value):
    if isinstance(value, float) and math.isinf(value):
        return "inf"
    return f"{value:.6e}"


def print_initial_conditions(p, hist):
    print("\n" + "=" * 90)
    print("INITIAL CONDITIONS")
    print("=" * 90)
    print(f"{'temperature T':>28s} = {p.temperature:.6e} K")
    print(f"{'fission rate F':>28s} = {p.fission_rate:.6e} fiss/(m3 s)")
    print(f"{'grain radius':>28s} = {p.grain_radius:.6e} m")
    print(f"{'final time':>28s} = {p.final_time:.6e} s = {p.final_time/3600.0:.6e} h")
    print(f"{'target burnup':>28s} = {hist['burnup_percent_fima'][-1]:.6e} %FIMA")
    print(f"{'time step dt':>28s} = {p.dt:.6e} s")
    print(f"{'n_modes':>28s} = {p.n_modes:d}")
    print("-" * 90)
    print(f"{'c0':>28s} = {p.c0:.6e} at/m3")
    print(f"{'mb0':>28s} = {p.mb0:.6e} at/m3")
    print(f"{'md0':>28s} = {p.md0:.6e} at/m3")
    print(f"{'N_b0':>28s} = {p.N_b:.6e} bub/m3")
    print(f"{'N_d0':>28s} = {p.N_d:.6e} bub/m3")
    print(f"{'K_d':>28s} = {p.K_d:.6e} bub/m")
    print(f"{'K_d*rho_d':>28s} = {(p.K_d*p.rho_d):.6e} bub/m3")
    print(f"{'R_b0':>28s} = {p.R_b:.6e} m = {p.R_b*1e9:.6e} nm")
    print(f"{'R_d0':>28s} = {p.R_d:.6e} m = {p.R_d*1e9:.6e} nm")
    print(f"{'V_b0':>28s} = {hist['Vb'][0]:.6e} m3/bub")
    print(f"{'V_d0':>28s} = {hist['Vd'][0]:.6e} m3/bub")
    print(f"{'nvb0 initialized':>28s} = {hist['nvb'][0]:.6e} vac/m3")
    print(f"{'nvd0 initialized':>28s} = {hist['nvd'][0]:.6e} vac/m3")
    print(f"{'rho_d':>28s} = {p.rho_d:.6e} m-2")
    print(f"{'r_d':>28s} = {p.r_d:.6e} m")
    print(f"{'Z_d':>28s} = {p.Z_d:.6e}")
    print(f"{'gamma_b':>28s} = {p.gamma_b:.6e} J/m2")
    print(f"{'omega_fg':>28s} = {p.omega_fg:.6e} m3/atom")
    print(f"{'Omega matrix':>28s} = {omega_matrix(p):.6e} m3")
    print(f"{'N_U = 1/Omega':>28s} = {(1.0/omega_matrix(p)):.6e} U atoms/m3")
    print(f"{'vacancy_absorption_only':>28s} = {p.vacancy_absorption_only}")
    print(f"{'update_bulk_vacancies':>28s} = {p.update_bulk_vacancies}")


def print_final_state(hist, rates):
    print("\n" + "=" * 90)
    print("FINAL INTRAGRANULAR STATE")
    print("=" * 90)
    for key in [
        "burnup_percent_fima",
        "c",
        "mb",
        "md",
        "Nb",
        "Nd",
        "Vb",
        "Vd",
        "Rb",
        "Rd",
        "nvb",
        "nvd",
        "generated",
        "retained",
        "q_gb",
        "qdot_gb",
        "swelling_b",
        "swelling_d",
        "swelling_ig",
        "p_b",
        "p_d",
        "p_b_eq",
        "p_d_eq",
        "lambda_d",
    ]:
        print(f"{key:>24s} = {safe_print_value(hist[key][-1])}")

    print("\n" + "=" * 90)
    print("LAST-STEP RATES / COEFFICIENTS")
    print("=" * 90)
    for key in [
        "D1",
        "D2",
        "D3",
        "Dg",
        "Dv1",
        "Dv2",
        "Dv",
        "beta",
        "g_b",
        "g_d",
        "term_bubbles",
        "term_dislocation",
        "b_b",
        "b_d",
        "nu_b",
        "phi_b",
        "lambda_d",
        "dVd_growth_dt",
        "dnvb_dt",
        "dnvd_dt",
        "dnvb_dt_explicit_diagnostic",
        "dnvd_dt_explicit_diagnostic",
        "free_dislocation",
        "Gamma_d",
        "den",
    ]:
        if key in rates:
            print(f"{key:>32s} = {safe_print_value(rates[key])}")


def selected_indices(hist, n_rows=6):
    n = len(hist["time"])
    if n <= n_rows:
        return list(range(n))

    raw = [
        0,
        int(0.2 * (n - 1)),
        int(0.4 * (n - 1)),
        int(0.6 * (n - 1)),
        int(0.8 * (n - 1)),
        n - 1,
    ]

    indices = []
    for idx in raw:
        if idx not in indices:
            indices.append(idx)

    return indices


def print_table(title, hist, columns, n_rows=6):
    print("\n" + "-" * 90)
    print(title)
    print("-" * 90)

    indices = selected_indices(hist, n_rows=n_rows)

    header = f"{'time_h':>12s} | {'BU_%FIMA':>12s}"
    for name, _, _ in columns:
        header += f" | {name:>15s}"
    print(header)
    print("-" * len(header))

    for idx in indices:
        row = (
            f"{hist['time'][idx] / 3600.0:12.6e} | "
            f"{hist['burnup_percent_fima'][idx]:12.6e}"
        )
        for _, key, scale in columns:
            value = hist[key][idx] * scale
            if isinstance(value, float) and math.isinf(value):
                row += f" | {'inf':>15s}"
            else:
                row += f" | {value:15.6e}"
        print(row)


def print_mass_balance_summary(hist):
    print("\n" + "-" * 90)
    print("MASS BALANCE SUMMARY")
    print("-" * 90)
    print(f"{'generated final':>25s} = {hist['generated'][-1]:.6e} at/m3")
    print(f"{'retained final':>25s} = {hist['retained'][-1]:.6e} at/m3")
    print(f"{'q_gb final':>25s} = {hist['q_gb'][-1]:.6e} at/m3")
    print(f"{'retained + q_gb':>25s} = {(hist['retained'][-1] + hist['q_gb'][-1]):.6e} at/m3")
    print(f"{'balance error':>25s} = {(hist['generated'][-1] - hist['retained'][-1] - hist['q_gb'][-1]):.6e} at/m3")


def print_physical_sanity_checks(p, hist):
    print("\n" + "-" * 90)
    print("PHYSICAL SANITY CHECKS")
    print("-" * 90)

    final_Rb_nm = hist["Rb"][-1] * 1e9
    final_Rd_nm = hist["Rd"][-1] * 1e9
    final_sw = hist["swelling_ig"][-1]
    final_xi_d = hist["Vd"][-1] * hist["Nd"][-1]

    print(f"{'final R_b':>25s} = {final_Rb_nm:.6e} nm")
    print(f"{'final R_d':>25s} = {final_Rd_nm:.6e} nm")
    print(f"{'final xi_d = V_d*N_d':>25s} = {final_xi_d:.6e}")
    print(f"{'final gas swelling':>25s} = {final_sw:.6e}")

    if hist["Rd"][-1] > p.grain_radius:
        print("WARNING: R_d is larger than the grain radius. Result is physically suspicious.")

    if final_xi_d > 0.1:
        print("WARNING: xi_d is larger than 0.1. Coalescence correction may be outside its reliable range.")

    if final_sw > 0.5:
        print("WARNING: intragranular gas swelling is larger than 50%. Result is physically suspicious.")

    if hist["lambda_d"][-1] > 1.0e6:
        print("WARNING: lambda_d is extremely large. This usually means xi_d is close to 1.")


def plot_and_print_results(p, hist):
    hours = [t / 3600.0 for t in hist["time"]]

    total_grain = [
        c + mb + md
        for c, mb, md in zip(hist["c"], hist["mb"], hist["md"])
    ]

    print_table(
        "NUMERIC VALUES FOR PLOT 1: gas concentrations and gas to grain face",
        hist,
        [
            ("c", "c", 1.0),
            ("m_b", "mb", 1.0),
            ("m_d", "md", 1.0),
            ("retained", "retained", 1.0),
            ("q_gb", "q_gb", 1.0),
            ("qdot_gb", "qdot_gb", 1.0),
        ],
    )

    if plt is not None:
        plt.figure(figsize=(9, 5))
        plt.plot(hours, hist["c"], label="c: gas in solution")
        plt.plot(hours, hist["mb"], label="m_b: gas in bulk bubbles")
        plt.plot(hours, hist["md"], label="m_d: gas in dislocation bubbles")
        plt.plot(hours, total_grain, "--", label="total retained in grain")
        plt.plot(hours, hist["q_gb"], ":", label="q_gb to grain face")
        plt.title(f"Gas concentrations - {case_label(p)}")
        plt.xlabel("time (h)")
        plt.ylabel("concentration (at/m3)")
        plt.yscale("symlog", linthresh=1.0)
        plt.grid(True, which="both", alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()

    print_table(
        "NUMERIC VALUES FOR PLOT 2: bubble number densities",
        hist,
        [
            ("N_b", "Nb", 1.0),
            ("N_d", "Nd", 1.0),
            ("lambda_d", "lambda_d", 1.0),
            ("phi_b", "phi_b", 1.0),
            ("nu_b", "nu_b", 1.0),
        ],
    )

    if plt is not None:
        plt.figure(figsize=(9, 5))
        plt.plot(hours, hist["Nb"], label="N_b: bulk bubbles")
        plt.plot(hours, hist["Nd"], label="N_d: dislocation bubbles")
        plt.title(f"Bubble number densities - {case_label(p)}")
        plt.xlabel("time (h)")
        plt.ylabel("bubble number density (bub/m3)")
        plt.yscale("symlog", linthresh=1.0)
        plt.grid(True, which="both", alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()

    print_table(
        "NUMERIC VALUES FOR PLOT 3: bubble radii and volumes",
        hist,
        [
            ("R_b_nm", "Rb", 1.0e9),
            ("R_d_nm", "Rd", 1.0e9),
            ("V_b", "Vb", 1.0),
            ("V_d", "Vd", 1.0),
            ("dVd_dt", "dVd_growth_dt", 1.0),
        ],
    )

    if plt is not None:
        plt.figure(figsize=(9, 5))
        plt.plot(hours, [r * 1e9 for r in hist["Rb"]], label="R_b")
        plt.plot(hours, [r * 1e9 for r in hist["Rd"]], label="R_d")
        plt.title(f"Bubble radii - {case_label(p)}")
        plt.xlabel("time (h)")
        plt.ylabel("bubble radius (nm)")
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()

    print_table(
        "NUMERIC VALUES FOR PLOT 4: vacancy concentrations and pressure state",
        hist,
        [
            ("n_b", "nvb", 1.0),
            ("n_d", "nvd", 1.0),
            ("p_b", "p_b", 1.0),
            ("p_b_eq", "p_b_eq", 1.0),
            ("p_d", "p_d", 1.0),
            ("p_d_eq", "p_d_eq", 1.0),
        ],
    )

    if plt is not None:
        plt.figure(figsize=(9, 5))
        plt.plot(hours, hist["nvb"], label="n_b: vacancies in bulk bubbles")
        plt.plot(hours, hist["nvd"], label="n_d: vacancies in dislocation bubbles")
        plt.title(f"Vacancy concentrations - {case_label(p)}")
        plt.xlabel("time (h)")
        plt.ylabel("vacancy concentration (vac/m3)")
        plt.yscale("symlog", linthresh=1.0)
        plt.grid(True, which="both", alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()

        plt.figure(figsize=(9, 5))
        plt.plot(hours, hist["p_b"], label="p_b")
        plt.plot(hours, hist["p_b_eq"], "--", label="p_b_eq")
        plt.plot(hours, hist["p_d"], label="p_d")
        plt.plot(hours, hist["p_d_eq"], "--", label="p_d_eq")
        plt.title(f"Bubble pressures - {case_label(p)}")
        plt.xlabel("time (h)")
        plt.ylabel("pressure (Pa)")
        plt.yscale("symlog", linthresh=1.0)
        plt.grid(True, which="both", alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()

    print_table(
        "NUMERIC VALUES FOR PLOT 5: intragranular gaseous swelling indicators",
        hist,
        [
            ("sw_b", "swelling_b", 1.0),
            ("sw_d", "swelling_d", 1.0),
            ("sw_ig", "swelling_ig", 1.0),
        ],
    )

    if plt is not None:
        plt.figure(figsize=(9, 5))
        plt.plot(hours, hist["swelling_b"], label="bulk bubble swelling")
        plt.plot(hours, hist["swelling_d"], label="dislocation bubble swelling")
        plt.plot(hours, hist["swelling_ig"], "--", label="total intragranular gas swelling")
        plt.title(f"Intragranular gaseous swelling - {case_label(p)}")
        plt.xlabel("time (h)")
        plt.ylabel("swelling fraction (-)")
        plt.yscale("symlog", linthresh=1.0e-12)
        plt.grid(True, which="both", alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()


def run_temperature_sweep(
    base_params: UNParameters,
    temperatures: Sequence[float],
    burnup_percent_fima: float,
):
    results = []

    print("\n" + "=" * 100)
    print(f"TEMPERATURE SWEEP AT FIXED BURNUP = {burnup_percent_fima:.6e} %FIMA")
    print("=" * 100)
    print(
        "Nota: il modello intergranulare/grain-boundary non è ancora implementato. "
        "Qui vengono riportate solo bulk, dislocation e totale intragranulare."
    )

    for T in temperatures:
        p_T = replace(
            base_params,
            temperature=float(T),
            target_burnup_percent_fima=burnup_percent_fima,
        )

        hist, rates = solve_UN_sciantix_intragranular_extended(p_T)

        generated = hist["generated"][-1]
        q_fraction = hist["q_gb"][-1] / generated if generated > 0.0 else 0.0

        row = {
            "T": T,
            "burnup_percent_fima": hist["burnup_percent_fima"][-1],
            "time_h": p_T.final_time / 3600.0,
            "time_d": p_T.final_time / (3600.0 * 24.0),
            "swelling_b": hist["swelling_b"][-1],
            "swelling_d": hist["swelling_d"][-1],
            "swelling_ig": hist["swelling_ig"][-1],
            "swelling_b_percent": 100.0 * hist["swelling_b"][-1],
            "swelling_d_percent": 100.0 * hist["swelling_d"][-1],
            "swelling_ig_percent": 100.0 * hist["swelling_ig"][-1],
            "q_gb": hist["q_gb"][-1],
            "q_gb_fraction": q_fraction,
            "Rb_nm": hist["Rb"][-1] * 1.0e9,
            "Rd_nm": hist["Rd"][-1] * 1.0e9,
            "Nb": hist["Nb"][-1],
            "Nd": hist["Nd"][-1],
            "Dg": rates["Dg"],
            "Dv": rates["Dv"],
        }

        results.append(row)

    print(
        f"{'T_K':>10s} | {'BU_%':>12s} | {'time_d':>12s} | "
        f"{'bulk_%':>12s} | {'disl_%':>12s} | {'intra_%':>12s} | "
        f"{'qgb/gen':>12s} | {'Rb_nm':>12s} | {'Rd_nm':>12s} | "
        f"{'Nb':>12s} | {'Nd':>12s}"
    )
    print("-" * 150)

    for r in results:
        print(
            f"{r['T']:10.2f} | "
            f"{r['burnup_percent_fima']:12.6e} | "
            f"{r['time_d']:12.6e} | "
            f"{r['swelling_b_percent']:12.6e} | "
            f"{r['swelling_d_percent']:12.6e} | "
            f"{r['swelling_ig_percent']:12.6e} | "
            f"{r['q_gb_fraction']:12.6e} | "
            f"{r['Rb_nm']:12.6e} | "
            f"{r['Rd_nm']:12.6e} | "
            f"{r['Nb']:12.6e} | "
            f"{r['Nd']:12.6e}"
        )

    if plt is not None:
        Ts = [r["T"] for r in results]

        plt.figure(figsize=(8, 5))
        plt.plot(
            Ts,
            [r["swelling_b_percent"] for r in results],
            marker="o",
            linestyle=":",
            label="Bulk"
        )
        plt.plot(
            Ts,
            [r["swelling_d_percent"] for r in results],
            marker="o",
            linestyle="--",
            label="Dislocation"
        )
        plt.plot(
            Ts,
            [r["swelling_ig_percent"] for r in results],
            marker="o",
            linestyle="-",
            label="Bulk + dislocation"
        )

        plt.title(
            f"Intragranular fission gas swelling at {burnup_percent_fima:.2f}% FIMA"
        )
        plt.xlabel("T [K]")
        plt.ylabel("Fission gas swelling [%]")
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()

        plt.figure(figsize=(8, 5))
        plt.plot(
            Ts,
            [r["q_gb_fraction"] for r in results],
            marker="o",
            linestyle="-",
            label="q_gb / generated gas"
        )
        plt.title(
            f"Gas transferred to grain face at {burnup_percent_fima:.2f}% FIMA"
        )
        plt.xlabel("T [K]")
        plt.ylabel("fraction of generated gas reaching grain face [-]")
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()

    return results


if __name__ == "__main__":
    # Single run example.
    p = UNParameters(
        temperature=1600.0,
        fission_rate=5.0e19,  # for Rizk validation experiments
        grain_radius=5.0e-6,
        target_burnup_percent_fima=None,
        final_time=24.0 * 3600.0,
        dt=60.0,
        R_b=0.0,
        N_b=0.0,
        R_d=0.0,
        N_d=None,
        c0=0.0,
        mb0=0.0,
        md0=0.0,
        vacancy_absorption_only=True,
        update_bulk_vacancies=True,
    )

    hist, rates = solve_UN_sciantix_intragranular_extended(p)

    print_initial_conditions(p, hist)
    print_final_state(hist, rates)
    print_mass_balance_summary(hist)
    print_physical_sanity_checks(p, hist)
    plot_and_print_results(p, hist)

    # Temperature sweep at fixed burnup, analogous to the Rizk-style plot.
    # Grain-boundary/intergranular swelling is not yet included.
    base = UNParameters(
        fission_rate=5.0e19,  # for Rizk validation experiments
        grain_radius=5.0e-6,
        dt=3600.0,
        R_b=0.0,
        N_b=0.0,
        R_d=0.0,
        N_d=None,
        c0=0.0,
        mb0=0.0,
        md0=0.0,
        vacancy_absorption_only=True,
        update_bulk_vacancies=True,
    )

    temperatures = [float(T) for T in range(900, 2001, 50)]
    
    results_11_fima = run_temperature_sweep(
        base,
        temperatures,
        burnup_percent_fima=1.1,
    )

    if plt is None:
        print("\nmatplotlib is not installed: plots skipped, numeric tables printed.")
````

````python
# test inizializzazione / nucleazione bulk bubbles
#
# Da incollare sotto la cella principale del modello.
#
# Obiettivo:
# - testare vari R_b_seed per le bulk bubbles appena nucleate;
# - confrontare pressioni, swelling, N, R, gas partition con i grafici Rizk.
#
# Non modifica il solver principale: definisce un solver diagnostico separato.
# ============================================================

import math
from dataclasses import dataclass, replace

try:
    import matplotlib.pyplot as plt
except ModuleNotFoundError:
    plt = None


# ============================================================
# CONFIGURAZIONE
# ============================================================

FISSION_RATE = 5.0e19
GRAIN_RADIUS = 5.0e-6
DT_DIAGNOSTIC = 3600.0

TEMPERATURES_FINE = [float(T) for T in range(900, 2001, 50)]
TEMPERATURES_SCAN = [float(T) for T in range(900, 2001, 100)]

BURNUPS_SWELLING = [1.1, 1.3, 3.2]
BURNUP_PRESSURE = 3.2
BURNUP_NR = 1.3

# Target qualitativo per scegliere automaticamente un seed:
# in Rizk Fig. 5, bulk pressure a bassa T è sotto p_eq.
TARGET_PB_OVER_PEQ_AT_900K = 0.5


@dataclass
class SeedCase:
    name: str
    R_b_seed: float   # m, volume/vacancy seed per nuove bulk bubbles
    R_d0: float       # m, raggio iniziale dislocation bubbles
    description: str


SEED_CASES = [
    SeedCase(
        name="baseline",
        R_b_seed=0.0,
        R_d0=0.0,
        description="N_b0=0, R_b0=0, no bulk seed, R_d0=0"
    ),
    SeedCase(
        name="bulk_seed_0p5nm",
        R_b_seed=0.5e-9,
        R_d0=0.0,
        description="bulk seed 0.5 nm, R_d0=0"
    ),
    SeedCase(
        name="bulk_seed_1nm",
        R_b_seed=1.0e-9,
        R_d0=0.0,
        description="bulk seed 1 nm, R_d0=0"
    ),
    SeedCase(
        name="bulk_seed_2nm",
        R_b_seed=2.0e-9,
        R_d0=0.0,
        description="bulk seed 2 nm, R_d0=0"
    ),
    SeedCase(
        name="bulk_seed_3nm",
        R_b_seed=3.0e-9,
        R_d0=0.0,
        description="bulk seed 3 nm, R_d0=0"
    ),
    SeedCase(
        name="bulk_seed_1nm_Rd0_1nm",
        R_b_seed=1.0e-9,
        R_d0=1.0e-9,
        description="bulk seed 1 nm, dislocation bubbles initially R_d0=1 nm"
    ),
]


# ============================================================
# CONTROLLI PRELIMINARI
# ============================================================

_required_names = [
    "UNParameters",
    "omega_matrix",
    "sphere_volume",
    "radius_from_volume",
    "xe_diffusivity_UN",
    "b0_resolution",
    "resolution_rates_UN",
    "trapping_rates_UN",
    "beta_production",
    "nucleation_rate_bulk",
    "phi_bulk",
    "coalescence_lambda",
    "pressure_internal",
    "pressure_equilibrium",
    "wigner_seitz_delta",
    "zeta_geometry",
    "initialize_vacancy_concentration",
    "initialize_modes_from_average",
    "reconstruct_average",
    "sciantix_3x3_exchange_step",
    "time_to_burnup_percent",
]

_missing = [name for name in _required_names if name not in globals()]

if _missing:
    raise RuntimeError(
        "Prima devi eseguire la cella principale del modello. "
        f"Mancano questi nomi: {_missing}"
    )


# ============================================================
# DIFFUSIVITÀ VACANZE CORRETTA
# ============================================================

def vacancy_diffusivity_UN_corrected_for_diagnostic(p):
    """
    Diffusività corretta delle vacanze U.

    Usa:
    Dv = D1 + D2

    con A20 rifittato sulla curva 'Vacancies, bulk' di Rizk Fig. 4.
    """
    T = p.temperature
    F = p.fission_rate
    kBT = p.kB_eV * T

    D1 = p.D10_vU * math.exp(-p.Q1_vU / kBT)

    A20_vU_fig4_refit = 4.6304523933553033e-29

    D2 = math.sqrt(F) * A20_vU_fig4_refit * math.exp(
        p.B21_vU / kBT
        + p.B22_vU / (kBT**2)
    )

    Dv = D1 + D2

    return Dv, {
        "Dv1": D1,
        "Dv2": D2,
        "Dv": Dv,
    }


# ============================================================
# VACANCY UPDATE LOCALE PER IL SOLVER DIAGNOSTICO
# ============================================================

def gas_only_radius_for_population(p, m_gas, N):
    if m_gas <= 0.0 or N <= 0.0:
        return 0.0

    V_gas_per_bubble = p.omega_fg * m_gas / N

    if V_gas_per_bubble <= 0.0:
        return 0.0

    return radius_from_volume(V_gas_per_bubble)


def radius_for_vacancy_update(p, R_old, N, m_gas):
    if R_old > 0.0:
        return R_old

    R_gas = gas_only_radius_for_population(p, m_gas, N)

    if R_gas > 0.0:
        return R_gas

    return 0.0


def pressure_equilibrium_for_vacancy_update(p, R_update):
    if R_update <= 0.0:
        return math.inf

    return 2.0 * p.gamma_b / R_update - p.hydrostatic_stress


def vacancy_rate_explicit_diagnostic_local(p, Dv, R, N, m_gas, n_vac):
    if N <= 0.0 or m_gas <= 0.0:
        return 0.0

    R_update = radius_for_vacancy_update(p, R, N, m_gas)

    if R_update <= 0.0:
        return 0.0

    p_int = pressure_internal(p, m_gas, n_vac)
    p_eq = pressure_equilibrium_for_vacancy_update(p, R_update)

    driving = p_int - p_eq

    if p.vacancy_absorption_only and driving <= 0.0:
        return 0.0

    if not math.isfinite(driving):
        return math.inf

    delta = wigner_seitz_delta(N)
    zeta = zeta_geometry(R_update, N)

    return (
        2.0
        * math.pi
        * Dv
        * delta
        * N
        / (p.kB_J * p.temperature * zeta)
        * driving
    )


def vacancy_concentration_implicit_step_local(p, Dv, R, N, m_gas, n_old, dt):
    if N <= 0.0 or m_gas <= 0.0:
        return n_old, 0.0

    R_update = radius_for_vacancy_update(p, R, N, m_gas)

    if R_update <= 0.0:
        return n_old, 0.0

    p_eq = pressure_equilibrium_for_vacancy_update(p, R_update)
    p_int_old = pressure_internal(p, m_gas, n_old)

    if p.vacancy_absorption_only and p_int_old <= p_eq:
        return n_old, 0.0

    delta = wigner_seitz_delta(N)
    zeta = zeta_geometry(R_update, N)

    A = (
        2.0
        * math.pi
        * Dv
        * delta
        * N
        / (p.kB_J * p.temperature * zeta)
    )

    C = p.kB_J * p.temperature * m_gas / omega_matrix(p)

    # Backward Euler:
    # n_new = n_old + dt*A*(C/n_new - p_eq)
    # n_new^2 - (n_old - dt*A*p_eq)*n_new - dt*A*C = 0
    B = n_old - dt * A * p_eq
    discriminant = B * B + 4.0 * dt * A * C

    if discriminant < 0.0:
        raise ValueError(
            f"Negative discriminant in vacancy implicit step: {discriminant:g}"
        )

    sqrt_disc = math.sqrt(discriminant)

    if B >= 0.0:
        n_new = 0.5 * (B + sqrt_disc)
    else:
        denominator = sqrt_disc - B
        if denominator <= 0.0:
            n_new = 0.0
        else:
            n_new = (2.0 * dt * A * C) / denominator

    if p.vacancy_absorption_only:
        n_new = max(n_new, n_old)

    dn_dt = (n_new - n_old) / dt

    return n_new, dn_dt


# ============================================================
# SOLVER DIAGNOSTICO CON BULK SEED
# ============================================================

def solve_UN_seeded_bulk_nucleation(p, R_b_seed=0.0):
    """
    Solver diagnostico.

    Differenza rispetto al solver principale:
    quando N_b aumenta per nucleazione, aggiunge un volume/vacancy seed
    alle nuove bulk bubbles:

        Delta n_b_seed = Delta N_b_positive * V_seed / Omega

    dove:

        V_seed = 4/3 pi R_b_seed^3

    Tutte le altre equazioni restano uguali.
    """

    modes_c = initialize_modes_from_average(p.c0, p.n_modes)
    modes_mb = initialize_modes_from_average(p.mb0, p.n_modes)
    modes_md = initialize_modes_from_average(p.md0, p.n_modes)

    R_b = p.R_b
    R_d = p.R_d

    N_b = p.N_b
    N_d = p.N_d

    V_b = sphere_volume(R_b)
    V_d = sphere_volume(R_d)

    if p.nvb0 is None:
        nvb = initialize_vacancy_concentration(p, N_b, R_b, p.mb0)
    else:
        nvb = p.nvb0

    if p.nvd0 is None:
        nvd = initialize_vacancy_concentration(p, N_d, R_d, p.md0)
    else:
        nvd = p.nvd0

    beta = beta_production(p)

    initial_gas = p.c0 + p.mb0 + p.md0
    generated = 0.0
    q_gb = 0.0
    retained = initial_gas

    seed_vacancies_added_total = 0.0

    hist = {
        "time": [0.0],
        "burnup_percent_fima": [
            time_to_burnup_percent(0.0, p.fission_rate, p.lattice_parameter)
        ],
        "c": [reconstruct_average(modes_c)],
        "mb": [reconstruct_average(modes_mb)],
        "md": [reconstruct_average(modes_md)],
        "Nb": [N_b],
        "Nd": [N_d],
        "Vb": [V_b],
        "Vd": [V_d],
        "Rb": [R_b],
        "Rd": [R_d],
        "nvb": [nvb],
        "nvd": [nvd],
        "generated": [generated],
        "retained": [retained],
        "q_gb": [q_gb],
        "qdot_gb": [0.0],
        "swelling_b": [N_b * V_b],
        "swelling_d": [N_d * V_d],
        "swelling_ig": [N_b * V_b + N_d * V_d],
        "p_b": [pressure_internal(p, p.mb0, nvb)],
        "p_d": [pressure_internal(p, p.md0, nvd)],
        "p_b_eq": [pressure_equilibrium(p, R_b)],
        "p_d_eq": [pressure_equilibrium(p, R_d)],
        "lambda_d": [coalescence_lambda(V_d, N_d)],
        "nu_b": [0.0],
        "phi_b": [0.0],
        "dnvb_dt": [0.0],
        "dnvd_dt": [0.0],
        "dnvb_dt_explicit_diagnostic": [0.0],
        "dnvd_dt_explicit_diagnostic": [0.0],
        "dVd_growth_dt": [0.0],
        "seed_vacancies_added_total": [seed_vacancies_added_total],
    }

    last_rates = {}

    n_steps = int(math.ceil(p.final_time / p.dt))

    for _ in range(n_steps):
        t_old = hist["time"][-1]
        dt = min(p.dt, p.final_time - t_old)

        if dt <= 0.0:
            break

        c_old = hist["c"][-1]
        mb_old = hist["mb"][-1]
        md_old = hist["md"][-1]

        Nb_old = N_b
        Nd_old = N_d
        Vd_old = V_d

        Dg, D_parts = xe_diffusivity_UN(p)
        Dv, Dv_parts = vacancy_diffusivity_UN_corrected_for_diagnostic(p)

        b_b, b_d = resolution_rates_UN(p, R_b, R_d)

        g_b, g_d, trapping_parts = trapping_rates_UN(
            p,
            Dg,
            R_b,
            Nb_old,
            R_d,
            Nd_old,
        )

        # ------------------------------------------------------------
        # SCIANTIX-like order: update N_b before gas solve
        # ------------------------------------------------------------
        nu_b = nucleation_rate_bulk(p, Dg, c_old)
        phi_b = phi_bulk(mb_old, Nb_old)

        N_b = (Nb_old + dt * nu_b) / (1.0 + dt * b_b * phi_b)
        N_b = max(N_b, p.min_number_density)

        # ------------------------------------------------------------
        # NEW DIAGNOSTIC PART:
        # seed volume/vacancies for newly nucleated bulk bubbles
        # ------------------------------------------------------------
        dN_b_positive = max(N_b - Nb_old, 0.0)
        seed_vacancies_added_step = 0.0

        if dN_b_positive > 0.0 and R_b_seed > 0.0:
            V_seed = sphere_volume(R_b_seed)
            seed_vacancies_added_step = dN_b_positive * V_seed / omega_matrix(p)
            nvb += seed_vacancies_added_step
            seed_vacancies_added_total += seed_vacancies_added_step

        Nb_for_microstructure = N_b

        # ------------------------------------------------------------
        # Gas solve
        # ------------------------------------------------------------
        c_new, mb_new, md_new = sciantix_3x3_exchange_step(
            modes_c,
            modes_mb,
            modes_md,
            Dg,
            p.grain_radius,
            beta,
            g_b,
            g_d,
            b_b,
            b_d,
            dt,
        )

        dmb_dt = (mb_new - mb_old) / dt
        dmd_dt = (md_new - md_old) / dt

        # ------------------------------------------------------------
        # Vacancy equations
        # ------------------------------------------------------------
        if p.update_bulk_vacancies:
            dnvb_dt_explicit = vacancy_rate_explicit_diagnostic_local(
                p,
                Dv,
                R_b,
                Nb_for_microstructure,
                mb_new,
                nvb,
            )

            nvb, dnvb_dt = vacancy_concentration_implicit_step_local(
                p,
                Dv,
                R_b,
                Nb_for_microstructure,
                mb_new,
                nvb,
                dt,
            )
        else:
            dnvb_dt_explicit = 0.0
            dnvb_dt = 0.0

        dnvd_dt_explicit = vacancy_rate_explicit_diagnostic_local(
            p,
            Dv,
            R_d,
            Nd_old,
            md_new,
            nvd,
        )

        nvd, dnvd_dt = vacancy_concentration_implicit_step_local(
            p,
            Dv,
            R_d,
            Nd_old,
            md_new,
            nvd,
            dt,
        )

        # ------------------------------------------------------------
        # Volume growth before dislocation coalescence
        # ------------------------------------------------------------
        if Nb_for_microstructure > 0.0:
            V_b_growth = V_b + dt * (
                p.omega_fg / Nb_for_microstructure * dmb_dt
                + omega_matrix(p) / Nb_for_microstructure * dnvb_dt
            )
            V_b_growth = max(V_b_growth, p.min_volume)
        else:
            V_b_growth = V_b

        if Nd_old > 0.0:
            dVd_growth_dt = (
                p.omega_fg / Nd_old * dmd_dt
                + omega_matrix(p) / Nd_old * dnvd_dt
            )
            V_d_growth = V_d + dt * dVd_growth_dt
            V_d_growth = max(V_d_growth, p.min_volume)
        else:
            dVd_growth_dt = 0.0
            V_d_growth = V_d

        # ------------------------------------------------------------
        # Dislocation bubble coalescence
        # ------------------------------------------------------------
        lambda_d = coalescence_lambda(Vd_old, Nd_old)
        dVd_positive = max(V_d_growth - Vd_old, 0.0)

        if dVd_positive > 0.0 and Nd_old > 0.0:
            N_d = Nd_old / (1.0 + 4.0 * lambda_d * Nd_old * dVd_positive)
        else:
            N_d = Nd_old

        N_d = max(N_d, p.min_number_density)

        # ------------------------------------------------------------
        # Reconstruct volumes
        # ------------------------------------------------------------
        if N_b > 0.0:
            V_b = (
                p.omega_fg * max(mb_new, 0.0)
                + omega_matrix(p) * nvb
            ) / N_b
        else:
            V_b = 0.0

        if N_d > 0.0:
            V_d = (
                p.omega_fg * max(md_new, 0.0)
                + omega_matrix(p) * nvd
            ) / N_d
        else:
            V_d = 0.0

        V_b = max(V_b, p.min_volume)
        V_d = max(V_d, p.min_volume)

        R_b = radius_from_volume(V_b)
        R_d = radius_from_volume(V_d)

        # ------------------------------------------------------------
        # Mass balance to grain face
        # ------------------------------------------------------------
        generated += beta * dt
        retained = c_new + mb_new + md_new

        q_gb_old = q_gb
        q_gb = max(initial_gas + generated - retained, 0.0)
        qdot_gb = max((q_gb - q_gb_old) / dt, 0.0)

        hist["time"].append(t_old + dt)
        hist["burnup_percent_fima"].append(
            time_to_burnup_percent(t_old + dt, p.fission_rate, p.lattice_parameter)
        )
        hist["c"].append(c_new)
        hist["mb"].append(mb_new)
        hist["md"].append(md_new)
        hist["Nb"].append(N_b)
        hist["Nd"].append(N_d)
        hist["Vb"].append(V_b)
        hist["Vd"].append(V_d)
        hist["Rb"].append(R_b)
        hist["Rd"].append(R_d)
        hist["nvb"].append(nvb)
        hist["nvd"].append(nvd)
        hist["generated"].append(generated)
        hist["retained"].append(retained)
        hist["q_gb"].append(q_gb)
        hist["qdot_gb"].append(qdot_gb)
        hist["swelling_b"].append(N_b * V_b)
        hist["swelling_d"].append(N_d * V_d)
        hist["swelling_ig"].append(N_b * V_b + N_d * V_d)
        hist["p_b"].append(pressure_internal(p, mb_new, nvb))
        hist["p_d"].append(pressure_internal(p, md_new, nvd))
        hist["p_b_eq"].append(pressure_equilibrium(p, R_b))
        hist["p_d_eq"].append(pressure_equilibrium(p, R_d))
        hist["lambda_d"].append(lambda_d)
        hist["nu_b"].append(nu_b)
        hist["phi_b"].append(phi_b)
        hist["dnvb_dt"].append(dnvb_dt)
        hist["dnvd_dt"].append(dnvd_dt)
        hist["dnvb_dt_explicit_diagnostic"].append(dnvb_dt_explicit)
        hist["dnvd_dt_explicit_diagnostic"].append(dnvd_dt_explicit)
        hist["dVd_growth_dt"].append(dVd_growth_dt)
        hist["seed_vacancies_added_total"].append(seed_vacancies_added_total)

        last_rates = {
            "Dg": Dg,
            "Dv": Dv,
            "beta": beta,
            "g_b": g_b,
            "g_d": g_d,
            "b_b": b_b,
            "b_d": b_d,
            "nu_b": nu_b,
            "phi_b": phi_b,
            "lambda_d": lambda_d,
            "dVd_growth_dt": dVd_growth_dt,
            "dnvb_dt": dnvb_dt,
            "dnvd_dt": dnvd_dt,
            "dnvb_dt_explicit_diagnostic": dnvb_dt_explicit,
            "dnvd_dt_explicit_diagnostic": dnvd_dt_explicit,
            "p_b": pressure_internal(p, mb_new, nvb),
            "p_d": pressure_internal(p, md_new, nvd),
            "p_b_eq": pressure_equilibrium(p, R_b),
            "p_d_eq": pressure_equilibrium(p, R_d),
            "seed_vacancies_added_step": seed_vacancies_added_step,
            "seed_vacancies_added_total": seed_vacancies_added_total,
            **D_parts,
            **Dv_parts,
            **trapping_parts,
        }

    return hist, last_rates


# ============================================================
# RUN HELPERS
# ============================================================

def make_base_params_for_case(case):
    return UNParameters(
        fission_rate=FISSION_RATE,
        grain_radius=GRAIN_RADIUS,
        dt=DT_DIAGNOSTIC,
        R_b=0.0,
        N_b=0.0,
        R_d=case.R_d0,
        N_d=None,
        c0=0.0,
        mb0=0.0,
        md0=0.0,
        vacancy_absorption_only=True,
        update_bulk_vacancies=True,
    )


_run_cache = {}


def run_case(case, burnup_percent_fima, T):
    key = (case.name, float(burnup_percent_fima), float(T))

    if key in _run_cache:
        return _run_cache[key]

    base = make_base_params_for_case(case)

    p_T = replace(
        base,
        temperature=float(T),
        target_burnup_percent_fima=float(burnup_percent_fima),
    )

    hist, rates = solve_UN_seeded_bulk_nucleation(
        p_T,
        R_b_seed=case.R_b_seed,
    )

    generated = hist["generated"][-1]

    if generated > 0.0:
        matrix_gas_percent = 100.0 * hist["c"][-1] / generated
        bulk_gas_percent = 100.0 * hist["mb"][-1] / generated
        dislocation_gas_percent = 100.0 * hist["md"][-1] / generated
        qgb_gas_percent = 100.0 * hist["q_gb"][-1] / generated
        retained_gas_percent = 100.0 * hist["retained"][-1] / generated
    else:
        matrix_gas_percent = 0.0
        bulk_gas_percent = 0.0
        dislocation_gas_percent = 0.0
        qgb_gas_percent = 0.0
        retained_gas_percent = 0.0

    p_b = hist["p_b"][-1]
    p_b_eq = hist["p_b_eq"][-1]
    p_d = hist["p_d"][-1]
    p_d_eq = hist["p_d_eq"][-1]

    row = {
        "case": case.name,
        "description": case.description,
        "R_b_seed_nm": case.R_b_seed * 1.0e9,
        "R_d0_nm": case.R_d0 * 1.0e9,
        "burnup_percent_fima": burnup_percent_fima,
        "T": float(T),
        "time_d": p_T.final_time / (3600.0 * 24.0),

        "p_b": p_b,
        "p_b_eq": p_b_eq,
        "p_d": p_d,
        "p_d_eq": p_d_eq,
        "p_b_over_p_b_eq": p_b / p_b_eq if p_b_eq > 0.0 else math.nan,
        "p_d_over_p_d_eq": p_d / p_d_eq if p_d_eq > 0.0 else math.nan,

        "swelling_b_percent": 100.0 * hist["swelling_b"][-1],
        "swelling_d_percent": 100.0 * hist["swelling_d"][-1],
        "swelling_ig_percent": 100.0 * hist["swelling_ig"][-1],

        "Rb_nm": hist["Rb"][-1] * 1.0e9,
        "Rd_nm": hist["Rd"][-1] * 1.0e9,
        "Nb": hist["Nb"][-1],
        "Nd": hist["Nd"][-1],
        "Vb": hist["Vb"][-1],
        "Vd": hist["Vd"][-1],
        "nvb": hist["nvb"][-1],
        "nvd": hist["nvd"][-1],

        "matrix_gas_percent": matrix_gas_percent,
        "bulk_gas_percent": bulk_gas_percent,
        "dislocation_gas_percent": dislocation_gas_percent,
        "qgb_gas_percent": qgb_gas_percent,
        "retained_gas_percent": retained_gas_percent,

        "Dg": rates["Dg"],
        "Dv": rates["Dv"],
        "Dv1": rates["Dv1"],
        "Dv2": rates["Dv2"],
        "g_b": rates["g_b"],
        "g_d": rates["g_d"],
        "b_b": rates["b_b"],
        "b_d": rates["b_d"],
        "nu_b": rates["nu_b"],
        "phi_b": rates["phi_b"],
        "seed_vacancies_added_total": rates["seed_vacancies_added_total"],

        "hist": hist,
        "rates": rates,
    }

    _run_cache[key] = row

    return row


def run_grid(case, burnup_percent_fima, temperatures):
    return [
        run_case(case, burnup_percent_fima, T)
        for T in temperatures
    ]


def print_seed_scan_table(seed_scan):
    print("\n" + "=" * 150)
    print("SEED SCAN: pressione bulk a 3.2% FIMA")
    print("=" * 150)
    print(
        f"{'case':>24s} | "
        f"{'Rb_seed_nm':>10s} | "
        f"{'Rd0_nm':>8s} | "
        f"{'T_K':>8s} | "
        f"{'p_b/p_eq':>12s} | "
        f"{'p_d/p_eq':>12s} | "
        f"{'p_b':>12s} | "
        f"{'p_b_eq':>12s} | "
        f"{'sw_b_%':>10s} | "
        f"{'sw_d_%':>10s} | "
        f"{'Rb_nm':>10s} | "
        f"{'Rd_nm':>10s}"
    )
    print("-" * 150)

    for case_name, rows in seed_scan.items():
        for row in rows:
            if row["T"] in [900.0, 1200.0, 1600.0, 2000.0]:
                print(
                    f"{case_name:>24s} | "
                    f"{row['R_b_seed_nm']:10.3f} | "
                    f"{row['R_d0_nm']:8.3f} | "
                    f"{row['T']:8.1f} | "
                    f"{row['p_b_over_p_b_eq']:12.6e} | "
                    f"{row['p_d_over_p_d_eq']:12.6e} | "
                    f"{row['p_b']:12.6e} | "
                    f"{row['p_b_eq']:12.6e} | "
                    f"{row['swelling_b_percent']:10.4e} | "
                    f"{row['swelling_d_percent']:10.4e} | "
                    f"{row['Rb_nm']:10.4e} | "
                    f"{row['Rd_nm']:10.4e}"
                )


def choose_seed_case(seed_scan):
    best_name = None
    best_score = None

    for case_name, rows in seed_scan.items():
        row_900 = min(rows, key=lambda r: abs(r["T"] - 900.0))
        ratio = row_900["p_b_over_p_b_eq"]

        if ratio <= 0.0 or not math.isfinite(ratio):
            continue

        score = abs(math.log10(ratio / TARGET_PB_OVER_PEQ_AT_900K))

        if best_score is None or score < best_score:
            best_score = score
            best_name = case_name

    for case in SEED_CASES:
        if case.name == best_name:
            return case

    return SEED_CASES[0]


# ============================================================
# PLOT HELPERS
# ============================================================

def plot_seed_scan_pressure_ratios(seed_scan):
    if plt is None:
        return

    plt.figure(figsize=(9, 5))

    for case_name, rows in seed_scan.items():
        Ts = [r["T"] for r in rows]
        ratios = [r["p_b_over_p_b_eq"] for r in rows]
        label = case_name.replace("_", " ")

        plt.plot(Ts, ratios, marker="o", label=label)

    plt.axhline(1.0, linestyle="--", linewidth=1.0)
    plt.xlabel("T [K]")
    plt.ylabel(r"$p_b / p_{b,eq}$ [-]")
    plt.title("Bulk pressure ratio at 3.2% FIMA: seed scan")
    plt.grid(True, alpha=0.3)
    plt.legend(fontsize=8)
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(9, 5))

    for case_name, rows in seed_scan.items():
        Ts = [r["T"] for r in rows]
        ratios = [r["p_d_over_p_d_eq"] for r in rows]
        label = case_name.replace("_", " ")

        plt.plot(Ts, ratios, marker="o", label=label)

    plt.axhline(1.0, linestyle="--", linewidth=1.0)
    plt.xlabel("T [K]")
    plt.ylabel(r"$p_d / p_{d,eq}$ [-]")
    plt.title("Dislocation pressure ratio at 3.2% FIMA: seed scan")
    plt.grid(True, alpha=0.3)
    plt.legend(fontsize=8)
    plt.tight_layout()
    plt.show()


def plot_pressure_fig5_style(rows, title_suffix):
    if plt is None:
        return

    Ts = [r["T"] for r in rows]

    plt.figure(figsize=(8, 5))
    plt.plot(Ts, [r["p_b"] for r in rows], label="Bulk bubble pressure")
    plt.plot(Ts, [r["p_b_eq"] for r in rows], linestyle=":", label="Bulk bubble equilibrium pressure")
    plt.plot(Ts, [r["p_d"] for r in rows], label="Dislocation bubble pressure")
    plt.plot(Ts, [r["p_d_eq"] for r in rows], linestyle="--", label="Dislocation bubble equilibrium pressure")

    plt.yscale("log")
    plt.xlabel("T [K]")
    plt.ylabel("Bubble pressure [Pa]")
    plt.title(f"Bubble pressures at 3.2% FIMA - {title_suffix}")
    plt.grid(True, which="both", alpha=0.3)
    plt.legend(fontsize=8)
    plt.tight_layout()
    plt.show()


def plot_swelling_fig3_style(results_by_burnup, title_suffix):
    if plt is None:
        return

    for burnup, rows in results_by_burnup.items():
        Ts = [r["T"] for r in rows]
        bulk = [r["swelling_b_percent"] for r in rows]
        disl = [r["swelling_d_percent"] for r in rows]

        plt.figure(figsize=(8, 5))
        plt.plot(Ts, bulk, marker="o", linestyle=":", label="Bulk")
        plt.plot(Ts, disl, marker="o", linestyle="--", label="Dislocation")

        plt.xlabel("T [K]")
        plt.ylabel("Fission gas swelling [%]")
        plt.title(f"Swelling at {burnup:.1f}% FIMA - {title_suffix}")
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()


def plot_number_density_fig7_style(rows, title_suffix):
    if plt is None:
        return

    Ts = [r["T"] for r in rows]

    plt.figure(figsize=(8, 5))
    plt.plot(Ts, [r["Nb"] for r in rows], linestyle=":", marker="o", label="Bulk")
    plt.plot(Ts, [r["Nd"] for r in rows], linestyle="--", marker="o", label="Dislocation")

    plt.yscale("log")
    plt.xlabel("T [K]")
    plt.ylabel("Bubble concentration [m$^{-3}$]")
    plt.title(f"Bubble number density at 1.3% FIMA - {title_suffix}")
    plt.grid(True, which="both", alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()


def plot_radius_fig8_style(rows, title_suffix):
    if plt is None:
        return

    Ts = [r["T"] for r in rows]

    plt.figure(figsize=(8, 5))
    plt.plot(Ts, [r["Rb_nm"] for r in rows], linestyle=":", marker="o", label="Bulk")
    plt.plot(Ts, [r["Rd_nm"] for r in rows], linestyle="--", marker="o", label="Dislocation")

    plt.yscale("log")
    plt.xlabel("T [K]")
    plt.ylabel("Bubble radius [nm]")
    plt.title(f"Bubble radii at 1.3% FIMA - {title_suffix}")
    plt.grid(True, which="both", alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()


def plot_gas_partition_fig9_style(rows, burnup, title_suffix):
    if plt is None:
        return

    Ts = [r["T"] for r in rows]

    plt.figure(figsize=(8, 5))
    plt.plot(Ts, [r["matrix_gas_percent"] for r in rows], label="Matrix")
    plt.plot(Ts, [r["bulk_gas_percent"] for r in rows], linestyle=":", label="Bulk bubbles")
    plt.plot(Ts, [r["dislocation_gas_percent"] for r in rows], linestyle="--", label="Dislocation bubbles")
    plt.plot(Ts, [r["qgb_gas_percent"] for r in rows], linestyle="-.", label="Gas to grain face")

    plt.xlabel("T [K]")
    plt.ylabel("Amount of generated gas [%]")
    plt.title(f"Gas partition at {burnup:.1f}% FIMA - {title_suffix}")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()


def print_rows_compact(title, rows):
    print("\n" + "=" * 150)
    print(title)
    print("=" * 150)
    print(
        f"{'T_K':>8s} | "
        f"{'p_b/p_eq':>12s} | "
        f"{'p_d/p_eq':>12s} | "
        f"{'sw_b_%':>10s} | "
        f"{'sw_d_%':>10s} | "
        f"{'Rb_nm':>10s} | "
        f"{'Rd_nm':>10s} | "
        f"{'Nb':>12s} | "
        f"{'Nd':>12s} | "
        f"{'matrix_%':>10s} | "
        f"{'bulkgas_%':>10s} | "
        f"{'dislgas_%':>10s} | "
        f"{'qgb_%':>10s}"
    )
    print("-" * 150)

    for r in rows:
        print(
            f"{r['T']:8.1f} | "
            f"{r['p_b_over_p_b_eq']:12.6e} | "
            f"{r['p_d_over_p_d_eq']:12.6e} | "
            f"{r['swelling_b_percent']:10.4e} | "
            f"{r['swelling_d_percent']:10.4e} | "
            f"{r['Rb_nm']:10.4e} | "
            f"{r['Rd_nm']:10.4e} | "
            f"{r['Nb']:12.6e} | "
            f"{r['Nd']:12.6e} | "
            f"{r['matrix_gas_percent']:10.4e} | "
            f"{r['bulk_gas_percent']:10.4e} | "
            f"{r['dislocation_gas_percent']:10.4e} | "
            f"{r['qgb_gas_percent']:10.4e}"
        )


# ============================================================
# ESECUZIONE COMPLETA
# ============================================================

print("\n" + "#" * 150)
print("START FULL INITIALIZATION / PRESSURE DIAGNOSTIC")
print("#" * 150)

# ------------------------------------------------------------
# 1. Seed scan sulla pressione bulk a 3.2% FIMA
# ------------------------------------------------------------

seed_scan = {}

for case in SEED_CASES:
    print(f"\nRunning seed scan case: {case.name} -- {case.description}")

    rows = run_grid(
        case,
        BURNUP_PRESSURE,
        TEMPERATURES_SCAN,
    )

    seed_scan[case.name] = rows

print_seed_scan_table(seed_scan)
plot_seed_scan_pressure_ratios(seed_scan)

selected_case = choose_seed_case(seed_scan)

print("\n" + "=" * 150)
print("SELECTED CASE FOR DETAILED RIZK-STYLE DIAGNOSTICS")
print("=" * 150)
print(f"selected case = {selected_case.name}")
print(f"description   = {selected_case.description}")
print(f"R_b_seed      = {selected_case.R_b_seed:.6e} m = {selected_case.R_b_seed*1e9:.3f} nm")
print(f"R_d0          = {selected_case.R_d0:.6e} m = {selected_case.R_d0*1e9:.3f} nm")
print("=" * 150)

# ------------------------------------------------------------
# 2. Pressione tipo Fig. 5 a 3.2% FIMA
# ------------------------------------------------------------

pressure_rows = run_grid(
    selected_case,
    BURNUP_PRESSURE,
    TEMPERATURES_FINE,
)

print_rows_compact(
    f"DETAILED PRESSURE / MICROSTRUCTURE TABLE - {selected_case.name} - 3.2% FIMA",
    pressure_rows,
)

plot_pressure_fig5_style(
    pressure_rows,
    selected_case.name,
)

# ------------------------------------------------------------
# 3. Swelling vs T a 1.1, 1.3, 3.2% FIMA
# ------------------------------------------------------------

results_by_burnup = {}

for burnup in BURNUPS_SWELLING:
    rows = run_grid(
        selected_case,
        burnup,
        TEMPERATURES_FINE,
    )

    results_by_burnup[burnup] = rows

    print_rows_compact(
        f"SWELLING / MICROSTRUCTURE TABLE - {selected_case.name} - {burnup:.1f}% FIMA",
        rows,
    )

plot_swelling_fig3_style(
    results_by_burnup,
    selected_case.name,
)

# ------------------------------------------------------------
# 4. Bubble concentration and radius at 1.3% FIMA
# ------------------------------------------------------------

rows_13 = results_by_burnup[1.3]

plot_number_density_fig7_style(
    rows_13,
    selected_case.name,
)

plot_radius_fig8_style(
    rows_13,
    selected_case.name,
)

# ------------------------------------------------------------
# 5. Gas partition at 1.1% and 3.2% FIMA
# ------------------------------------------------------------

rows_11 = results_by_burnup[1.1]
rows_32 = results_by_burnup[3.2]

plot_gas_partition_fig9_style(
    rows_11,
    1.1,
    selected_case.name,
)

plot_gas_partition_fig9_style(
    rows_32,
    3.2,
    selected_case.name,
)

print("\n" + "#" * 150)
print("END FULL INITIALIZATION / PRESSURE DIAGNOSTIC")
print("#" * 150)

if plt is None:
    print("\nmatplotlib is not installed: plots skipped, numeric tables printed.")
````

````python
# FULL DIAGNOSTIC CELL v3
#
# Obiettivo:
# capire perché il modello Python non riproduce i grafici Rizk.
#
# Questa cella NON richiede modifiche sparse nel codice principale.
# Definisce un solver diagnostico separato con:
# - solver gas 3eq tipo Sciantix corrente
# - solver gas Polypole1-like tipo Rizk
# - sensitivity su f_n, K_d, g_b, g_d, seed bulk, denominatore dislocation
#
# Grafici prodotti:
# 1) dislocation swelling vs T, 3.2% FIMA
# 2) bulk swelling vs T, 3.2% FIMA
# 3) gas partition vs T, 3.2% FIMA
# 4) N_b, N_d vs T, 1.3% FIMA
# 5) R_b, R_d vs T, 1.3% FIMA
# 6) pressure ratios p/p_eq, 3.2% FIMA
# 7) grafici finali per alcuni casi selezionati a 1.1, 1.3, 3.2% FIMA
# ============================================================

import math
from dataclasses import dataclass, replace

try:
    import matplotlib.pyplot as plt
except ModuleNotFoundError:
    plt = None


# ============================================================
# CHECK: devi aver già eseguito la cella principale
# ============================================================

_REQUIRED = [
    "UNParameters",
    "omega_matrix",
    "sphere_volume",
    "radius_from_volume",
    "xe_diffusivity_UN",
    "resolution_rates_UN",
    "beta_production",
    "nucleation_rate_bulk",
    "phi_bulk",
    "coalescence_lambda",
    "pressure_internal",
    "pressure_equilibrium",
    "wigner_seitz_delta",
    "zeta_geometry",
    "initialize_vacancy_concentration",
    "initialize_modes_from_average",
    "reconstruct_average",
    "sciantix_3x3_exchange_step",
    "time_to_burnup_percent",
]

_missing = [name for name in _REQUIRED if name not in globals()]

if _missing:
    raise RuntimeError(
        "Prima devi eseguire la cella principale del modello. "
        f"Mancano: {_missing}"
    )


# ============================================================
# IMPOSTAZIONI GLOBALI
# ============================================================

FISSION_RATE = 5.0e19
DT = 3600.0

# Rizk Table 1 usa circa r_gr = 6 micrometri.
# Teniamo anche un caso 5 micrometri per confrontare col codice precedente.
GRAIN_RADIUS_RIZK = 6.0e-6
GRAIN_RADIUS_OLD = 5.0e-6

TEMPERATURES_FULL = [float(T) for T in range(800, 2601, 100)]
TEMPERATURES_FINE = [float(T) for T in range(900, 2001, 50)]

BURNUPS = [1.1, 1.3, 3.2]


# ============================================================
# VARIANTI DA TESTARE
# ============================================================

@dataclass
class Variant:
    name: str
    gas_solver: str = "3eq"       # "3eq" oppure "polypole1"
    f_n: float = 1.0e-6
    K_d: float = 5.0e5
    grain_radius: float = GRAIN_RADIUS_RIZK
    R_b_seed: float = 0.0
    R_d0: float = 0.0
    gb_scale: float = 1.0
    gd_scale: float = 1.0
    den_subtract: float = 3.0 / 5.0
    description: str = ""


VARIANTS = [
    Variant(
        name="old_3eq_rgr5",
        gas_solver="3eq",
        f_n=1.0e-6,
        K_d=5.0e5,
        grain_radius=GRAIN_RADIUS_OLD,
        description="codice precedente circa: 3eq, r_gr=5um, f_n=1e-6"
    ),
    Variant(
        name="rizk_table_3eq",
        gas_solver="3eq",
        f_n=1.0e-6,
        K_d=5.0e5,
        grain_radius=GRAIN_RADIUS_RIZK,
        description="parametri Rizk tabella, ma solver gas 3eq"
    ),
    Variant(
        name="rizk_table_polypole1",
        gas_solver="polypole1",
        f_n=1.0e-6,
        K_d=5.0e5,
        grain_radius=GRAIN_RADIUS_RIZK,
        description="parametri Rizk tabella + gas solver Polypole1-like"
    ),
    Variant(
        name="bulk_seed_2nm_poly",
        gas_solver="polypole1",
        f_n=1.0e-6,
        K_d=5.0e5,
        grain_radius=GRAIN_RADIUS_RIZK,
        R_b_seed=2.0e-9,
        description="Polypole1 + seed bulk 2 nm"
    ),
    Variant(
        name="low_fn_1e-7_poly",
        gas_solver="polypole1",
        f_n=1.0e-7,
        K_d=5.0e5,
        grain_radius=GRAIN_RADIUS_RIZK,
        description="Polypole1 + nucleazione bulk ridotta 10x"
    ),
    Variant(
        name="low_fn_1e-8_poly",
        gas_solver="polypole1",
        f_n=1.0e-8,
        K_d=5.0e5,
        grain_radius=GRAIN_RADIUS_RIZK,
        description="Polypole1 + nucleazione bulk ridotta 100x"
    ),
    Variant(
        name="gd_x5_poly",
        gas_solver="polypole1",
        f_n=1.0e-6,
        K_d=5.0e5,
        grain_radius=GRAIN_RADIUS_RIZK,
        gd_scale=5.0,
        description="Polypole1 + trapping dislocation moltiplicato x5"
    ),
    Variant(
        name="gb_x0p2_poly",
        gas_solver="polypole1",
        f_n=1.0e-6,
        K_d=5.0e5,
        grain_radius=GRAIN_RADIUS_RIZK,
        gb_scale=0.2,
        description="Polypole1 + trapping bulk ridotto a 0.2"
    ),
    Variant(
        name="Kd_1e6_poly",
        gas_solver="polypole1",
        f_n=1.0e-6,
        K_d=1.0e6,
        grain_radius=GRAIN_RADIUS_RIZK,
        description="Polypole1 + K_d = 1e6 invece di 5e5"
    ),
    Variant(
        name="den_5over3_poly",
        gas_solver="polypole1",
        f_n=1.0e-6,
        K_d=5.0e5,
        grain_radius=GRAIN_RADIUS_RIZK,
        den_subtract=5.0 / 3.0,
        description="Polypole1 + denominatore dislocation con -5/3"
    ),
]


# Casi per cui facciamo anche i grafici finali completi
FULL_PLOT_VARIANTS = [
    "rizk_table_3eq",
    "rizk_table_polypole1",
    "low_fn_1e-7_poly",
    "low_fn_1e-8_poly",
    "gd_x5_poly",
    "gb_x0p2_poly",
]


# ============================================================
# DIFFUSIVITÀ VACANZE CORRETTA
# ============================================================

def vacancy_diffusivity_UN_corrected(p):
    """
    Dv = D1 + D2_corrected.
    D2 usa il prefattore rifittato alla curva Rizk Fig. 4 'Vacancies, bulk'.
    """
    T = p.temperature
    F = p.fission_rate
    kBT = p.kB_eV * T

    D1 = p.D10_vU * math.exp(-p.Q1_vU / kBT)

    A20_vU_fig4_refit = 4.6304523933553033e-29

    D2 = math.sqrt(F) * A20_vU_fig4_refit * math.exp(
        p.B21_vU / kBT
        + p.B22_vU / (kBT**2)
    )

    Dv = D1 + D2

    return Dv, {
        "Dv1": D1,
        "Dv2": D2,
        "Dv": Dv,
    }


# ============================================================
# TRAPPING VARIANT
# ============================================================

def trapping_rates_variant(p, Dg, R_b, N_b, R_d, N_d, variant):
    Rb_eff = R_b + p.radius_in_lattice
    Rd_eff = R_d + p.radius_in_lattice

    if N_b <= 0.0:
        g_b = 0.0
    else:
        g_b = 4.0 * math.pi * Dg * Rb_eff * N_b

    Gamma_d = 1.0 / math.sqrt(math.pi * p.rho_d)
    den = math.log(Gamma_d / (p.Z_d * p.r_d)) - variant.den_subtract

    if den <= 0.0:
        raise ValueError(f"Invalid dislocation denominator: {den:g}")

    free_dislocation = max(p.rho_d - 2.0 * R_d * N_d, 0.0)

    term_bubbles = 4.0 * math.pi * Dg * Rd_eff * N_d
    term_dislocation = (2.0 * math.pi * Dg / den) * free_dislocation

    g_d = term_bubbles + term_dislocation

    g_b *= variant.gb_scale
    g_d *= variant.gd_scale

    return g_b, g_d, {
        "Gamma_d": Gamma_d,
        "den": den,
        "free_dislocation": free_dislocation,
        "term_bubbles": term_bubbles,
        "term_dislocation": term_dislocation,
    }


# ============================================================
# RADIUS REGULARIZATION FOR VACANCY UPDATE
# ============================================================

def gas_only_radius_for_population(p, m_gas, N):
    if m_gas <= 0.0 or N <= 0.0:
        return 0.0

    V_gas_per_bubble = p.omega_fg * m_gas / N

    if V_gas_per_bubble <= 0.0:
        return 0.0

    return radius_from_volume(V_gas_per_bubble)


def radius_for_vacancy_update(p, R_old, N, m_gas):
    if R_old > 0.0:
        return R_old

    R_gas = gas_only_radius_for_population(p, m_gas, N)

    if R_gas > 0.0:
        return R_gas

    return 0.0


def pressure_equilibrium_for_vacancy_update(p, R_update):
    if R_update <= 0.0:
        return math.inf

    return 2.0 * p.gamma_b / R_update - p.hydrostatic_stress


def vacancy_concentration_implicit_step_local(p, Dv, R, N, m_gas, n_old, dt):
    if N <= 0.0 or m_gas <= 0.0:
        return n_old, 0.0

    R_update = radius_for_vacancy_update(p, R, N, m_gas)

    if R_update <= 0.0:
        return n_old, 0.0

    p_eq = pressure_equilibrium_for_vacancy_update(p, R_update)
    p_int_old = pressure_internal(p, m_gas, n_old)

    if p.vacancy_absorption_only and p_int_old <= p_eq:
        return n_old, 0.0

    delta = wigner_seitz_delta(N)
    zeta = zeta_geometry(R_update, N)

    A = (
        2.0
        * math.pi
        * Dv
        * delta
        * N
        / (p.kB_J * p.temperature * zeta)
    )

    C = p.kB_J * p.temperature * m_gas / omega_matrix(p)

    # Backward Euler:
    # n_new = n_old + dt*A*(C/n_new - p_eq)
    B = n_old - dt * A * p_eq
    discriminant = B * B + 4.0 * dt * A * C

    if discriminant < 0.0:
        raise ValueError(
            f"Negative discriminant in vacancy implicit step: {discriminant:g}"
        )

    sqrt_disc = math.sqrt(discriminant)

    if B >= 0.0:
        n_new = 0.5 * (B + sqrt_disc)
    else:
        denominator = sqrt_disc - B
        if denominator <= 0.0:
            n_new = 0.0
        else:
            n_new = (2.0 * dt * A * C) / denominator

    if p.vacancy_absorption_only:
        n_new = max(n_new, n_old)

    dn_dt = (n_new - n_old) / dt

    return n_new, dn_dt


# ============================================================
# POLYPOLE1-LIKE SINGLE TOTAL GAS DIFFUSION STEP
# ============================================================

def single_diffusion_step(modes, D_eff, R_grain, beta, dt):
    """
    Risolve:
        d c_t / dt = D_eff Laplacian(c_t) + beta

    con lo stesso stile spettrale usato dal solver 3eq.
    """
    projection_coeff = -2.0 * math.sqrt(2.0 / math.pi)
    diffusion_rate_coeff = math.pi**2 * D_eff / R_grain**2

    for i in range(len(modes)):
        n = i + 1
        n_coeff = (-1.0) ** n / n

        diffusion_rate = diffusion_rate_coeff * n**2
        source_rate = projection_coeff * beta * n_coeff

        modes[i] = (modes[i] + source_rate * dt) / (1.0 + diffusion_rate * dt)

    return reconstruct_average(modes)


def polypole1_partition(Dg, g_b, g_d, b_b, b_d, c_total):
    """
    Quasi-stationary approximation:

        m_b = (g_b/b_b) c
        m_d = (g_d/b_d) c

        c_total = c + m_b + m_d

    quindi:

        c = c_total / (1 + g_b/b_b + g_d/b_d)
    """
    eps = 1.0e-300

    a_b = g_b / max(b_b, eps)
    a_d = g_d / max(b_d, eps)

    fraction_solution = 1.0 / (1.0 + a_b + a_d)

    c = fraction_solution * c_total
    m_b = a_b * c
    m_d = a_d * c

    D_eff = Dg * fraction_solution

    return c, m_b, m_d, D_eff, a_b, a_d, fraction_solution


# ============================================================
# SOLVER DIAGNOSTICO GENERALE
# ============================================================

def make_params_for_variant(variant, T, burnup):
    return UNParameters(
        temperature=float(T),
        fission_rate=FISSION_RATE,
        grain_radius=variant.grain_radius,
        target_burnup_percent_fima=float(burnup),
        dt=DT,
        n_modes=40,
        R_b=0.0,
        N_b=0.0,
        R_d=variant.R_d0,
        N_d=None,
        K_d=variant.K_d,
        f_n=variant.f_n,
        c0=0.0,
        mb0=0.0,
        md0=0.0,
        vacancy_absorption_only=True,
        update_bulk_vacancies=True,
    )


def solve_variant(p, variant):
    # Gas modes for 3eq solver
    modes_c = initialize_modes_from_average(p.c0, p.n_modes)
    modes_mb = initialize_modes_from_average(p.mb0, p.n_modes)
    modes_md = initialize_modes_from_average(p.md0, p.n_modes)

    # Total gas modes for Polypole1-like solver
    modes_ct = initialize_modes_from_average(p.c0 + p.mb0 + p.md0, p.n_modes)

    R_b = p.R_b
    R_d = p.R_d

    N_b = p.N_b
    N_d = p.N_d

    V_b = sphere_volume(R_b)
    V_d = sphere_volume(R_d)

    if p.nvb0 is None:
        nvb = initialize_vacancy_concentration(p, N_b, R_b, p.mb0)
    else:
        nvb = p.nvb0

    if p.nvd0 is None:
        nvd = initialize_vacancy_concentration(p, N_d, R_d, p.md0)
    else:
        nvd = p.nvd0

    c = p.c0
    mb = p.mb0
    md = p.md0

    beta = beta_production(p)

    initial_gas = c + mb + md
    generated = 0.0
    q_gb = 0.0

    hist = {
        "time": [0.0],
        "burnup_percent_fima": [0.0],
        "c": [c],
        "mb": [mb],
        "md": [md],
        "retained": [c + mb + md],
        "generated": [generated],
        "q_gb": [q_gb],
        "Nb": [N_b],
        "Nd": [N_d],
        "Vb": [V_b],
        "Vd": [V_d],
        "Rb": [R_b],
        "Rd": [R_d],
        "nvb": [nvb],
        "nvd": [nvd],
        "swelling_b": [N_b * V_b],
        "swelling_d": [N_d * V_d],
        "swelling_ig": [N_b * V_b + N_d * V_d],
        "p_b": [pressure_internal(p, mb, nvb)],
        "p_b_eq": [pressure_equilibrium(p, R_b)],
        "p_d": [pressure_internal(p, md, nvd)],
        "p_d_eq": [pressure_equilibrium(p, R_d)],
        "gb": [0.0],
        "gd": [0.0],
        "bb": [0.0],
        "bd": [0.0],
        "Dg": [0.0],
        "Dv": [0.0],
        "Deff": [0.0],
        "alpha_b": [0.0],
        "alpha_d": [0.0],
        "nu_b": [0.0],
        "phi_b": [0.0],
        "coverage_d": [0.0],
    }

    last_rates = {}

    n_steps = int(math.ceil(p.final_time / p.dt))

    for _ in range(n_steps):
        t_old = hist["time"][-1]
        dt = min(p.dt, p.final_time - t_old)

        if dt <= 0.0:
            break

        c_old = c
        mb_old = mb
        md_old = md

        Nb_old = N_b
        Nd_old = N_d
        Vd_old = V_d

        Dg, D_parts = xe_diffusivity_UN(p)
        Dv, Dv_parts = vacancy_diffusivity_UN_corrected(p)

        b_b, b_d = resolution_rates_UN(p, R_b, R_d)

        g_b, g_d, trapping_parts = trapping_rates_variant(
            p,
            Dg,
            R_b,
            Nb_old,
            R_d,
            Nd_old,
            variant,
        )

        # --------------------------------------------------------
        # Update N_b prima del gas solver
        # --------------------------------------------------------
        nu_b = nucleation_rate_bulk(p, Dg, c_old)
        phi_b = phi_bulk(mb_old, Nb_old)

        N_b = (Nb_old + dt * nu_b) / (1.0 + dt * b_b * phi_b)
        N_b = max(N_b, p.min_number_density)

        # Seed volume/vacanze per bolle bulk appena nucleate
        dN_b_positive = max(N_b - Nb_old, 0.0)

        if dN_b_positive > 0.0 and variant.R_b_seed > 0.0:
            nvb += dN_b_positive * sphere_volume(variant.R_b_seed) / omega_matrix(p)

        # --------------------------------------------------------
        # Gas solver
        # --------------------------------------------------------
        if variant.gas_solver == "3eq":
            c_new, mb_new, md_new = sciantix_3x3_exchange_step(
                modes_c,
                modes_mb,
                modes_md,
                Dg,
                p.grain_radius,
                beta,
                g_b,
                g_d,
                b_b,
                b_d,
                dt,
            )

            D_eff = math.nan
            alpha_b = g_b / max(b_b, 1.0e-300)
            alpha_d = g_d / max(b_d, 1.0e-300)

        elif variant.gas_solver == "polypole1":
            alpha_b_tmp = g_b / max(b_b, 1.0e-300)
            alpha_d_tmp = g_d / max(b_d, 1.0e-300)

            D_eff = Dg / (1.0 + alpha_b_tmp + alpha_d_tmp)

            c_total_new = single_diffusion_step(
                modes_ct,
                D_eff,
                p.grain_radius,
                beta,
                dt,
            )

            c_new, mb_new, md_new, D_eff, alpha_b, alpha_d, _ = polypole1_partition(
                Dg,
                g_b,
                g_d,
                b_b,
                b_d,
                c_total_new,
            )

        else:
            raise ValueError(f"Unknown gas_solver: {variant.gas_solver}")

        dmb_dt = (mb_new - mb_old) / dt
        dmd_dt = (md_new - md_old) / dt

        c = c_new
        mb = mb_new
        md = md_new

        # --------------------------------------------------------
        # Vacancy update
        # --------------------------------------------------------
        if p.update_bulk_vacancies:
            nvb, dnvb_dt = vacancy_concentration_implicit_step_local(
                p,
                Dv,
                R_b,
                N_b,
                mb,
                nvb,
                dt,
            )
        else:
            dnvb_dt = 0.0

        nvd, dnvd_dt = vacancy_concentration_implicit_step_local(
            p,
            Dv,
            R_d,
            Nd_old,
            md,
            nvd,
            dt,
        )

        # --------------------------------------------------------
        # Dislocation bubble growth/coalescence
        # --------------------------------------------------------
        if Nd_old > 0.0:
            dVd_growth_dt = (
                p.omega_fg / Nd_old * dmd_dt
                + omega_matrix(p) / Nd_old * dnvd_dt
            )
            V_d_growth = V_d + dt * dVd_growth_dt
            V_d_growth = max(V_d_growth, p.min_volume)
        else:
            dVd_growth_dt = 0.0
            V_d_growth = V_d

        lambda_d = coalescence_lambda(Vd_old, Nd_old)
        dVd_positive = max(V_d_growth - Vd_old, 0.0)

        if dVd_positive > 0.0 and Nd_old > 0.0:
            N_d = Nd_old / (1.0 + 4.0 * lambda_d * Nd_old * dVd_positive)
        else:
            N_d = Nd_old

        N_d = max(N_d, p.min_number_density)

        # --------------------------------------------------------
        # Reconstruct volumes/radii
        # --------------------------------------------------------
        if N_b > 0.0:
            V_b = (
                p.omega_fg * max(mb, 0.0)
                + omega_matrix(p) * nvb
            ) / N_b
        else:
            V_b = 0.0

        if N_d > 0.0:
            V_d = (
                p.omega_fg * max(md, 0.0)
                + omega_matrix(p) * nvd
            ) / N_d
        else:
            V_d = 0.0

        V_b = max(V_b, p.min_volume)
        V_d = max(V_d, p.min_volume)

        R_b = radius_from_volume(V_b)
        R_d = radius_from_volume(V_d)

        # --------------------------------------------------------
        # Gas to grain face by mass balance
        # --------------------------------------------------------
        generated += beta * dt
        retained = c + mb + md

        q_gb = max(initial_gas + generated - retained, 0.0)

        burnup_now = time_to_burnup_percent(
            t_old + dt,
            p.fission_rate,
            p.lattice_parameter,
        )

        coverage_d = 0.0
        if p.rho_d > 0.0:
            coverage_d = 2.0 * R_d * N_d / p.rho_d

        hist["time"].append(t_old + dt)
        hist["burnup_percent_fima"].append(burnup_now)
        hist["c"].append(c)
        hist["mb"].append(mb)
        hist["md"].append(md)
        hist["retained"].append(retained)
        hist["generated"].append(generated)
        hist["q_gb"].append(q_gb)
        hist["Nb"].append(N_b)
        hist["Nd"].append(N_d)
        hist["Vb"].append(V_b)
        hist["Vd"].append(V_d)
        hist["Rb"].append(R_b)
        hist["Rd"].append(R_d)
        hist["nvb"].append(nvb)
        hist["nvd"].append(nvd)
        hist["swelling_b"].append(N_b * V_b)
        hist["swelling_d"].append(N_d * V_d)
        hist["swelling_ig"].append(N_b * V_b + N_d * V_d)
        hist["p_b"].append(pressure_internal(p, mb, nvb))
        hist["p_b_eq"].append(pressure_equilibrium(p, R_b))
        hist["p_d"].append(pressure_internal(p, md, nvd))
        hist["p_d_eq"].append(pressure_equilibrium(p, R_d))
        hist["gb"].append(g_b)
        hist["gd"].append(g_d)
        hist["bb"].append(b_b)
        hist["bd"].append(b_d)
        hist["Dg"].append(Dg)
        hist["Dv"].append(Dv)
        hist["Deff"].append(D_eff)
        hist["alpha_b"].append(alpha_b)
        hist["alpha_d"].append(alpha_d)
        hist["nu_b"].append(nu_b)
        hist["phi_b"].append(phi_b)
        hist["coverage_d"].append(coverage_d)

        last_rates = {
            "Dg": Dg,
            "Dv": Dv,
            "Deff": D_eff,
            "g_b": g_b,
            "g_d": g_d,
            "b_b": b_b,
            "b_d": b_d,
            "alpha_b": alpha_b,
            "alpha_d": alpha_d,
            "nu_b": nu_b,
            "phi_b": phi_b,
            "coverage_d": coverage_d,
            **D_parts,
            **Dv_parts,
            **trapping_parts,
        }

    return hist, last_rates


# ============================================================
# SUMMARY FUNCTIONS
# ============================================================

_CACHE = {}

def run_one(variant, T, burnup):
    key = (variant.name, float(T), float(burnup))

    if key in _CACHE:
        return _CACHE[key]

    p = make_params_for_variant(variant, T, burnup)

    hist, rates = solve_variant(p, variant)

    generated = hist["generated"][-1]

    if generated > 0.0:
        c_pct = 100.0 * hist["c"][-1] / generated
        mb_pct = 100.0 * hist["mb"][-1] / generated
        md_pct = 100.0 * hist["md"][-1] / generated
        qgb_pct = 100.0 * hist["q_gb"][-1] / generated
        retained_pct = 100.0 * hist["retained"][-1] / generated
    else:
        c_pct = mb_pct = md_pct = qgb_pct = retained_pct = 0.0

    p_b = hist["p_b"][-1]
    p_b_eq = hist["p_b_eq"][-1]
    p_d = hist["p_d"][-1]
    p_d_eq = hist["p_d_eq"][-1]

    row = {
        "variant": variant.name,
        "gas_solver": variant.gas_solver,
        "T": float(T),
        "burnup": float(burnup),
        "time_d": p.final_time / 86400.0,

        "sw_b_pct": 100.0 * hist["swelling_b"][-1],
        "sw_d_pct": 100.0 * hist["swelling_d"][-1],
        "sw_ig_pct": 100.0 * hist["swelling_ig"][-1],

        "Rb_nm": hist["Rb"][-1] * 1.0e9,
        "Rd_nm": hist["Rd"][-1] * 1.0e9,
        "Nb": hist["Nb"][-1],
        "Nd": hist["Nd"][-1],

        "p_b": p_b,
        "p_b_eq": p_b_eq,
        "p_d": p_d,
        "p_d_eq": p_d_eq,
        "p_b_ratio": p_b / p_b_eq if p_b_eq > 0.0 else math.nan,
        "p_d_ratio": p_d / p_d_eq if p_d_eq > 0.0 else math.nan,

        "matrix_pct": c_pct,
        "bulk_gas_pct": mb_pct,
        "disl_gas_pct": md_pct,
        "qgb_pct": qgb_pct,
        "retained_pct": retained_pct,

        "Dg": rates["Dg"],
        "Dv": rates["Dv"],
        "Deff": rates["Deff"],
        "gb": rates["g_b"],
        "gd": rates["g_d"],
        "bb": rates["b_b"],
        "bd": rates["b_d"],
        "alpha_b": rates["alpha_b"],
        "alpha_d": rates["alpha_d"],
        "coverage_d_pct": 100.0 * rates["coverage_d"],

        "hist": hist,
        "rates": rates,
    }

    _CACHE[key] = row

    return row


def run_grid(variant, burnup, temperatures):
    return [run_one(variant, T, burnup) for T in temperatures]


def print_variant_summary_at_key_temperatures(rows_by_variant, burnup):
    key_Ts = [900.0, 1500.0, 1700.0, 2000.0, 2400.0, 2600.0]

    print("\n" + "=" * 170)
    print(f"SUMMARY AT SELECTED TEMPERATURES - burnup = {burnup:.1f}% FIMA")
    print("=" * 170)
    print(
        f"{'variant':>24s} | {'T_K':>7s} | "
        f"{'sw_b_%':>10s} | {'sw_d_%':>10s} | {'qgb_%':>10s} | "
        f"{'bulk_gas_%':>12s} | {'disl_gas_%':>12s} | "
        f"{'Rb_nm':>10s} | {'Rd_nm':>10s} | "
        f"{'Nb':>12s} | {'Nd':>12s} | "
        f"{'pb/peq':>10s} | {'pd/peq':>10s} | {'cov_d_%':>10s}"
    )
    print("-" * 170)

    for variant_name, rows in rows_by_variant.items():
        for T in key_Ts:
            row = min(rows, key=lambda r: abs(r["T"] - T))

            print(
                f"{variant_name:>24s} | "
                f"{row['T']:7.1f} | "
                f"{row['sw_b_pct']:10.4e} | "
                f"{row['sw_d_pct']:10.4e} | "
                f"{row['qgb_pct']:10.4e} | "
                f"{row['bulk_gas_pct']:12.4e} | "
                f"{row['disl_gas_pct']:12.4e} | "
                f"{row['Rb_nm']:10.4e} | "
                f"{row['Rd_nm']:10.4e} | "
                f"{row['Nb']:12.4e} | "
                f"{row['Nd']:12.4e} | "
                f"{row['p_b_ratio']:10.4e} | "
                f"{row['p_d_ratio']:10.4e} | "
                f"{row['coverage_d_pct']:10.4e}"
            )


# ============================================================
# PLOTTING
# ============================================================

def plot_variant_scan(rows_by_variant, burnup):
    if plt is None:
        print("matplotlib assente: grafici saltati.")
        return

    # 1) Dislocation swelling
    plt.figure(figsize=(9, 5))
    for name, rows in rows_by_variant.items():
        Ts = [r["T"] for r in rows]
        ys = [r["sw_d_pct"] for r in rows]
        plt.plot(Ts, ys, marker="o", label=name)
    plt.xlabel("T [K]")
    plt.ylabel("Dislocation bubble swelling [%]")
    plt.title(f"Dislocation swelling sensitivity at {burnup:.1f}% FIMA")
    plt.grid(True, alpha=0.3)
    plt.legend(fontsize=8)
    plt.tight_layout()
    plt.show()

    # 2) Bulk swelling
    plt.figure(figsize=(9, 5))
    for name, rows in rows_by_variant.items():
        Ts = [r["T"] for r in rows]
        ys = [r["sw_b_pct"] for r in rows]
        plt.plot(Ts, ys, marker="o", label=name)
    plt.xlabel("T [K]")
    plt.ylabel("Bulk bubble swelling [%]")
    plt.title(f"Bulk swelling sensitivity at {burnup:.1f}% FIMA")
    plt.grid(True, alpha=0.3)
    plt.legend(fontsize=8)
    plt.tight_layout()
    plt.show()

    # 3) Dislocation gas percentage
    plt.figure(figsize=(9, 5))
    for name, rows in rows_by_variant.items():
        Ts = [r["T"] for r in rows]
        ys = [r["disl_gas_pct"] for r in rows]
        plt.plot(Ts, ys, marker="o", label=name)
    plt.xlabel("T [K]")
    plt.ylabel("Gas in dislocation bubbles [% generated]")
    plt.title(f"Gas partition to dislocation bubbles at {burnup:.1f}% FIMA")
    plt.grid(True, alpha=0.3)
    plt.legend(fontsize=8)
    plt.tight_layout()
    plt.show()

    # 4) qgb
    plt.figure(figsize=(9, 5))
    for name, rows in rows_by_variant.items():
        Ts = [r["T"] for r in rows]
        ys = [r["qgb_pct"] for r in rows]
        plt.plot(Ts, ys, marker="o", label=name)
    plt.xlabel("T [K]")
    plt.ylabel("Gas to grain face [% generated]")
    plt.title(f"Gas to grain face sensitivity at {burnup:.1f}% FIMA")
    plt.grid(True, alpha=0.3)
    plt.legend(fontsize=8)
    plt.tight_layout()
    plt.show()

    # 5) Pressure ratio bulk
    plt.figure(figsize=(9, 5))
    for name, rows in rows_by_variant.items():
        Ts = [r["T"] for r in rows]
        ys = [r["p_b_ratio"] for r in rows]
        plt.plot(Ts, ys, marker="o", label=name)
    plt.axhline(1.0, linestyle="--")
    plt.xlabel("T [K]")
    plt.ylabel("p_b / p_b_eq [-]")
    plt.title(f"Bulk pressure ratio at {burnup:.1f}% FIMA")
    plt.grid(True, alpha=0.3)
    plt.legend(fontsize=8)
    plt.tight_layout()
    plt.show()

    # 6) Coverage dislocation
    plt.figure(figsize=(9, 5))
    for name, rows in rows_by_variant.items():
        Ts = [r["T"] for r in rows]
        ys = [r["coverage_d_pct"] for r in rows]
        plt.plot(Ts, ys, marker="o", label=name)
    plt.xlabel("T [K]")
    plt.ylabel("Dislocation coverage [%]")
    plt.title(f"Dislocation coverage at {burnup:.1f}% FIMA")
    plt.grid(True, alpha=0.3)
    plt.legend(fontsize=8)
    plt.tight_layout()
    plt.show()


def plot_final_rizk_style(rows_by_burnup, variant_name):
    if plt is None:
        return

    for burnup, rows in rows_by_burnup.items():
        Ts = [r["T"] for r in rows]

        plt.figure(figsize=(8, 5))
        plt.plot(Ts, [r["sw_b_pct"] for r in rows], marker="o", linestyle=":", label="Bulk")
        plt.plot(Ts, [r["sw_d_pct"] for r in rows], marker="o", linestyle="--", label="Dislocation")
        plt.plot(Ts, [r["sw_ig_pct"] for r in rows], marker="o", linestyle="-", label="Bulk + Dislocation")
        plt.xlabel("T [K]")
        plt.ylabel("Fission gas swelling [%]")
        plt.title(f"Swelling at {burnup:.1f}% FIMA - {variant_name}")
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()

    # Number density at 1.3% FIMA
    if 1.3 in rows_by_burnup:
        rows = rows_by_burnup[1.3]
        Ts = [r["T"] for r in rows]

        plt.figure(figsize=(8, 5))
        plt.plot(Ts, [r["Nb"] for r in rows], marker="o", linestyle=":", label="Bulk")
        plt.plot(Ts, [r["Nd"] for r in rows], marker="o", linestyle="--", label="Dislocation")
        plt.yscale("log")
        plt.xlabel("T [K]")
        plt.ylabel("Bubble concentration [m$^{-3}$]")
        plt.title(f"Bubble concentrations at 1.3% FIMA - {variant_name}")
        plt.grid(True, which="both", alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()

        plt.figure(figsize=(8, 5))
        plt.plot(Ts, [r["Rb_nm"] for r in rows], marker="o", linestyle=":", label="Bulk")
        plt.plot(Ts, [r["Rd_nm"] for r in rows], marker="o", linestyle="--", label="Dislocation")
        plt.yscale("log")
        plt.xlabel("T [K]")
        plt.ylabel("Bubble radius [nm]")
        plt.title(f"Bubble radii at 1.3% FIMA - {variant_name}")
        plt.grid(True, which="both", alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()

    # Gas partition at 3.2% FIMA up to 2600 K
    if 3.2 in rows_by_burnup:
        rows = rows_by_burnup[3.2]
        Ts = [r["T"] for r in rows]

        plt.figure(figsize=(8, 5))
        plt.plot(Ts, [r["matrix_pct"] for r in rows], label="Matrix")
        plt.plot(Ts, [r["bulk_gas_pct"] for r in rows], linestyle=":", label="Bulk bubbles")
        plt.plot(Ts, [r["disl_gas_pct"] for r in rows], linestyle="--", label="Dislocation bubbles")
        plt.plot(Ts, [r["qgb_pct"] for r in rows], linestyle="-.", label="Gas to grain face")
        plt.xlabel("T [K]")
        plt.ylabel("Amount of generated gas [%]")
        plt.title(f"Gas partition at 3.2% FIMA - {variant_name}")
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()

        plt.figure(figsize=(8, 5))
        plt.plot(Ts, [r["p_b"] for r in rows], label="Bulk pressure")
        plt.plot(Ts, [r["p_b_eq"] for r in rows], linestyle=":", label="Bulk equilibrium pressure")
        plt.plot(Ts, [r["p_d"] for r in rows], label="Dislocation pressure")
        plt.plot(Ts, [r["p_d_eq"] for r in rows], linestyle="--", label="Dislocation equilibrium pressure")
        plt.yscale("log")
        plt.xlabel("T [K]")
        plt.ylabel("Pressure [Pa]")
        plt.title(f"Bubble pressures at 3.2% FIMA - {variant_name}")
        plt.grid(True, which="both", alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()


# ============================================================
# RUN
# ============================================================

print("\n" + "#" * 150)
print("START FULL MODEL DIAGNOSTIC v3")
print("#" * 150)

print("\nVARIANTS:")
for v in VARIANTS:
    print(f"- {v.name:24s}: {v.description}")

# ------------------------------------------------------------
# 1. Variant scan at 3.2% FIMA up to 2600 K
# ------------------------------------------------------------

burnup_scan = 3.2
rows_by_variant_32 = {}

for variant in VARIANTS:
    print(f"\nRunning variant scan: {variant.name}")
    rows_by_variant_32[variant.name] = run_grid(
        variant,
        burnup_scan,
        TEMPERATURES_FULL,
    )

print_variant_summary_at_key_temperatures(
    rows_by_variant_32,
    burnup_scan,
)

plot_variant_scan(
    rows_by_variant_32,
    burnup_scan,
)

# ------------------------------------------------------------
# 2. Full Rizk-style plots for selected variants
# ------------------------------------------------------------

for variant in VARIANTS:
    if variant.name not in FULL_PLOT_VARIANTS:
        continue

    print("\n" + "=" * 150)
    print(f"FULL RIZK-STYLE PLOTS FOR VARIANT: {variant.name}")
    print("=" * 150)
    print(variant.description)

    rows_by_burnup = {}

    for burnup in BURNUPS:
        rows_by_burnup[burnup] = run_grid(
            variant,
            burnup,
            TEMPERATURES_FULL,
        )

    for burnup, rows in rows_by_burnup.items():
        print_variant_summary_at_key_temperatures(
            {variant.name: rows},
            burnup,
        )

    plot_final_rizk_style(
        rows_by_burnup,
        variant.name,
    )

print("\n" + "#" * 150)
print("END FULL MODEL DIAGNOSTIC v3")
print("#" * 150)
````

````python
# FULL f_n DIAGNOSTIC - 3x3 SOLVER ONLY
#
# Obiettivo:
# capire se il disaccordo con Rizk viene da f_n / nucleazione bulk.
#
# Questa cella:
# - NON modifica il codice principale;
# - usa il solver 3x3;
# - usa la diffusività vacancy corretta;
# - testa diversi f_n e due condizioni di nucleazione bulk:
#       R_b_seed = 0 nm
#       R_b_seed = 2 nm
#
# Da incollare sotto la cella principale del modello.
# ============================================================

import math

try:
    import matplotlib.pyplot as plt
except ModuleNotFoundError:
    plt = None


# ============================================================
# CHECK NOMI NECESSARI
# ============================================================

_required_names = [
    "UNParameters",
    "omega_matrix",
    "sphere_volume",
    "radius_from_volume",
    "xe_diffusivity_UN",
    "resolution_rates_UN",
    "trapping_rates_UN",
    "beta_production",
    "nucleation_rate_bulk",
    "phi_bulk",
    "coalescence_lambda",
    "pressure_internal",
    "pressure_equilibrium",
    "wigner_seitz_delta",
    "zeta_geometry",
    "initialize_vacancy_concentration",
    "initialize_modes_from_average",
    "reconstruct_average",
    "sciantix_3x3_exchange_step",
    "time_to_burnup_percent",
]

_missing = [name for name in _required_names if name not in globals()]
if _missing:
    raise RuntimeError(
        "Prima devi eseguire la cella principale del modello. "
        f"Mancano questi nomi: {_missing}"
    )


# ============================================================
# CONFIGURAZIONE
# ============================================================

FISSION_RATE = 5.0e19
GRAIN_RADIUS = 6.0e-6       # Rizk table: circa 6 micrometri
DT = 3600.0                 # 1 h, già visto quasi convergente

BURNUPS = [1.1, 1.3, 3.2]

TEMPERATURES = [float(T) for T in range(900, 2601, 100)]

FN_VALUES = [
    1.0e-6,
    3.0e-7,
    1.0e-7,
    3.0e-8,
    1.0e-8,
    3.0e-9,
    1.0e-9,
]

RB_SEED_VALUES_NM = [
    0.0,
    2.0,
]

KEY_TEMPERATURES = [
    900.0,
    1500.0,
    1700.0,
    2000.0,
    2400.0,
    2600.0,
]


# ============================================================
# DIFFUSIVITÀ VACANZE CORRETTA
# ============================================================

def vacancy_diffusivity_UN_corrected(p):
    """
    Dv = D1 + D2_corrected.

    D1: mantenuto da Rizk.
    D2: usa forma con coefficienti firmati e prefattore rifittato
        alla curva 'Vacancies, bulk' di Rizk Fig. 4.
    """

    T = p.temperature
    F = p.fission_rate
    kBT = p.kB_eV * T

    D1 = p.D10_vU * math.exp(-p.Q1_vU / kBT)

    A20_vU_fig4_refit = 4.6304523933553033e-29

    D2 = math.sqrt(F) * A20_vU_fig4_refit * math.exp(
        p.B21_vU / kBT
        + p.B22_vU / (kBT**2)
    )

    Dv = D1 + D2

    return Dv, {
        "Dv1": D1,
        "Dv2": D2,
        "Dv": Dv,
    }


# ============================================================
# REGOLARIZZAZIONE SOLO PER UPDATE VACANCY
# ============================================================

def gas_only_radius_for_population(p, m_gas, N):
    if m_gas <= 0.0 or N <= 0.0:
        return 0.0

    V_gas_per_bubble = p.omega_fg * m_gas / N

    if V_gas_per_bubble <= 0.0:
        return 0.0

    return radius_from_volume(V_gas_per_bubble)


def radius_for_vacancy_update(p, R_old, N, m_gas):
    if R_old > 0.0:
        return R_old

    R_gas = gas_only_radius_for_population(p, m_gas, N)

    if R_gas > 0.0:
        return R_gas

    return 0.0


def pressure_equilibrium_for_vacancy_update(p, R_update):
    if R_update <= 0.0:
        return math.inf

    return 2.0 * p.gamma_b / R_update - p.hydrostatic_stress


def vacancy_concentration_implicit_step_local(p, Dv, R, N, m_gas, n_old, dt):
    if N <= 0.0 or m_gas <= 0.0:
        return n_old, 0.0

    R_update = radius_for_vacancy_update(p, R, N, m_gas)

    if R_update <= 0.0:
        return n_old, 0.0

    p_eq = pressure_equilibrium_for_vacancy_update(p, R_update)
    p_int_old = pressure_internal(p, m_gas, n_old)

    if p.vacancy_absorption_only and p_int_old <= p_eq:
        return n_old, 0.0

    delta = wigner_seitz_delta(N)
    zeta = zeta_geometry(R_update, N)

    A = (
        2.0
        * math.pi
        * Dv
        * delta
        * N
        / (p.kB_J * p.temperature * zeta)
    )

    C = p.kB_J * p.temperature * m_gas / omega_matrix(p)

    # Backward Euler:
    # n_new = n_old + dt*A*(C/n_new - p_eq)
    # n_new^2 - (n_old - dt*A*p_eq)*n_new - dt*A*C = 0
    B = n_old - dt * A * p_eq
    discriminant = B * B + 4.0 * dt * A * C

    if discriminant < 0.0:
        raise ValueError(f"Negative discriminant in vacancy step: {discriminant:g}")

    sqrt_disc = math.sqrt(discriminant)

    if B >= 0.0:
        n_new = 0.5 * (B + sqrt_disc)
    else:
        denominator = sqrt_disc - B
        if denominator <= 0.0:
            n_new = 0.0
        else:
            n_new = (2.0 * dt * A * C) / denominator

    if p.vacancy_absorption_only:
        n_new = max(n_new, n_old)

    dn_dt = (n_new - n_old) / dt

    return n_new, dn_dt


# ============================================================
# SOLVER 3x3 DIAGNOSTICO CON f_n E R_b_seed VARIABILI
# ============================================================

def make_params_fn_case(T, burnup_percent_fima, f_n):
    return UNParameters(
        temperature=float(T),
        fission_rate=FISSION_RATE,
        grain_radius=GRAIN_RADIUS,
        target_burnup_percent_fima=float(burnup_percent_fima),
        dt=DT,
        n_modes=40,
        f_n=float(f_n),
        R_b=0.0,
        N_b=0.0,
        R_d=0.0,
        N_d=None,
        c0=0.0,
        mb0=0.0,
        md0=0.0,
        vacancy_absorption_only=True,
        update_bulk_vacancies=True,
    )


def solve_UN_3x3_fn_seed(p, R_b_seed_nm=0.0):
    R_b_seed = R_b_seed_nm * 1.0e-9

    modes_c = initialize_modes_from_average(p.c0, p.n_modes)
    modes_mb = initialize_modes_from_average(p.mb0, p.n_modes)
    modes_md = initialize_modes_from_average(p.md0, p.n_modes)

    R_b = p.R_b
    R_d = p.R_d

    N_b = p.N_b
    N_d = p.N_d

    V_b = sphere_volume(R_b)
    V_d = sphere_volume(R_d)

    if p.nvb0 is None:
        nvb = initialize_vacancy_concentration(p, N_b, R_b, p.mb0)
    else:
        nvb = p.nvb0

    if p.nvd0 is None:
        nvd = initialize_vacancy_concentration(p, N_d, R_d, p.md0)
    else:
        nvd = p.nvd0

    beta = beta_production(p)

    c = p.c0
    mb = p.mb0
    md = p.md0

    initial_gas = c + mb + md
    generated = 0.0
    q_gb = 0.0

    hist = {
        "time": [0.0],
        "burnup_percent_fima": [0.0],
        "c": [c],
        "mb": [mb],
        "md": [md],
        "retained": [c + mb + md],
        "generated": [generated],
        "q_gb": [q_gb],
        "Nb": [N_b],
        "Nd": [N_d],
        "Rb": [R_b],
        "Rd": [R_d],
        "Vb": [V_b],
        "Vd": [V_d],
        "nvb": [nvb],
        "nvd": [nvd],
        "swelling_b": [N_b * V_b],
        "swelling_d": [N_d * V_d],
        "swelling_ig": [N_b * V_b + N_d * V_d],
        "p_b": [pressure_internal(p, mb, nvb)],
        "p_b_eq": [pressure_equilibrium(p, R_b)],
        "p_d": [pressure_internal(p, md, nvd)],
        "p_d_eq": [pressure_equilibrium(p, R_d)],
        "g_b": [0.0],
        "g_d": [0.0],
        "b_b": [0.0],
        "b_d": [0.0],
        "Dg": [0.0],
        "Dv": [0.0],
        "nu_b": [0.0],
        "phi_b": [0.0],
        "coverage_d": [0.0],
    }

    last_rates = {}

    n_steps = int(math.ceil(p.final_time / p.dt))

    for _ in range(n_steps):
        t_old = hist["time"][-1]
        dt = min(p.dt, p.final_time - t_old)

        if dt <= 0.0:
            break

        c_old = c
        mb_old = mb
        md_old = md

        Nb_old = N_b
        Nd_old = N_d
        Vd_old = V_d

        Dg, D_parts = xe_diffusivity_UN(p)
        Dv, Dv_parts = vacancy_diffusivity_UN_corrected(p)

        b_b, b_d = resolution_rates_UN(p, R_b, R_d)

        g_b, g_d, trapping_parts = trapping_rates_UN(
            p,
            Dg,
            R_b,
            Nb_old,
            R_d,
            Nd_old,
        )

        # --------------------------------------------------------
        # SCIANTIX-like order:
        # update N_b before gas solver, with old c and old mb.
        # --------------------------------------------------------
        nu_b = nucleation_rate_bulk(p, Dg, c_old)
        phi_b = phi_bulk(mb_old, Nb_old)

        N_b = (Nb_old + dt * nu_b) / (1.0 + dt * b_b * phi_b)
        N_b = max(N_b, p.min_number_density)

        # --------------------------------------------------------
        # Optional seed volume/vacancies for newly nucleated bulk bubbles.
        # --------------------------------------------------------
        dN_b_positive = max(N_b - Nb_old, 0.0)

        if dN_b_positive > 0.0 and R_b_seed > 0.0:
            nvb += dN_b_positive * sphere_volume(R_b_seed) / omega_matrix(p)

        # --------------------------------------------------------
        # Gas 3x3 solve with old coefficients.
        # --------------------------------------------------------
        c_new, mb_new, md_new = sciantix_3x3_exchange_step(
            modes_c,
            modes_mb,
            modes_md,
            Dg,
            p.grain_radius,
            beta,
            g_b,
            g_d,
            b_b,
            b_d,
            dt,
        )

        dmb_dt = (mb_new - mb_old) / dt
        dmd_dt = (md_new - md_old) / dt

        c = c_new
        mb = mb_new
        md = md_new

        # --------------------------------------------------------
        # Vacancy absorption for bulk and dislocation bubbles.
        # --------------------------------------------------------
        if p.update_bulk_vacancies:
            nvb, dnvb_dt = vacancy_concentration_implicit_step_local(
                p,
                Dv,
                R_b,
                N_b,
                mb,
                nvb,
                dt,
            )
        else:
            dnvb_dt = 0.0

        nvd, dnvd_dt = vacancy_concentration_implicit_step_local(
            p,
            Dv,
            R_d,
            Nd_old,
            md,
            nvd,
            dt,
        )

        # --------------------------------------------------------
        # Volume growth and dislocation coalescence.
        # --------------------------------------------------------
        if Nd_old > 0.0:
            dVd_growth_dt = (
                p.omega_fg / Nd_old * dmd_dt
                + omega_matrix(p) / Nd_old * dnvd_dt
            )
            V_d_growth = V_d + dt * dVd_growth_dt
            V_d_growth = max(V_d_growth, p.min_volume)
        else:
            dVd_growth_dt = 0.0
            V_d_growth = V_d

        lambda_d = coalescence_lambda(Vd_old, Nd_old)
        dVd_positive = max(V_d_growth - Vd_old, 0.0)

        if dVd_positive > 0.0 and Nd_old > 0.0:
            N_d = Nd_old / (1.0 + 4.0 * lambda_d * Nd_old * dVd_positive)
        else:
            N_d = Nd_old

        N_d = max(N_d, p.min_number_density)

        # --------------------------------------------------------
        # Reconstruct bubble volumes/radii.
        # --------------------------------------------------------
        if N_b > 0.0:
            V_b = (
                p.omega_fg * max(mb, 0.0)
                + omega_matrix(p) * nvb
            ) / N_b
        else:
            V_b = 0.0

        if N_d > 0.0:
            V_d = (
                p.omega_fg * max(md, 0.0)
                + omega_matrix(p) * nvd
            ) / N_d
        else:
            V_d = 0.0

        V_b = max(V_b, p.min_volume)
        V_d = max(V_d, p.min_volume)

        R_b = radius_from_volume(V_b)
        R_d = radius_from_volume(V_d)

        # --------------------------------------------------------
        # Mass balance to grain face.
        # --------------------------------------------------------
        generated += beta * dt
        retained = c + mb + md
        q_gb = max(initial_gas + generated - retained, 0.0)

        coverage_d = 0.0
        if p.rho_d > 0.0:
            coverage_d = 2.0 * R_d * N_d / p.rho_d

        burnup_now = time_to_burnup_percent(
            t_old + dt,
            p.fission_rate,
            p.lattice_parameter,
        )

        hist["time"].append(t_old + dt)
        hist["burnup_percent_fima"].append(burnup_now)
        hist["c"].append(c)
        hist["mb"].append(mb)
        hist["md"].append(md)
        hist["retained"].append(retained)
        hist["generated"].append(generated)
        hist["q_gb"].append(q_gb)
        hist["Nb"].append(N_b)
        hist["Nd"].append(N_d)
        hist["Rb"].append(R_b)
        hist["Rd"].append(R_d)
        hist["Vb"].append(V_b)
        hist["Vd"].append(V_d)
        hist["nvb"].append(nvb)
        hist["nvd"].append(nvd)
        hist["swelling_b"].append(N_b * V_b)
        hist["swelling_d"].append(N_d * V_d)
        hist["swelling_ig"].append(N_b * V_b + N_d * V_d)
        hist["p_b"].append(pressure_internal(p, mb, nvb))
        hist["p_b_eq"].append(pressure_equilibrium(p, R_b))
        hist["p_d"].append(pressure_internal(p, md, nvd))
        hist["p_d_eq"].append(pressure_equilibrium(p, R_d))
        hist["g_b"].append(g_b)
        hist["g_d"].append(g_d)
        hist["b_b"].append(b_b)
        hist["b_d"].append(b_d)
        hist["Dg"].append(Dg)
        hist["Dv"].append(Dv)
        hist["nu_b"].append(nu_b)
        hist["phi_b"].append(phi_b)
        hist["coverage_d"].append(coverage_d)

        last_rates = {
            "Dg": Dg,
            "Dv": Dv,
            "g_b": g_b,
            "g_d": g_d,
            "b_b": b_b,
            "b_d": b_d,
            "nu_b": nu_b,
            "phi_b": phi_b,
            "coverage_d": coverage_d,
            **D_parts,
            **Dv_parts,
            **trapping_parts,
        }

    return hist, last_rates


# ============================================================
# CACHE + SUMMARY
# ============================================================

_RUN_CACHE = {}


def run_one_case(T, burnup, f_n, R_b_seed_nm):
    key = (
        float(T),
        float(burnup),
        float(f_n),
        float(R_b_seed_nm),
    )

    if key in _RUN_CACHE:
        return _RUN_CACHE[key]

    p = make_params_fn_case(
        T=T,
        burnup_percent_fima=burnup,
        f_n=f_n,
    )

    hist, rates = solve_UN_3x3_fn_seed(
        p,
        R_b_seed_nm=R_b_seed_nm,
    )

    generated = hist["generated"][-1]

    if generated > 0.0:
        matrix_pct = 100.0 * hist["c"][-1] / generated
        bulk_gas_pct = 100.0 * hist["mb"][-1] / generated
        disl_gas_pct = 100.0 * hist["md"][-1] / generated
        qgb_pct = 100.0 * hist["q_gb"][-1] / generated
        retained_pct = 100.0 * hist["retained"][-1] / generated
    else:
        matrix_pct = 0.0
        bulk_gas_pct = 0.0
        disl_gas_pct = 0.0
        qgb_pct = 0.0
        retained_pct = 0.0

    p_b = hist["p_b"][-1]
    p_b_eq = hist["p_b_eq"][-1]
    p_d = hist["p_d"][-1]
    p_d_eq = hist["p_d_eq"][-1]

    row = {
        "T": float(T),
        "burnup": float(burnup),
        "f_n": float(f_n),
        "R_b_seed_nm": float(R_b_seed_nm),
        "time_d": p.final_time / 86400.0,

        "sw_b_pct": 100.0 * hist["swelling_b"][-1],
        "sw_d_pct": 100.0 * hist["swelling_d"][-1],
        "sw_ig_pct": 100.0 * hist["swelling_ig"][-1],

        "Rb_nm": hist["Rb"][-1] * 1.0e9,
        "Rd_nm": hist["Rd"][-1] * 1.0e9,
        "Nb": hist["Nb"][-1],
        "Nd": hist["Nd"][-1],

        "p_b": p_b,
        "p_b_eq": p_b_eq,
        "p_d": p_d,
        "p_d_eq": p_d_eq,
        "p_b_ratio": p_b / p_b_eq if p_b_eq > 0.0 else math.nan,
        "p_d_ratio": p_d / p_d_eq if p_d_eq > 0.0 else math.nan,

        "matrix_pct": matrix_pct,
        "bulk_gas_pct": bulk_gas_pct,
        "disl_gas_pct": disl_gas_pct,
        "qgb_pct": qgb_pct,
        "retained_pct": retained_pct,

        "Dg": rates["Dg"],
        "Dv": rates["Dv"],
        "g_b": rates["g_b"],
        "g_d": rates["g_d"],
        "b_b": rates["b_b"],
        "b_d": rates["b_d"],
        "nu_b": rates["nu_b"],
        "phi_b": rates["phi_b"],
        "coverage_d_pct": 100.0 * rates["coverage_d"],

        "hist": hist,
        "rates": rates,
    }

    _RUN_CACHE[key] = row

    return row


def run_grid_for(burnup, f_n, R_b_seed_nm):
    return [
        run_one_case(T, burnup, f_n, R_b_seed_nm)
        for T in TEMPERATURES
    ]


def nearest_row(rows, T_target):
    return min(rows, key=lambda r: abs(r["T"] - T_target))


def crossing_temperature(rows, key, threshold):
    rows_sorted = sorted(rows, key=lambda r: r["T"])

    for i in range(1, len(rows_sorted)):
        T0 = rows_sorted[i - 1]["T"]
        T1 = rows_sorted[i]["T"]
        y0 = rows_sorted[i - 1][key]
        y1 = rows_sorted[i][key]

        if y0 <= threshold <= y1:
            if y1 == y0:
                return T1
            return T0 + (threshold - y0) * (T1 - T0) / (y1 - y0)

    return None


# ============================================================
# RUN COMPLETO
# ============================================================

print("\n" + "#" * 150)
print("START f_n DIAGNOSTIC - 3x3 SOLVER ONLY")
print("#" * 150)

all_results = {}

for R_b_seed_nm in RB_SEED_VALUES_NM:
    for burnup in BURNUPS:
        for f_n in FN_VALUES:
            label = (R_b_seed_nm, burnup, f_n)
            print(
                f"Running: R_b_seed = {R_b_seed_nm:.1f} nm, "
                f"BU = {burnup:.1f}% FIMA, f_n = {f_n:.1e}"
            )
            all_results[label] = run_grid_for(
                burnup=burnup,
                f_n=f_n,
                R_b_seed_nm=R_b_seed_nm,
            )


# ============================================================
# TABELLA 1: valori a temperature chiave
# ============================================================

for R_b_seed_nm in RB_SEED_VALUES_NM:
    for burnup in BURNUPS:
        print("\n" + "=" * 170)
        print(
            f"KEY TEMPERATURE TABLE | R_b_seed = {R_b_seed_nm:.1f} nm | "
            f"burnup = {burnup:.1f}% FIMA"
        )
        print("=" * 170)

        print(
            f"{'f_n':>12s} | {'T_K':>7s} | "
            f"{'sw_b_%':>10s} | {'sw_d_%':>10s} | {'sw_ig_%':>10s} | "
            f"{'bulk_gas_%':>12s} | {'disl_gas_%':>12s} | {'qgb_%':>10s} | "
            f"{'Rb_nm':>10s} | {'Rd_nm':>10s} | "
            f"{'Nb':>12s} | {'Nd':>12s} | "
            f"{'pb/peq':>10s} | {'pd/peq':>10s} | "
            f"{'g_b/g_d':>10s}"
        )
        print("-" * 170)

        for f_n in FN_VALUES:
            rows = all_results[(R_b_seed_nm, burnup, f_n)]

            for T_key in KEY_TEMPERATURES:
                r = nearest_row(rows, T_key)
                g_ratio = r["g_b"] / r["g_d"] if r["g_d"] > 0.0 else math.inf

                print(
                    f"{f_n:12.1e} | "
                    f"{r['T']:7.1f} | "
                    f"{r['sw_b_pct']:10.4e} | "
                    f"{r['sw_d_pct']:10.4e} | "
                    f"{r['sw_ig_pct']:10.4e} | "
                    f"{r['bulk_gas_pct']:12.4e} | "
                    f"{r['disl_gas_pct']:12.4e} | "
                    f"{r['qgb_pct']:10.4e} | "
                    f"{r['Rb_nm']:10.4e} | "
                    f"{r['Rd_nm']:10.4e} | "
                    f"{r['Nb']:12.4e} | "
                    f"{r['Nd']:12.4e} | "
                    f"{r['p_b_ratio']:10.4e} | "
                    f"{r['p_d_ratio']:10.4e} | "
                    f"{g_ratio:10.4e}"
                )


# ============================================================
# TABELLA 2: temperature di salita dislocation swelling
# ============================================================

thresholds = [0.1, 0.5, 1.0, 2.0, 5.0]

for R_b_seed_nm in RB_SEED_VALUES_NM:
    for burnup in BURNUPS:
        print("\n" + "=" * 120)
        print(
            f"DISLOCATION SWELLING THRESHOLD TEMPERATURES | "
            f"R_b_seed = {R_b_seed_nm:.1f} nm | burnup = {burnup:.1f}% FIMA"
        )
        print("=" * 120)

        header = f"{'f_n':>12s}"
        for th in thresholds:
            header += f" | {'T@sw_d='+str(th)+'%':>14s}"
        print(header)
        print("-" * len(header))

        for f_n in FN_VALUES:
            rows = all_results[(R_b_seed_nm, burnup, f_n)]

            line = f"{f_n:12.1e}"
            for th in thresholds:
                T_cross = crossing_temperature(rows, "sw_d_pct", th)

                if T_cross is None:
                    line += f" | {'not reached':>14s}"
                else:
                    line += f" | {T_cross:14.1f}"

            print(line)


# ============================================================
# PLOTS
# ============================================================

if plt is not None:

    # --------------------------------------------------------
    # 1) Swelling bulk/dislocation per burnup e seed
    # --------------------------------------------------------
    for R_b_seed_nm in RB_SEED_VALUES_NM:
        for burnup in BURNUPS:

            plt.figure(figsize=(9, 5))
            for f_n in FN_VALUES:
                rows = all_results[(R_b_seed_nm, burnup, f_n)]
                Ts = [r["T"] for r in rows]
                ys = [r["sw_d_pct"] for r in rows]
                plt.plot(Ts, ys, marker="o", label=f"f_n={f_n:.0e}")

            plt.xlabel("T [K]")
            plt.ylabel("Dislocation bubble swelling [%]")
            plt.title(
                f"Dislocation swelling vs T | BU={burnup:.1f}% | "
                f"R_b_seed={R_b_seed_nm:.1f} nm"
            )
            plt.grid(True, alpha=0.3)
            plt.legend(fontsize=8)
            plt.tight_layout()
            plt.show()

            plt.figure(figsize=(9, 5))
            for f_n in FN_VALUES:
                rows = all_results[(R_b_seed_nm, burnup, f_n)]
                Ts = [r["T"] for r in rows]
                ys = [r["sw_b_pct"] for r in rows]
                plt.plot(Ts, ys, marker="o", label=f"f_n={f_n:.0e}")

            plt.xlabel("T [K]")
            plt.ylabel("Bulk bubble swelling [%]")
            plt.title(
                f"Bulk swelling vs T | BU={burnup:.1f}% | "
                f"R_b_seed={R_b_seed_nm:.1f} nm"
            )
            plt.grid(True, alpha=0.3)
            plt.legend(fontsize=8)
            plt.tight_layout()
            plt.show()

    # --------------------------------------------------------
    # 2) Gas partition at 3.2% FIMA fino a 2600 K
    # --------------------------------------------------------
    burnup = 3.2

    for R_b_seed_nm in RB_SEED_VALUES_NM:

        plt.figure(figsize=(9, 5))
        for f_n in FN_VALUES:
            rows = all_results[(R_b_seed_nm, burnup, f_n)]
            Ts = [r["T"] for r in rows]
            ys = [r["disl_gas_pct"] for r in rows]
            plt.plot(Ts, ys, marker="o", label=f"f_n={f_n:.0e}")

        plt.xlabel("T [K]")
        plt.ylabel("Gas in dislocation bubbles [% generated]")
        plt.title(
            f"Gas to dislocation bubbles | BU=3.2% | "
            f"R_b_seed={R_b_seed_nm:.1f} nm"
        )
        plt.grid(True, alpha=0.3)
        plt.legend(fontsize=8)
        plt.tight_layout()
        plt.show()

        plt.figure(figsize=(9, 5))
        for f_n in FN_VALUES:
            rows = all_results[(R_b_seed_nm, burnup, f_n)]
            Ts = [r["T"] for r in rows]
            ys = [r["bulk_gas_pct"] for r in rows]
            plt.plot(Ts, ys, marker="o", label=f"f_n={f_n:.0e}")

        plt.xlabel("T [K]")
        plt.ylabel("Gas in bulk bubbles [% generated]")
        plt.title(
            f"Gas to bulk bubbles | BU=3.2% | "
            f"R_b_seed={R_b_seed_nm:.1f} nm"
        )
        plt.grid(True, alpha=0.3)
        plt.legend(fontsize=8)
        plt.tight_layout()
        plt.show()

        plt.figure(figsize=(9, 5))
        for f_n in FN_VALUES:
            rows = all_results[(R_b_seed_nm, burnup, f_n)]
            Ts = [r["T"] for r in rows]
            ys = [r["qgb_pct"] for r in rows]
            plt.plot(Ts, ys, marker="o", label=f"f_n={f_n:.0e}")

        plt.xlabel("T [K]")
        plt.ylabel("Gas to grain face [% generated]")
        plt.title(
            f"Gas to grain face | BU=3.2% | "
            f"R_b_seed={R_b_seed_nm:.1f} nm"
        )
        plt.grid(True, alpha=0.3)
        plt.legend(fontsize=8)
        plt.tight_layout()
        plt.show()

    # --------------------------------------------------------
    # 3) Bubble number density and radius at 1.3% FIMA
    # --------------------------------------------------------
    burnup = 1.3

    for R_b_seed_nm in RB_SEED_VALUES_NM:

        plt.figure(figsize=(9, 5))
        for f_n in FN_VALUES:
            rows = all_results[(R_b_seed_nm, burnup, f_n)]
            Ts = [r["T"] for r in rows]
            ys = [r["Nb"] for r in rows]
            plt.plot(Ts, ys, marker="o", label=f"bulk, f_n={f_n:.0e}")

        rows_ref = all_results[(R_b_seed_nm, burnup, FN_VALUES[0])]
        Ts_ref = [r["T"] for r in rows_ref]
        Nd_ref = [r["Nd"] for r in rows_ref]
        plt.plot(Ts_ref, Nd_ref, "k--", label="dislocation N_d")

        plt.yscale("log")
        plt.xlabel("T [K]")
        plt.ylabel("Bubble concentration [m$^{-3}$]")
        plt.title(
            f"Bubble concentrations | BU=1.3% | "
            f"R_b_seed={R_b_seed_nm:.1f} nm"
        )
        plt.grid(True, which="both", alpha=0.3)
        plt.legend(fontsize=8)
        plt.tight_layout()
        plt.show()

        plt.figure(figsize=(9, 5))
        for f_n in FN_VALUES:
            rows = all_results[(R_b_seed_nm, burnup, f_n)]
            Ts = [r["T"] for r in rows]
            Rb = [r["Rb_nm"] for r in rows]
            Rd = [r["Rd_nm"] for r in rows]
            plt.plot(Ts, Rb, marker="o", linestyle=":", label=f"R_b, f_n={f_n:.0e}")
            plt.plot(Ts, Rd, marker="o", linestyle="--", label=f"R_d, f_n={f_n:.0e}")

        plt.yscale("log")
        plt.xlabel("T [K]")
        plt.ylabel("Bubble radius [nm]")
        plt.title(
            f"Bubble radii | BU=1.3% | "
            f"R_b_seed={R_b_seed_nm:.1f} nm"
        )
        plt.grid(True, which="both", alpha=0.3)
        plt.legend(fontsize=7)
        plt.tight_layout()
        plt.show()

    # --------------------------------------------------------
    # 4) Pressure ratios at 3.2% FIMA
    # --------------------------------------------------------
    burnup = 3.2

    for R_b_seed_nm in RB_SEED_VALUES_NM:

        plt.figure(figsize=(9, 5))
        for f_n in FN_VALUES:
            rows = all_results[(R_b_seed_nm, burnup, f_n)]
            Ts = [r["T"] for r in rows]
            ys = [r["p_b_ratio"] for r in rows]
            plt.plot(Ts, ys, marker="o", label=f"bulk, f_n={f_n:.0e}")

        plt.axhline(1.0, linestyle="--", linewidth=1.0)
        plt.xlabel("T [K]")
        plt.ylabel(r"$p_b/p_{b,eq}$ [-]")
        plt.title(
            f"Bulk pressure ratio | BU=3.2% | "
            f"R_b_seed={R_b_seed_nm:.1f} nm"
        )
        plt.grid(True, alpha=0.3)
        plt.legend(fontsize=8)
        plt.tight_layout()
        plt.show()

        plt.figure(figsize=(9, 5))
        for f_n in FN_VALUES:
            rows = all_results[(R_b_seed_nm, burnup, f_n)]
            Ts = [r["T"] for r in rows]
            ys = [r["p_d_ratio"] for r in rows]
            plt.plot(Ts, ys, marker="o", label=f"dislocation, f_n={f_n:.0e}")

        plt.axhline(1.0, linestyle="--", linewidth=1.0)
        plt.xlabel("T [K]")
        plt.ylabel(r"$p_d/p_{d,eq}$ [-]")
        plt.title(
            f"Dislocation pressure ratio | BU=3.2% | "
            f"R_b_seed={R_b_seed_nm:.1f} nm"
        )
        plt.grid(True, alpha=0.3)
        plt.legend(fontsize=8)
        plt.tight_layout()
        plt.show()

else:
    print("\nmatplotlib is not installed: plots skipped.")


print("\n" + "#" * 150)
print("END f_n DIAGNOSTIC - 3x3 SOLVER ONLY")
print("#" * 150)
````

````python
# LIGHTWEIGHT TARGETED CALIBRATION CELL
#
# Versione leggera:
# - niente cache di hist completi
# - salva solo valori finali
# - stampa avanzamento a ogni run
# - sensitivity principale con xe_yield = 0.24
# - una sola prova rapida con xe_yield = 0.475
# ============================================================

import math
import gc

try:
    import matplotlib.pyplot as plt
except ModuleNotFoundError:
    plt = None


# ============================================================
# CHECK
# ============================================================

required_names = [
    "UNParameters",
    "solve_UN_sciantix_intragranular_extended",
]

missing = [name for name in required_names if name not in globals()]
if missing:
    raise RuntimeError(
        "Prima devi eseguire la cella principale del modello. "
        f"Mancano: {missing}"
    )


# ============================================================
# SETTINGS
# ============================================================

FISSION_RATE = 5.0e19
GRAIN_RADIUS = 6.0e-6
DT = 3600.0

# Sensitivity principale: yield Xe-only
YIELD_MAIN = 0.24

# Test rapido: gas + volatili, NON definitivo
YIELD_TEST = 0.475

FN_VALUES = [
    1.0e-9,
    3.0e-9,
    1.0e-8,
]

KD_VALUES = [
    5.0e5,
    1.2e6,
]

BURNUPS = [
    1.1,
    1.3,
    3.2,
]

# Temperature mirate, non tutto lo sweep denso
TEMPS_BY_BURNUP = {
    1.1: [900.0, 1300.0, 1500.0, 1550.0, 1600.0, 1650.0, 1700.0, 1800.0],
    1.3: [900.0, 1300.0, 1500.0, 1550.0, 1600.0, 1650.0, 1700.0, 1800.0],
    3.2: [900.0, 1300.0, 1500.0, 1550.0, 1600.0, 1650.0, 1700.0, 1800.0, 2000.0, 2400.0, 2600.0],
}


# ============================================================
# FUNZIONI
# ============================================================

def run_final_only(
    T,
    burnup,
    f_n,
    K_d,
    xe_yield,
):
    p = UNParameters(
        temperature=float(T),
        fission_rate=FISSION_RATE,
        grain_radius=GRAIN_RADIUS,
        target_burnup_percent_fima=float(burnup),
        dt=DT,
        n_modes=40,
        xe_yield=float(xe_yield),
        f_n=float(f_n),
        K_d=float(K_d),
        R_b=0.0,
        N_b=0.0,
        R_d=0.0,
        N_d=None,
        c0=0.0,
        mb0=0.0,
        md0=0.0,
        vacancy_absorption_only=True,
        update_bulk_vacancies=True,
    )

    hist, rates = solve_UN_sciantix_intragranular_extended(p)

    generated = hist["generated"][-1]

    if generated > 0.0:
        matrix_pct = 100.0 * hist["c"][-1] / generated
        bulk_gas_pct = 100.0 * hist["mb"][-1] / generated
        disl_gas_pct = 100.0 * hist["md"][-1] / generated
        qgb_pct = 100.0 * hist["q_gb"][-1] / generated
    else:
        matrix_pct = 0.0
        bulk_gas_pct = 0.0
        disl_gas_pct = 0.0
        qgb_pct = 0.0

    p_b = hist["p_b"][-1]
    p_b_eq = hist["p_b_eq"][-1]
    p_d = hist["p_d"][-1]
    p_d_eq = hist["p_d_eq"][-1]

    row = {
        "T": float(T),
        "burnup": float(burnup),
        "f_n": float(f_n),
        "K_d": float(K_d),
        "N_d0": float(K_d) * p.rho_d,
        "xe_yield": float(xe_yield),
        "time_d": p.final_time / 86400.0,

        "sw_b_pct": 100.0 * hist["swelling_b"][-1],
        "sw_d_pct": 100.0 * hist["swelling_d"][-1],
        "sw_ig_pct": 100.0 * hist["swelling_ig"][-1],

        "Rb_nm": hist["Rb"][-1] * 1.0e9,
        "Rd_nm": hist["Rd"][-1] * 1.0e9,
        "Nb": hist["Nb"][-1],
        "Nd": hist["Nd"][-1],

        "matrix_pct": matrix_pct,
        "bulk_gas_pct": bulk_gas_pct,
        "disl_gas_pct": disl_gas_pct,
        "qgb_pct": qgb_pct,

        "p_b": p_b,
        "p_b_eq": p_b_eq,
        "p_d": p_d,
        "p_d_eq": p_d_eq,
        "p_b_ratio": p_b / p_b_eq if p_b_eq > 0.0 else math.nan,
        "p_d_ratio": p_d / p_d_eq if p_d_eq > 0.0 else math.nan,

        "Dg": rates.get("Dg", math.nan),
        "Dv": rates.get("Dv", math.nan),
        "g_b": rates.get("g_b", math.nan),
        "g_d": rates.get("g_d", math.nan),
        "b_b": rates.get("b_b", math.nan),
        "b_d": rates.get("b_d", math.nan),
    }

    del hist
    del rates
    gc.collect()

    return row


def print_rows_table(rows, title):
    print("\n" + "=" * 170)
    print(title)
    print("=" * 170)
    print(
        f"{'yield':>8s} | {'BU':>5s} | {'T_K':>7s} | "
        f"{'f_n':>10s} | {'K_d':>10s} | "
        f"{'sw_b_%':>10s} | {'sw_d_%':>10s} | {'sw_ig_%':>10s} | "
        f"{'bulk_gas_%':>12s} | {'disl_gas_%':>12s} | {'qgb_%':>10s} | "
        f"{'Rb_nm':>10s} | {'Rd_nm':>10s} | "
        f"{'Nb':>12s} | {'Nd':>12s} | "
        f"{'pb/peq':>10s} | {'pd/peq':>10s}"
    )
    print("-" * 170)

    for r in rows:
        print(
            f"{r['xe_yield']:8.3f} | "
            f"{r['burnup']:5.1f} | "
            f"{r['T']:7.1f} | "
            f"{r['f_n']:10.1e} | "
            f"{r['K_d']:10.1e} | "
            f"{r['sw_b_pct']:10.4e} | "
            f"{r['sw_d_pct']:10.4e} | "
            f"{r['sw_ig_pct']:10.4e} | "
            f"{r['bulk_gas_pct']:12.4e} | "
            f"{r['disl_gas_pct']:12.4e} | "
            f"{r['qgb_pct']:10.4e} | "
            f"{r['Rb_nm']:10.4e} | "
            f"{r['Rd_nm']:10.4e} | "
            f"{r['Nb']:12.4e} | "
            f"{r['Nd']:12.4e} | "
            f"{r['p_b_ratio']:10.4e} | "
            f"{r['p_d_ratio']:10.4e}"
        )


def rows_for_case(rows, burnup, f_n, K_d, xe_yield):
    return [
        r for r in rows
        if (
            abs(r["burnup"] - burnup) < 1.0e-12
            and abs(r["f_n"] - f_n) / f_n < 1.0e-12
            and abs(r["K_d"] - K_d) / K_d < 1.0e-12
            and abs(r["xe_yield"] - xe_yield) < 1.0e-12
        )
    ]


# ============================================================
# MAIN RUN: xe_yield = 0.24
# ============================================================

main_rows = []

total_runs = 0
for burnup in BURNUPS:
    total_runs += len(TEMPS_BY_BURNUP[burnup]) * len(FN_VALUES) * len(KD_VALUES)

run_counter = 0

print("\n" + "#" * 150)
print("START LIGHTWEIGHT MAIN SENSITIVITY")
print(f"xe_yield = {YIELD_MAIN}")
print(f"Total runs = {total_runs}")
print("#" * 150)

for burnup in BURNUPS:
    for f_n in FN_VALUES:
        for K_d in KD_VALUES:
            for T in TEMPS_BY_BURNUP[burnup]:
                run_counter += 1

                print(
                    f"[{run_counter:03d}/{total_runs:03d}] "
                    f"yield={YIELD_MAIN:.3f}, "
                    f"BU={burnup:.1f}, T={T:.0f} K, "
                    f"f_n={f_n:.1e}, K_d={K_d:.1e}",
                    flush=True,
                )

                row = run_final_only(
                    T=T,
                    burnup=burnup,
                    f_n=f_n,
                    K_d=K_d,
                    xe_yield=YIELD_MAIN,
                )

                main_rows.append(row)


print_rows_table(
    main_rows,
    "MAIN SENSITIVITY RESULTS - xe_yield = 0.24",
)


# ============================================================
# QUICK TEST: xe_yield = 0.475, one selected case only
# ============================================================

yield_test_rows = []

# scelgo un caso promettente e non troppo estremo
YIELD_TEST_FN = 3.0e-9
YIELD_TEST_KD = 5.0e5
YIELD_TEST_BURNUP = 3.2
YIELD_TEST_TEMPS = [1500.0, 1600.0, 1700.0, 1800.0, 2000.0]

print("\n" + "#" * 150)
print("START QUICK YIELD TEST")
print(
    f"xe_yield = {YIELD_TEST}, "
    f"f_n = {YIELD_TEST_FN:.1e}, "
    f"K_d = {YIELD_TEST_KD:.1e}, "
    f"BU = {YIELD_TEST_BURNUP:.1f}%"
)
print("#" * 150)

for i, T in enumerate(YIELD_TEST_TEMPS, start=1):
    print(
        f"[yield-test {i}/{len(YIELD_TEST_TEMPS)}] "
        f"yield={YIELD_TEST:.3f}, T={T:.0f} K",
        flush=True,
    )

    row = run_final_only(
        T=T,
        burnup=YIELD_TEST_BURNUP,
        f_n=YIELD_TEST_FN,
        K_d=YIELD_TEST_KD,
        xe_yield=YIELD_TEST,
    )

    yield_test_rows.append(row)


print_rows_table(
    yield_test_rows,
    "QUICK TEST RESULTS - xe_yield = 0.475, one selected case",
)


# ============================================================
# PLOTS
# ============================================================

if plt is not None:

    # --------------------------------------------------------
    # 1) Dislocation swelling vs T, grouped by burnup
    # --------------------------------------------------------
    for burnup in BURNUPS:
        plt.figure(figsize=(9, 5))

        for f_n in FN_VALUES:
            for K_d in KD_VALUES:
                rows = rows_for_case(
                    main_rows,
                    burnup=burnup,
                    f_n=f_n,
                    K_d=K_d,
                    xe_yield=YIELD_MAIN,
                )

                rows = sorted(rows, key=lambda r: r["T"])

                Ts = [r["T"] for r in rows]
                ys = [r["sw_d_pct"] for r in rows]

                plt.plot(
                    Ts,
                    ys,
                    marker="o",
                    label=f"f_n={f_n:.0e}, K_d={K_d:.1e}"
                )

        plt.xlabel("T [K]")
        plt.ylabel("Dislocation bubble swelling [%]")
        plt.title(f"Dislocation swelling sensitivity, BU = {burnup:.1f}% FIMA")
        plt.grid(True, alpha=0.3)
        plt.legend(fontsize=8)
        plt.tight_layout()
        plt.show()


    # --------------------------------------------------------
    # 2) Bulk swelling vs T, grouped by burnup
    # --------------------------------------------------------
    for burnup in BURNUPS:
        plt.figure(figsize=(9, 5))

        for f_n in FN_VALUES:
            for K_d in KD_VALUES:
                rows = rows_for_case(
                    main_rows,
                    burnup=burnup,
                    f_n=f_n,
                    K_d=K_d,
                    xe_yield=YIELD_MAIN,
                )

                rows = sorted(rows, key=lambda r: r["T"])

                Ts = [r["T"] for r in rows]
                ys = [r["sw_b_pct"] for r in rows]

                plt.plot(
                    Ts,
                    ys,
                    marker="o",
                    label=f"f_n={f_n:.0e}, K_d={K_d:.1e}"
                )

        plt.xlabel("T [K]")
        plt.ylabel("Bulk bubble swelling [%]")
        plt.title(f"Bulk swelling sensitivity, BU = {burnup:.1f}% FIMA")
        plt.grid(True, alpha=0.3)
        plt.legend(fontsize=8)
        plt.tight_layout()
        plt.show()


    # --------------------------------------------------------
    # 3) Gas partition at 3.2% FIMA
    # --------------------------------------------------------
    burnup = 3.2

    for f_n in FN_VALUES:
        for K_d in KD_VALUES:
            rows = rows_for_case(
                main_rows,
                burnup=burnup,
                f_n=f_n,
                K_d=K_d,
                xe_yield=YIELD_MAIN,
            )

            rows = sorted(rows, key=lambda r: r["T"])

            Ts = [r["T"] for r in rows]

            plt.figure(figsize=(9, 5))
            plt.plot(Ts, [r["matrix_pct"] for r in rows], marker="o", label="Matrix")
            plt.plot(Ts, [r["bulk_gas_pct"] for r in rows], marker="o", linestyle=":", label="Bulk bubbles")
            plt.plot(Ts, [r["disl_gas_pct"] for r in rows], marker="o", linestyle="--", label="Dislocation bubbles")
            plt.plot(Ts, [r["qgb_pct"] for r in rows], marker="o", linestyle="-.", label="Gas to grain face")

            plt.xlabel("T [K]")
            plt.ylabel("Generated gas fraction [%]")
            plt.title(
                f"Gas partition at 3.2% FIMA\n"
                f"xe_yield=0.24, f_n={f_n:.0e}, K_d={K_d:.1e}"
            )
            plt.grid(True, alpha=0.3)
            plt.legend()
            plt.tight_layout()
            plt.show()


    # --------------------------------------------------------
    # 4) Yield quick comparison
    # --------------------------------------------------------
    rows_024 = rows_for_case(
        main_rows,
        burnup=YIELD_TEST_BURNUP,
        f_n=YIELD_TEST_FN,
        K_d=YIELD_TEST_KD,
        xe_yield=YIELD_MAIN,
    )

    rows_024 = [
        r for r in rows_024
        if r["T"] in YIELD_TEST_TEMPS
    ]

    rows_024 = sorted(rows_024, key=lambda r: r["T"])
    rows_0475 = sorted(yield_test_rows, key=lambda r: r["T"])

    plt.figure(figsize=(9, 5))
    plt.plot(
        [r["T"] for r in rows_024],
        [r["sw_d_pct"] for r in rows_024],
        marker="o",
        label="xe_yield = 0.24"
    )
    plt.plot(
        [r["T"] for r in rows_0475],
        [r["sw_d_pct"] for r in rows_0475],
        marker="o",
        label="xe_yield = 0.475"
    )

    plt.xlabel("T [K]")
    plt.ylabel("Dislocation bubble swelling [%]")
    plt.title(
        f"Quick yield comparison, BU={YIELD_TEST_BURNUP:.1f}% FIMA\n"
        f"f_n={YIELD_TEST_FN:.0e}, K_d={YIELD_TEST_KD:.1e}"
    )
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

else:
    print("\nmatplotlib is not installed: plots skipped.")


print("\n" + "#" * 150)
print("LIGHTWEIGHT CALIBRATION COMPLETE")
print("#" * 150)
````

````python
# LIGHTWEIGHT f_n CALIBRATION / RIZK-STYLE DIAGNOSTIC
#
# Richiede di avere già eseguito la cella principale del modello.
# NON modifica il solver principale: definisce un solver finale locale
# più leggero per fare sweep diagnostici.
#
# Scopo:
# - tenere K_d fisso = 5e5 bub/m;
# - variare prima solo f_n;
# - usare un seed bulk fisso opzionale per controllare la pressione bulk;
# - stampare/plotare le grandezze confrontabili con Rizk:
#   swelling, pressioni, N, R, gas partition.
# ============================================================

import math
from dataclasses import replace

try:
    import matplotlib.pyplot as plt
except ModuleNotFoundError:
    plt = None


# ============================================================
# CONFIGURAZIONE LEGGERA
# ============================================================

FISSION_RATE = 5.0e19
GRAIN_RADIUS = 5.0e-6
K_D_FIXED = 5.0e5
XE_YIELD = 0.24

# Diagnostica veloce. Per il fit finale puoi abbassare a 3600 s.
DT_FAST = 6.0 * 3600.0
N_MODES_FAST = 30

# Seed fisso: serve solo a evitare che le bulk bubbles nascano con n_vac=0.
# Dai test precedenti, 2 nm porta p_b < p_eq a bassa T, coerente con Fig. 5 Rizk.
R_B_SEED_NM = 2.0

F_N_LIST = [
    3.0e-10,
    5.0e-10,
    7.0e-10,
    1.0e-9,
    2.0e-9,
    3.0e-9,
    1.0e-8,
    1.0e-6,
]

BURNUPS_FOR_SCORE = [1.1, 1.3, 3.2]
TEMPS_SCORE = [1200.0, 1500.0, 1600.0, 1700.0, 1800.0, 2000.0]
TEMPS_FINE = [float(T) for T in range(900, 2001, 50)]
TEMPS_GAS_PARTITION = [float(T) for T in range(900, 2601, 100)]


# Target qualitativi digitizzati a mano/occhio dalle figure Rizk.
# Non sono dati sperimentali: servono solo per ordinare le prove.
TARGET_SW_D_AT_1600 = {
    1.1: 2.7,
    1.3: 3.3,
    3.2: 4.5,
}

TARGET_SW_D_AT_1700 = {
    1.1: 5.5,
    1.3: 6.5,
    3.2: 9.0,
}


# ============================================================
# CONTROLLI
# ============================================================

_required = [
    "UNParameters",
    "omega_matrix",
    "sphere_volume",
    "radius_from_volume",
    "xe_diffusivity_UN",
    "vacancy_diffusivity_UN",
    "resolution_rates_UN",
    "trapping_rates_UN",
    "beta_production",
    "nucleation_rate_bulk",
    "phi_bulk",
    "coalescence_lambda",
    "pressure_internal",
    "pressure_equilibrium",
    "wigner_seitz_delta",
    "zeta_geometry",
    "initialize_vacancy_concentration",
    "initialize_modes_from_average",
    "reconstruct_average",
    "sciantix_3x3_exchange_step",
    "time_to_burnup_percent",
]

_missing = [name for name in _required if name not in globals()]
if _missing:
    raise RuntimeError(
        "Prima devi eseguire la cella principale del modello. Mancano: "
        + ", ".join(_missing)
    )


# ============================================================
# FUNZIONI LOCALI: vacancy update con raggio gas-only se R=0
# ============================================================

def _gas_only_radius_for_population(p, m_gas, N):
    if m_gas <= 0.0 or N <= 0.0:
        return 0.0

    V_gas_per_bubble = p.omega_fg * m_gas / N
    return radius_from_volume(V_gas_per_bubble)


def _radius_for_vacancy_update(p, R_old, N, m_gas):
    if R_old > 0.0:
        return R_old

    R_gas = _gas_only_radius_for_population(p, m_gas, N)
    if R_gas > 0.0:
        return R_gas

    return 0.0


def _pressure_equilibrium_for_vacancy_update(p, R_update):
    if R_update <= 0.0:
        return math.inf
    return 2.0 * p.gamma_b / R_update - p.hydrostatic_stress


def _vacancy_concentration_implicit_step_local(p, Dv, R, N, m_gas, n_old, dt):
    if N <= 0.0 or m_gas <= 0.0:
        return n_old, 0.0

    R_update = _radius_for_vacancy_update(p, R, N, m_gas)
    if R_update <= 0.0:
        return n_old, 0.0

    p_eq = _pressure_equilibrium_for_vacancy_update(p, R_update)
    p_int_old = pressure_internal(p, m_gas, n_old)

    if p.vacancy_absorption_only and p_int_old <= p_eq:
        return n_old, 0.0

    delta = wigner_seitz_delta(N)
    zeta = zeta_geometry(R_update, N)

    A = (
        2.0
        * math.pi
        * Dv
        * delta
        * N
        / (p.kB_J * p.temperature * zeta)
    )

    C = p.kB_J * p.temperature * m_gas / omega_matrix(p)

    B = n_old - dt * A * p_eq
    discriminant = B * B + 4.0 * dt * A * C

    if discriminant < 0.0:
        raise ValueError(f"Negative discriminant in vacancy step: {discriminant:g}")

    sqrt_disc = math.sqrt(discriminant)

    if B >= 0.0:
        n_new = 0.5 * (B + sqrt_disc)
    else:
        denominator = sqrt_disc - B
        n_new = 0.0 if denominator <= 0.0 else (2.0 * dt * A * C) / denominator

    if p.vacancy_absorption_only:
        n_new = max(n_new, n_old)

    return n_new, (n_new - n_old) / dt


# ============================================================
# SOLVER FINALE LEGGERO
# ============================================================

def solve_UN_final_lightweight(p, R_b_seed_nm=0.0):
    """
    Solver diagnostico finale.

    Differenze rispetto al principale:
    - non salva tutta la storia temporale: salva solo lo stato finale;
    - aggiorna N_b prima del gas solver, come ordine SCIANTIX-like;
    - opzionale seed di volume/vacanze per nuove bulk bubbles;
    - tiene la fisica intragranulare del modello principale.
    """

    modes_c = initialize_modes_from_average(p.c0, p.n_modes)
    modes_mb = initialize_modes_from_average(p.mb0, p.n_modes)
    modes_md = initialize_modes_from_average(p.md0, p.n_modes)

    c = reconstruct_average(modes_c)
    mb = reconstruct_average(modes_mb)
    md = reconstruct_average(modes_md)

    R_b = p.R_b
    R_d = p.R_d
    N_b = p.N_b
    N_d = p.N_d

    V_b = sphere_volume(R_b)
    V_d = sphere_volume(R_d)

    if p.nvb0 is None:
        nvb = initialize_vacancy_concentration(p, N_b, R_b, p.mb0)
    else:
        nvb = p.nvb0

    if p.nvd0 is None:
        nvd = initialize_vacancy_concentration(p, N_d, R_d, p.md0)
    else:
        nvd = p.nvd0

    beta = beta_production(p)

    initial_gas = p.c0 + p.mb0 + p.md0
    generated = 0.0
    q_gb = 0.0
    retained = initial_gas
    seed_vacancies_added_total = 0.0

    n_steps = int(math.ceil(p.final_time / p.dt))
    t = 0.0

    last_rates = {}

    for _ in range(n_steps):
        dt = min(p.dt, p.final_time - t)
        if dt <= 0.0:
            break

        c_old = c
        mb_old = mb
        md_old = md

        Nb_old = N_b
        Nd_old = N_d
        Vd_old = V_d

        Dg, D_parts = xe_diffusivity_UN(p)
        Dv, Dv_parts = vacancy_diffusivity_UN(p)

        b_b, b_d = resolution_rates_UN(p, R_b, R_d)

        g_b, g_d, trapping_parts = trapping_rates_UN(
            p, Dg, R_b, Nb_old, R_d, Nd_old
        )

        # 1) N_b prima della diffusione gas
        nu_b = nucleation_rate_bulk(p, Dg, c_old)
        phi_b = phi_bulk(mb_old, Nb_old)

        N_b = (Nb_old + dt * nu_b) / (1.0 + dt * b_b * phi_b)
        N_b = max(N_b, p.min_number_density)

        # 2) seed volume/vacanze per le nuove bulk bubbles
        dN_b_positive = max(N_b - Nb_old, 0.0)
        seed_vacancies_added_step = 0.0

        if dN_b_positive > 0.0 and R_b_seed_nm > 0.0:
            V_seed = sphere_volume(R_b_seed_nm * 1.0e-9)
            seed_vacancies_added_step = dN_b_positive * V_seed / omega_matrix(p)
            nvb += seed_vacancies_added_step
            seed_vacancies_added_total += seed_vacancies_added_step

        # 3) gas diffusion / trapping / resolution
        c, mb, md = sciantix_3x3_exchange_step(
            modes_c,
            modes_mb,
            modes_md,
            Dg,
            p.grain_radius,
            beta,
            g_b,
            g_d,
            b_b,
            b_d,
            dt,
        )

        dmb_dt = (mb - mb_old) / dt
        dmd_dt = (md - md_old) / dt

        # 4) vacancy update
        if p.update_bulk_vacancies:
            nvb, dnvb_dt = _vacancy_concentration_implicit_step_local(
                p, Dv, R_b, N_b, mb, nvb, dt
            )
        else:
            dnvb_dt = 0.0

        nvd, dnvd_dt = _vacancy_concentration_implicit_step_local(
            p, Dv, R_d, Nd_old, md, nvd, dt
        )

        # 5) volume growth
        if N_b > 0.0:
            V_b_growth = V_b + dt * (
                p.omega_fg / N_b * dmb_dt
                + omega_matrix(p) / N_b * dnvb_dt
            )
            V_b_growth = max(V_b_growth, p.min_volume)
        else:
            V_b_growth = V_b

        if Nd_old > 0.0:
            dVd_growth_dt = (
                p.omega_fg / Nd_old * dmd_dt
                + omega_matrix(p) / Nd_old * dnvd_dt
            )
            V_d_growth = V_d + dt * dVd_growth_dt
            V_d_growth = max(V_d_growth, p.min_volume)
        else:
            dVd_growth_dt = 0.0
            V_d_growth = V_d

        # 6) coalescence dislocation bubbles
        lambda_d = coalescence_lambda(Vd_old, Nd_old)
        dVd_positive = max(V_d_growth - Vd_old, 0.0)

        if dVd_positive > 0.0 and Nd_old > 0.0:
            N_d = Nd_old / (1.0 + 4.0 * lambda_d * Nd_old * dVd_positive)
        else:
            N_d = Nd_old

        N_d = max(N_d, p.min_number_density)

        # 7) reconstruct volumes from gas + vacancies
        if N_b > 0.0:
            V_b = (
                p.omega_fg * max(mb, 0.0)
                + omega_matrix(p) * nvb
            ) / N_b
        else:
            V_b = 0.0

        if N_d > 0.0:
            V_d = (
                p.omega_fg * max(md, 0.0)
                + omega_matrix(p) * nvd
            ) / N_d
        else:
            V_d = 0.0

        V_b = max(V_b, p.min_volume)
        V_d = max(V_d, p.min_volume)

        R_b = radius_from_volume(V_b)
        R_d = radius_from_volume(V_d)

        generated += beta * dt
        retained = c + mb + md
        q_gb = max(initial_gas + generated - retained, 0.0)

        t += dt

        last_rates = {
            "Dg": Dg,
            "Dv": Dv,
            "beta": beta,
            "g_b": g_b,
            "g_d": g_d,
            "b_b": b_b,
            "b_d": b_d,
            "nu_b": nu_b,
            "phi_b": phi_b,
            "lambda_d": lambda_d,
            "dVd_growth_dt": dVd_growth_dt,
            "dnvb_dt": dnvb_dt,
            "dnvd_dt": dnvd_dt,
            "seed_vacancies_added_step": seed_vacancies_added_step,
            "seed_vacancies_added_total": seed_vacancies_added_total,
            **D_parts,
            **Dv_parts,
            **trapping_parts,
        }

    p_b = pressure_internal(p, mb, nvb)
    p_d = pressure_internal(p, md, nvd)
    p_b_eq = pressure_equilibrium(p, R_b)
    p_d_eq = pressure_equilibrium(p, R_d)

    return {
        "T": p.temperature,
        "burnup": time_to_burnup_percent(t, p.fission_rate, p.lattice_parameter),
        "time_d": t / (3600.0 * 24.0),

        "c": c,
        "mb": mb,
        "md": md,
        "generated": generated,
        "retained": retained,
        "q_gb": q_gb,

        "matrix_percent": 100.0 * c / generated if generated > 0.0 else 0.0,
        "bulk_gas_percent": 100.0 * mb / generated if generated > 0.0 else 0.0,
        "disl_gas_percent": 100.0 * md / generated if generated > 0.0 else 0.0,
        "qgb_percent": 100.0 * q_gb / generated if generated > 0.0 else 0.0,

        "Nb": N_b,
        "Nd": N_d,
        "Vb": V_b,
        "Vd": V_d,
        "Rb_nm": R_b * 1.0e9,
        "Rd_nm": R_d * 1.0e9,
        "nvb": nvb,
        "nvd": nvd,

        "sw_b_percent": 100.0 * N_b * V_b,
        "sw_d_percent": 100.0 * N_d * V_d,
        "sw_ig_percent": 100.0 * (N_b * V_b + N_d * V_d),

        "p_b": p_b,
        "p_b_eq": p_b_eq,
        "p_d": p_d,
        "p_d_eq": p_d_eq,
        "p_b_over_eq": p_b / p_b_eq if p_b_eq > 0.0 else math.nan,
        "p_d_over_eq": p_d / p_d_eq if p_d_eq > 0.0 else math.nan,

        "rates": last_rates,
    }


def run_one(f_n, T, burnup, R_b_seed_nm=R_B_SEED_NM, dt=DT_FAST, n_modes=N_MODES_FAST):
    p = UNParameters(
        temperature=float(T),
        fission_rate=FISSION_RATE,
        grain_radius=GRAIN_RADIUS,
        target_burnup_percent_fima=float(burnup),
        dt=float(dt),
        n_modes=int(n_modes),
        xe_yield=XE_YIELD,
        f_n=float(f_n),
        K_d=K_D_FIXED,
        R_b=0.0,
        N_b=0.0,
        R_d=0.0,
        N_d=None,
        c0=0.0,
        mb0=0.0,
        md0=0.0,
        vacancy_absorption_only=True,
        update_bulk_vacancies=True,
    )

    out = solve_UN_final_lightweight(p, R_b_seed_nm=R_b_seed_nm)
    out["f_n"] = float(f_n)
    out["R_b_seed_nm"] = float(R_b_seed_nm)
    out["K_d"] = K_D_FIXED
    return out


# ============================================================
# SCORE f_n, K_d FISSO
# ============================================================

def score_row(row):
    score = 0.0

    # Score principale: dislocation swelling a 1600 e 1700 K.
    for burn, target in TARGET_SW_D_AT_1600.items():
        if abs(row["burnup"] - burn) < 1.0e-6 and abs(row["T"] - 1600.0) < 1.0e-6:
            score += abs(math.log10(max(row["sw_d_percent"], 1.0e-30) / target))

    for burn, target in TARGET_SW_D_AT_1700.items():
        if abs(row["burnup"] - burn) < 1.0e-6 and abs(row["T"] - 1700.0) < 1.0e-6:
            score += 0.5 * abs(math.log10(max(row["sw_d_percent"], 1.0e-30) / target))

    return score


print("\n" + "=" * 140)
print("PHASE A - f_n SWEEP WITH K_d FIXED")
print("=" * 140)
print(f"K_d fixed       = {K_D_FIXED:.6e} bub/m")
print(f"R_b_seed fixed  = {R_B_SEED_NM:.3f} nm")
print(f"dt diagnostic   = {DT_FAST:.1f} s")
print(f"n_modes         = {N_MODES_FAST}")
print("-" * 140)

score_rows = []

for f_n in F_N_LIST:
    rows = []

    print(f"Running f_n = {f_n:.3e}")

    for burnup in BURNUPS_FOR_SCORE:
        for T in TEMPS_SCORE:
            rows.append(run_one(f_n, T, burnup))

    score = 0.0
    for r in rows:
        score += score_row(r)

    # Add pressure sanity at 3.2% / 900 K if available from an extra run.
    p900 = run_one(f_n, 900.0, 3.2)
    pressure_penalty = abs(math.log10(max(p900["p_b_over_eq"], 1.0e-30) / 0.75))
    score += 0.25 * pressure_penalty

    r1600_11 = [r for r in rows if abs(r["burnup"] - 1.1) < 1e-6 and r["T"] == 1600.0][0]
    r1600_13 = [r for r in rows if abs(r["burnup"] - 1.3) < 1e-6 and r["T"] == 1600.0][0]
    r1600_32 = [r for r in rows if abs(r["burnup"] - 3.2) < 1e-6 and r["T"] == 1600.0][0]
    r1700_32 = [r for r in rows if abs(r["burnup"] - 3.2) < 1e-6 and r["T"] == 1700.0][0]

    score_rows.append(
        {
            "f_n": f_n,
            "score": score,
            "sw_d_11_1600": r1600_11["sw_d_percent"],
            "sw_d_13_1600": r1600_13["sw_d_percent"],
            "sw_d_32_1600": r1600_32["sw_d_percent"],
            "sw_d_32_1700": r1700_32["sw_d_percent"],
            "sw_b_32_1600": r1600_32["sw_b_percent"],
            "pb_ratio_32_900": p900["p_b_over_eq"],
            "Rd_13_1600": r1600_13["Rd_nm"],
            "Nd_13_1600": r1600_13["Nd"],
        }
    )

score_rows = sorted(score_rows, key=lambda r: r["score"])

print("\n" + "=" * 140)
print("RANKING f_n - K_d FIXED")
print("=" * 140)
print(
    f"{'rank':>4s} | {'score':>11s} | {'f_n':>11s} | "
    f"{'swD_1.1@1600':>14s} | {'swD_1.3@1600':>14s} | "
    f"{'swD_3.2@1600':>14s} | {'swD_3.2@1700':>14s} | "
    f"{'swB_3.2@1600':>14s} | {'pb/peq@900':>12s} | "
    f"{'Rd_1.3@1600':>13s} | {'Nd_1.3@1600':>13s}"
)
print("-" * 140)

for i, r in enumerate(score_rows, start=1):
    print(
        f"{i:4d} | {r['score']:11.4e} | {r['f_n']:11.3e} | "
        f"{r['sw_d_11_1600']:14.6e} | {r['sw_d_13_1600']:14.6e} | "
        f"{r['sw_d_32_1600']:14.6e} | {r['sw_d_32_1700']:14.6e} | "
        f"{r['sw_b_32_1600']:14.6e} | {r['pb_ratio_32_900']:12.6e} | "
        f"{r['Rd_13_1600']:13.6e} | {r['Nd_13_1600']:13.6e}"
    )

best_f_n = score_rows[0]["f_n"]

print("\n" + "=" * 140)
print("SELECTED f_n FOR DETAILED DIAGNOSTICS")
print("=" * 140)
print(f"best f_n = {best_f_n:.6e}")
print(f"K_d      = {K_D_FIXED:.6e} bub/m")
print(f"R_seed   = {R_B_SEED_NM:.3f} nm")
print("=" * 140)


# ============================================================
# DETAILED RUNS FOR SELECTED f_n
# ============================================================

detailed = {}

for burnup in [1.1, 1.3, 3.2]:
    detailed[burnup] = [run_one(best_f_n, T, burnup) for T in TEMPS_FINE]

pressure_32 = detailed[3.2]
rows_13 = detailed[1.3]

gas_partition = {
    1.1: [run_one(best_f_n, T, 1.1) for T in TEMPS_GAS_PARTITION],
    3.2: [run_one(best_f_n, T, 3.2) for T in TEMPS_GAS_PARTITION],
}


# ============================================================
# TABLES
# ============================================================

def print_key_table(title, rows):
    print("\n" + "=" * 160)
    print(title)
    print("=" * 160)
    print(
        f"{'T_K':>8s} | {'sw_b_%':>10s} | {'sw_d_%':>10s} | "
        f"{'Rb_nm':>10s} | {'Rd_nm':>10s} | "
        f"{'Nb':>12s} | {'Nd':>12s} | "
        f"{'pb/peq':>10s} | {'pd/peq':>10s} | "
        f"{'matrix_%':>10s} | {'bulkGas_%':>10s} | "
        f"{'dislGas_%':>10s} | {'qgb_%':>10s}"
    )
    print("-" * 160)

    for r in rows:
        if r["T"] in [900.0, 1200.0, 1500.0, 1600.0, 1700.0, 1800.0, 2000.0]:
            print(
                f"{r['T']:8.1f} | {r['sw_b_percent']:10.4e} | {r['sw_d_percent']:10.4e} | "
                f"{r['Rb_nm']:10.4e} | {r['Rd_nm']:10.4e} | "
                f"{r['Nb']:12.6e} | {r['Nd']:12.6e} | "
                f"{r['p_b_over_eq']:10.4e} | {r['p_d_over_eq']:10.4e} | "
                f"{r['matrix_percent']:10.4e} | {r['bulk_gas_percent']:10.4e} | "
                f"{r['disl_gas_percent']:10.4e} | {r['qgb_percent']:10.4e}"
            )

for burnup, rows in detailed.items():
    print_key_table(f"DETAILED TABLE - best f_n = {best_f_n:.3e}, burnup = {burnup:.1f}% FIMA", rows)


# ============================================================
# PLOTS
# ============================================================

if plt is not None:
    # Fig. 3 style: swelling vs T, three burnups
    for burnup, rows in detailed.items():
        Ts = [r["T"] for r in rows]
        plt.figure(figsize=(8, 5))
        plt.plot(Ts, [r["sw_b_percent"] for r in rows], marker="o", linestyle=":", label="Bulk")
        plt.plot(Ts, [r["sw_d_percent"] for r in rows], marker="o", linestyle="--", label="Dislocation")
        plt.title(f"Fission gas swelling at {burnup:.1f}% FIMA | f_n={best_f_n:.1e}, K_d fixed")
        plt.xlabel("T [K]")
        plt.ylabel("Fission gas swelling [%]")
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()

    # Fig. 5 style: pressure
    Ts = [r["T"] for r in pressure_32]
    plt.figure(figsize=(8, 5))
    plt.plot(Ts, [r["p_b"] for r in pressure_32], label="Bulk bubble pressure")
    plt.plot(Ts, [r["p_b_eq"] for r in pressure_32], linestyle=":", label="Bulk equilibrium pressure")
    plt.plot(Ts, [r["p_d"] for r in pressure_32], label="Dislocation bubble pressure")
    plt.plot(Ts, [r["p_d_eq"] for r in pressure_32], linestyle="--", label="Dislocation equilibrium pressure")
    plt.yscale("log")
    plt.title(f"Bubble pressures at 3.2% FIMA | f_n={best_f_n:.1e}")
    plt.xlabel("T [K]")
    plt.ylabel("Bubble pressure [Pa]")
    plt.grid(True, which="both", alpha=0.3)
    plt.legend(fontsize=8)
    plt.tight_layout()
    plt.show()

    # Fig. 7 style: number density at 1.3% FIMA
    Ts = [r["T"] for r in rows_13]
    plt.figure(figsize=(8, 5))
    plt.plot(Ts, [r["Nb"] for r in rows_13], linestyle=":", marker="o", label="Bulk")
    plt.plot(Ts, [r["Nd"] for r in rows_13], linestyle="--", marker="o", label="Dislocation")
    plt.yscale("log")
    plt.title(f"Bubble number density at 1.3% FIMA | f_n={best_f_n:.1e}")
    plt.xlabel("T [K]")
    plt.ylabel("Bubble concentration [m$^{-3}$]")
    plt.grid(True, which="both", alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

    # Fig. 8 style: radius at 1.3% FIMA
    plt.figure(figsize=(8, 5))
    plt.plot(Ts, [r["Rb_nm"] for r in rows_13], linestyle=":", marker="o", label="Bulk")
    plt.plot(Ts, [r["Rd_nm"] for r in rows_13], linestyle="--", marker="o", label="Dislocation")
    plt.yscale("log")
    plt.title(f"Bubble radius at 1.3% FIMA | f_n={best_f_n:.1e}")
    plt.xlabel("T [K]")
    plt.ylabel("Bubble radius [nm]")
    plt.grid(True, which="both", alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

    # Fig. 9 style: gas partition, 1.1 and 3.2% FIMA
    for burnup, rows in gas_partition.items():
        Ts = [r["T"] for r in rows]
        plt.figure(figsize=(8, 5))
        plt.plot(Ts, [r["matrix_percent"] for r in rows], label="Matrix")
        plt.plot(Ts, [r["bulk_gas_percent"] for r in rows], linestyle=":", label="Bulk bubbles")
        plt.plot(Ts, [r["disl_gas_percent"] for r in rows], linestyle="--", label="Dislocation bubbles")
        plt.plot(Ts, [r["qgb_percent"] for r in rows], linestyle="-.", label="Gas to grain face q_gb")
        plt.title(f"Gas partition at {burnup:.1f}% FIMA | f_n={best_f_n:.1e}")
        plt.xlabel("T [K]")
        plt.ylabel("Amount of generated gas [%]")
        plt.xlim(900, 2600)
        plt.ylim(0, 100)
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()
else:
    print("\nmatplotlib is not installed: plots skipped.")
````

Questa nota raccoglie i punti sperimentali estratti graficamente dalle figure fornite da Rizk et al. per il confronto con il modello UN. La digitalizzazione è **preliminare**: è stata fatta dalle immagini caricate nel notebook/chat, non dal file dati originale. I valori sono quindi da usare per diagnosi e calibrazione iniziale; per la versione finale della tesi conviene rifare la digitalizzazione con WebPlotDigitizer direttamente dalle immagini full-size.

## Interpretazione fisica dei punti da confrontare

Nel lavoro di Rizk et al. le misure sperimentali riportate nelle figure sono associate alla popolazione di bolle intragranulari grandi, cioè la popolazione P2. Nel modello, questa popolazione viene interpretata come **dislocation bubbles**. Per questo motivo il confronto quantitativo principale del nostro modello deve essere fatto contro:

- swelling delle dislocation bubbles, non bulk + dislocation;
- concentrazione numerica delle dislocation bubbles, non bulk bubbles;
- raggio medio delle dislocation bubbles, non bulk bubbles.

Le curve bulk e grain-boundary riportate da Rizk sono invece risultati del modello e non punti sperimentali diretti. Servono per controllare la coerenza fisica globale, ma non vanno usate come primo target di calibrazione.

# 7. Strategia di calibrazione proposta

## Primo target: swelling sperimentale delle dislocation bubbles

Il primo parametro da analizzare è `f_n`, mantenendo:

```text
K_d = 5.0e5 bubble/m
rho_d = 3.0e13 m^-2
N_d0 = K_d * rho_d = 1.5e19 m^-3
xe_yield = 0.24
```

Motivo: `f_n` controlla quante bolle bulk nucleano. Se `f_n` è troppo alto, il gas viene intercettato troppo dalle bulk bubbles e le dislocation bubbles crescono troppo poco. Se `f_n` è troppo basso, le dislocation bubbles crescono meglio ma si rischia di svuotare eccessivamente il bulk, alterando la gas partition.

## Secondo target: consistenza di N_d e R_d

Dopo aver scelto un intervallo ragionevole di `f_n`, bisogna confrontare:

```text
N_d(T, 1.3% FIMA) contro Fig. 7
R_d(T, 1.3% FIMA) contro Fig. 8
```

Se lo swelling è corretto ma `N_d` e `R_d` sono sbagliati, allora il modello sta ottenendo lo swelling con una microstruttura non fisica.

## Terzo target: pressure e gas partition

Solo dopo la calibrazione sui dati sperimentali bisogna guardare:

```text
p_b, p_b_eq, p_d, p_d_eq a 3.2% FIMA
frazione gas in matrix, bulk, dislocation, q_gb
```

Queste grandezze non sono misure dirette nelle figure, ma servono per capire se il modello sta riproducendo il meccanismo fisico di Rizk.

## Cosa sospettare se `f_n` non basta

Se nessun valore di `f_n` riesce a riprodurre contemporaneamente swelling, `N_d` e `R_d`, allora il problema probabilmente non è solo la nucleazione bulk. I candidati successivi sono:

1. trapping verso dislocazioni `g_d`, cioè possibile sottostima del trasporto preferenziale verso le dislocazioni;
2. vacancy absorption e inizializzazione delle vacanze nelle bolle;
3. mancanza del modello intergranulare, che nel nostro codice è ancora rappresentato solo come gas arrivato al grain face (`q_gb`), non come vere grain-boundary bubbles;
4. uso di solo Xe stabile con yield 0.24, mentre alcune parti del modello di swelling possono implicitamente includere anche Kr e volatili.


````python
import math
from dataclasses import dataclass
from typing import Optional

try:
    import matplotlib.pyplot as plt
except ModuleNotFoundError:
    plt = None

# ============================================================
# CONFIGURAZIONE GENERALE
# ============================================================

FISSION_RATE = 5.0e19
GRAIN_RADIUS = 5.0e-6
K_D_FIXED = 5.0e5
XE_YIELD = 0.24

DT_SCAN_H = 12.0
N_MODES_SCAN = 25

# 0.0 = nessun seed artificiale. Metti 2.0 solo per diagnostica pressione bulk.
BULK_SEED_RADIUS_NM = 0.0

FN_GRID = [
    3.0e-10,
    5.0e-10,
    7.0e-10,
    1.0e-9,
    1.5e-9,
    2.0e-9,
    3.0e-9,
    5.0e-9,
    7.0e-9,
    1.0e-8,
    3.0e-8,
    1.0e-7,
    3.0e-7,
    1.0e-6,
]

# Se vuoi fare solo un test veloce, metti False e usa RUN_SINGLE_TEST=True.
RUN_CALIBRATION = True
RUN_SINGLE_TEST = False

# ============================================================
# DATI SPERIMENTALI DIGITALIZZATI DAL MARKDOWN
# ============================================================

EXP_SWELLING_T = [
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1127.0, "swelling": 0.68},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1228.0, "swelling": 0.59},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1312.0, "swelling": 0.43},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1402.0, "swelling": 0.58},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1485.0, "swelling": 1.22},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1549.0, "swelling": 1.84},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1598.0, "swelling": 1.66},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1632.0, "swelling": 2.13},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1669.0, "swelling": 3.60},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1685.0, "swelling": 2.72},

    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 899.0,  "swelling": 0.63},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1154.0, "swelling": 1.17},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1228.0, "swelling": 1.08},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1325.0, "swelling": 1.28},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1435.0, "swelling": 1.32},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1514.0, "swelling": 2.10},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1570.0, "swelling": 2.72},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1608.0, "swelling": 2.91},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1635.0, "swelling": 3.28},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1656.0, "swelling": 3.75},

    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1044.0, "swelling": 1.11},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1220.0, "swelling": 1.22},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1377.0, "swelling": 1.33},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1534.0, "swelling": 2.83},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1595.0, "swelling": 3.53},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1639.0, "swelling": 3.86},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1661.0, "swelling": 2.45},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1709.0, "swelling": 2.93},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1724.0, "swelling": 3.15},

    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 984.0,  "swelling": 0.72},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1056.0, "swelling": 1.06},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1126.0, "swelling": 1.26},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1247.0, "swelling": 1.58},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1343.0, "swelling": 1.79},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1420.0, "swelling": 2.08},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1459.0, "swelling": 2.40},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1511.0, "swelling": 2.83},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1557.0, "swelling": 3.31},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1590.0, "swelling": 3.75},
]

EXP_SWELLING_BURNUP_1600 = [
    {"burnup": 1.12, "swelling": 1.64, "series": "approx 100 kW/m"},
    {"burnup": 1.11, "swelling": 2.90, "series": "approx 119 kW/m"},
    {"burnup": 1.31, "swelling": 3.51, "series": "measurement"},
    {"burnup": 3.18, "swelling": 3.72, "series": "measurement"},
]

EXP_ND_T_13 = [
    {"T": 1153.0, "N": 5.22e19},
    {"T": 1202.0, "N": 3.69e19},
    {"T": 1227.0, "N": 3.00e19},
    {"T": 1235.0, "N": 3.57e19},
    {"T": 1248.0, "N": 2.44e19},
    {"T": 1330.0, "N": 2.90e19},
    {"T": 1338.0, "N": 4.10e19},
    {"T": 1377.0, "N": 2.71e19},
    {"T": 1408.0, "N": 1.67e19},
    {"T": 1427.0, "N": 1.85e19},
    {"T": 1493.0, "N": 3.00e19},
    {"T": 1507.0, "N": 2.44e19},
    {"T": 1524.0, "N": 1.56e19},
    {"T": 1538.0, "N": 7.54e18},
    {"T": 1555.0, "N": 7.80e18},
    {"T": 1561.0, "N": 1.92e19},
    {"T": 1599.0, "N": 5.34e18},
    {"T": 1622.0, "N": 7.80e18},
    {"T": 1628.0, "N": 1.73e19},
    {"T": 1649.0, "N": 5.34e18},
    {"T": 1656.0, "N": 3.65e18},
    {"T": 1656.0, "N": 2.41e18},
    {"T": 1669.0, "N": 1.96e18},
    {"T": 1682.0, "N": 2.44e19},
    {"T": 1685.0, "N": 1.40e19},
    {"T": 1723.0, "N": 7.70e17},
    {"T": 1742.0, "N": 1.02e18},
    {"T": 1739.0, "N": 1.65e18},
]

EXP_RD_T_13 = [
    {"T": 1045.0, "R_nm": 54.06},
    {"T": 1219.0, "R_nm": 60.37},
    {"T": 1374.0, "R_nm": 69.58},
    {"T": 1535.0, "R_nm": 104.85},
    {"T": 1594.0, "R_nm": 120.83},
    {"T": 1641.0, "R_nm": 143.73},
    {"T": 1663.0, "R_nm": 122.76},
    {"T": 1710.0, "R_nm": 157.99},
    {"T": 1725.0, "R_nm": 173.67},
]

# ============================================================
# PARAMETRI E FUNZIONI BASE
# ============================================================

@dataclass
class UNParameters:
    temperature: float = 1600.0
    fission_rate: float = 5.0e19
    grain_radius: float = 5.0e-6
    final_time: float = 24.0 * 3600.0
    target_burnup_percent_fima: Optional[float] = None
    dt: float = 60.0
    n_modes: int = 40

    xe_yield: float = 0.24
    precursor_factor: float = 1.0

    D10: float = 1.56e-3
    Q1: float = 4.94
    A20: float = 1.21e-67
    B21: float = 25.87
    B22: float = -1.49
    B23: float = 0.0
    A30: float = 1.85e-39

    kB_eV: float = 8.617333262e-5
    kB_J: float = 1.380649e-23

    use_sciantix_D2_zero: bool = True

    D10_vU: float = 1.35e-2
    Q1_vU: float = 5.66
    B21_vU_refit: float = -0.62
    B22_vU_refit: float = -0.04
    A20_vU_fig4_refit: float = 4.6304523933553033e-29

    radius_in_lattice: float = 0.21e-9
    omega_fg: float = 8.5e-29
    lattice_parameter: float = 4.889e-10
    gamma_b: float = 1.11
    hydrostatic_stress: float = 0.0
    min_radius_for_pressure: float = 1.0e-15

    f_n: float = 1.0e-6

    rho_d: float = 3.0e13
    K_d: float = 5.0e5
    r_d: float = 3.46e-10
    Z_d: float = 5.0

    R_b: float = 0.0
    N_b: float = 0.0
    R_d: float = 0.0
    N_d: Optional[float] = None

    c0: float = 0.0
    mb0: float = 0.0
    md0: float = 0.0

    nvb0: Optional[float] = None
    nvd0: Optional[float] = None

    diffusivity_sf: float = 1.0
    trapping_sf: float = 1.0
    resolution_sf: float = 1.0

    vacancy_absorption_only: bool = True
    update_bulk_vacancies: bool = True
    min_number_density: float = 0.0
    min_volume: float = 0.0

    bulk_seed_radius_nm: float = 0.0

    def __post_init__(self):
        if self.N_d is None:
            self.N_d = self.K_d * self.rho_d

        if self.target_burnup_percent_fima is not None:
            self.final_time = burnup_percent_to_time(
                self.target_burnup_percent_fima,
                self.fission_rate,
                self.lattice_parameter,
            )


def omega_matrix(p):
    return p.lattice_parameter**3 / 4.0


def uranium_atom_density_from_lattice(lattice_parameter):
    return 4.0 / lattice_parameter**3


def burnup_percent_to_time(burnup_percent_fima, fission_rate, lattice_parameter):
    if fission_rate <= 0.0:
        raise ValueError("fission_rate must be positive to compute time from burnup.")

    burnup_fraction_fima = burnup_percent_fima / 100.0
    uranium_density = uranium_atom_density_from_lattice(lattice_parameter)
    return burnup_fraction_fima * uranium_density / fission_rate


def time_to_burnup_percent(time, fission_rate, lattice_parameter):
    uranium_density = uranium_atom_density_from_lattice(lattice_parameter)
    return 100.0 * fission_rate * time / uranium_density


def sphere_volume(R):
    if R <= 0.0:
        return 0.0
    return (4.0 / 3.0) * math.pi * R**3


def radius_from_volume(V):
    if V <= 0.0:
        return 0.0
    return (3.0 * V / (4.0 * math.pi)) ** (1.0 / 3.0)


def xe_diffusivity_UN(p):
    T = p.temperature
    F = p.fission_rate
    kBT = p.kB_eV * T

    D1 = p.D10 * math.exp(-p.Q1 / kBT)

    if p.use_sciantix_D2_zero:
        D2 = 0.0
    else:
        D2 = p.A20 * math.sqrt(F) * math.exp(
            -p.B21 / kBT
            -p.B22 / (kBT**2)
            -p.B23 / (kBT**3)
        )

    D3 = p.A30 * F
    Dg = (D1 + D2 + D3) * p.diffusivity_sf * p.precursor_factor
    return Dg, {"D1": D1, "D2": D2, "D3": D3, "Dg": Dg}


def vacancy_diffusivity_UN(p):
    T = p.temperature
    F = p.fission_rate
    kBT = p.kB_eV * T

    D1 = p.D10_vU * math.exp(-p.Q1_vU / kBT)
    D2 = math.sqrt(F) * p.A20_vU_fig4_refit * math.exp(
        p.B21_vU_refit / kBT
        + p.B22_vU_refit / (kBT**2)
    )

    Dv = D1 + D2
    return Dv, {"Dv1": D1, "Dv2": D2, "Dv": Dv}


def b0_resolution(R):
    R = max(R, 1.0e-15)
    return 1.0e-25 * (2.64 - 2.02 * math.exp(-2.61e-9 / R))


def resolution_rates_UN(p, R_b, R_d):
    Rb_eff = R_b + p.radius_in_lattice
    Rd_eff = R_d + p.radius_in_lattice

    b_b = p.fission_rate * b0_resolution(Rb_eff) * p.resolution_sf
    b_d = p.fission_rate * b0_resolution(Rd_eff) * p.resolution_sf
    return b_b, b_d


def trapping_rates_UN(p, Dg, R_b, N_b, R_d, N_d):
    Rb_eff = R_b + p.radius_in_lattice
    Rd_eff = R_d + p.radius_in_lattice

    if N_b <= 0.0:
        g_b = 0.0
    else:
        g_b = 4.0 * math.pi * Dg * Rb_eff * N_b

    Gamma_d = 1.0 / math.sqrt(math.pi * p.rho_d)
    den = math.log(Gamma_d / (p.Z_d * p.r_d)) - 3.0 / 5.0

    if den <= 0.0:
        raise ValueError(f"Invalid dislocation sink denominator: {den:g}")

    free_dislocation = max(p.rho_d - 2.0 * R_d * N_d, 0.0)
    term_bubbles = 4.0 * math.pi * Dg * Rd_eff * N_d
    term_dislocation = (2.0 * math.pi * Dg / den) * free_dislocation
    g_d = term_bubbles + term_dislocation

    return p.trapping_sf * g_b, p.trapping_sf * g_d, {
        "Gamma_d": Gamma_d,
        "den": den,
        "free_dislocation": free_dislocation,
        "term_bubbles": term_bubbles,
        "term_dislocation": term_dislocation,
    }


def beta_production(p):
    return p.xe_yield * p.fission_rate


def nucleation_rate_bulk(p, Dg, c):
    return 8.0 * math.pi * p.f_n * Dg * p.omega_fg ** (1.0 / 3.0) * max(c, 0.0) ** 2


def phi_bulk(mb, Nb):
    if Nb <= 0.0:
        return 0.0

    atoms_per_bubble = mb / Nb
    if atoms_per_bubble <= 1.0:
        return 0.0

    return 1.0 / (atoms_per_bubble - 1.0)


def coalescence_lambda(Vd, Nd):
    xi = Vd * Nd
    xi = max(0.0, min(xi, 0.999999))
    return (2.0 - xi) / (2.0 * (1.0 - xi) ** 3)


def pressure_internal(p, m_gas, n_vac):
    if m_gas <= 0.0:
        return 0.0
    if n_vac <= 0.0:
        return math.inf

    denominator = n_vac * omega_matrix(p)
    if denominator <= 0.0:
        return math.inf

    return p.kB_J * p.temperature * m_gas / denominator


def pressure_equilibrium(p, R):
    R_eff = max(R, p.min_radius_for_pressure)
    return 2.0 * p.gamma_b / R_eff - p.hydrostatic_stress


def gas_only_radius_for_population(p, m_gas, N):
    if m_gas <= 0.0 or N <= 0.0:
        return 0.0
    V_gas_per_bubble = p.omega_fg * m_gas / N
    return radius_from_volume(V_gas_per_bubble)


def radius_for_vacancy_update(p, R_old, N, m_gas):
    if R_old > 0.0:
        return R_old
    return gas_only_radius_for_population(p, m_gas, N)


def pressure_equilibrium_for_vacancy_update(p, R_update):
    if R_update <= 0.0:
        return math.inf
    return 2.0 * p.gamma_b / R_update - p.hydrostatic_stress


def wigner_seitz_delta(N):
    return (3.0 / (4.0 * math.pi * max(N, 1.0))) ** (1.0 / 3.0)


def zeta_geometry(R, N):
    delta = wigner_seitz_delta(N)
    psi = max(R / delta, 1.0e-12)
    den = -psi**6 + 5.0 * psi**2 - 9.0 * psi + 5.0

    if den <= 1.0e-30:
        den = 1.0e-30

    return max(10.0 * psi * (1.0 + psi**3) / den, 1.0e-30)


def vacancy_concentration_implicit_step(p, Dv, R, N, m_gas, n_old, dt):
    if N <= 0.0 or m_gas <= 0.0:
        return n_old, 0.0

    R_update = radius_for_vacancy_update(p, R, N, m_gas)
    if R_update <= 0.0:
        return n_old, 0.0

    p_eq = pressure_equilibrium_for_vacancy_update(p, R_update)
    p_int_old = pressure_internal(p, m_gas, n_old)

    if p.vacancy_absorption_only and p_int_old <= p_eq:
        return n_old, 0.0

    delta = wigner_seitz_delta(N)
    zeta = zeta_geometry(R_update, N)

    A = 2.0 * math.pi * Dv * delta * N / (p.kB_J * p.temperature * zeta)
    C = p.kB_J * p.temperature * m_gas / omega_matrix(p)

    B = n_old - dt * A * p_eq
    discriminant = B * B + 4.0 * dt * A * C

    if discriminant < 0.0:
        raise ValueError(f"Negative discriminant in vacancy implicit step: {discriminant:g}")

    sqrt_disc = math.sqrt(discriminant)

    if B >= 0.0:
        n_new = 0.5 * (B + sqrt_disc)
    else:
        denominator = sqrt_disc - B
        if denominator <= 0.0:
            n_new = 0.0
        else:
            n_new = (2.0 * dt * A * C) / denominator

    if p.vacancy_absorption_only:
        n_new = max(n_new, n_old)

    dn_dt = (n_new - n_old) / dt
    return n_new, dn_dt


def initialize_vacancy_concentration(p, N, R, m_gas):
    if N <= 0.0 or R <= 0.0:
        return 0.0

    total_bubble_volume = N * sphere_volume(R)
    gas_volume = p.omega_fg * m_gas
    vacancy_volume = max(total_bubble_volume - gas_volume, 0.0)
    return vacancy_volume / omega_matrix(p)


def initialize_modes_from_average(average, n_modes, n_iter=20):
    modes = [0.0 for _ in range(n_modes)]
    projection_coeff = -math.sqrt(8.0 / math.pi)
    remainder = average

    for _ in range(n_iter):
        reconstructed = 0.0
        for i in range(n_modes):
            n = i + 1
            n_coeff = (-1.0) ** n / n
            modes[i] += projection_coeff * n_coeff * remainder
            reconstructed += projection_coeff * n_coeff * modes[i] * 3.0 / (4.0 * math.pi)
        remainder = average - reconstructed

    return modes


def reconstruct_average(modes):
    projection_coeff = -2.0 * math.sqrt(2.0 / math.pi)
    average = 0.0

    for i, value in enumerate(modes):
        n = i + 1
        n_coeff = (-1.0) ** n / n
        average += projection_coeff * n_coeff * value / ((4.0 / 3.0) * math.pi)

    return average


def det3(A):
    return (
        A[0][0] * (A[1][1] * A[2][2] - A[1][2] * A[2][1])
        - A[0][1] * (A[1][0] * A[2][2] - A[1][2] * A[2][0])
        + A[0][2] * (A[1][0] * A[2][1] - A[1][1] * A[2][0])
    )


def solve3x3_cramer(A, b):
    detA = det3(A)
    if abs(detA) < 1.0e-300:
        raise ZeroDivisionError("Singular 3x3 system in spectral mode solve")

    Ax = [[b[i], A[i][1], A[i][2]] for i in range(3)]
    Ay = [[A[i][0], b[i], A[i][2]] for i in range(3)]
    Az = [[A[i][0], A[i][1], b[i]] for i in range(3)]

    return [det3(Ax) / detA, det3(Ay) / detA, det3(Az) / detA]


def sciantix_3x3_exchange_step(modes_c, modes_mb, modes_md, Dg, R, beta, g_b, g_d, b_b, b_d, dt):
    projection_coeff = -2.0 * math.sqrt(2.0 / math.pi)
    diffusion_rate_coeff = math.pi**2 * Dg / R**2

    for i in range(len(modes_c)):
        n = i + 1
        n_coeff = (-1.0) ** n / n

        diffusion_rate = diffusion_rate_coeff * n**2
        source_rate = projection_coeff * beta * n_coeff

        A = [
            [1.0 + (diffusion_rate + g_b + g_d) * dt, -b_b * dt, -b_d * dt],
            [-g_b * dt, 1.0 + b_b * dt, 0.0],
            [-g_d * dt, 0.0, 1.0 + b_d * dt],
        ]
        rhs = [modes_c[i] + source_rate * dt, modes_mb[i], modes_md[i]]
        modes_c[i], modes_mb[i], modes_md[i] = solve3x3_cramer(A, rhs)

    return reconstruct_average(modes_c), reconstruct_average(modes_mb), reconstruct_average(modes_md)

# ============================================================
# SOLVER PRINCIPALE UNICO
# ============================================================

def solve_UN_sciantix_intragranular_extended(p):
    modes_c = initialize_modes_from_average(p.c0, p.n_modes)
    modes_mb = initialize_modes_from_average(p.mb0, p.n_modes)
    modes_md = initialize_modes_from_average(p.md0, p.n_modes)

    R_b = p.R_b
    R_d = p.R_d
    N_b = p.N_b
    N_d = p.N_d

    V_b = sphere_volume(R_b)
    V_d = sphere_volume(R_d)

    if p.nvb0 is None:
        nvb = initialize_vacancy_concentration(p, N_b, R_b, p.mb0)
    else:
        nvb = p.nvb0

    if p.nvd0 is None:
        nvd = initialize_vacancy_concentration(p, N_d, R_d, p.md0)
    else:
        nvd = p.nvd0

    beta = beta_production(p)
    initial_gas = p.c0 + p.mb0 + p.md0
    generated = 0.0
    q_gb = 0.0
    retained = initial_gas
    seed_vacancies_added_total = 0.0

    hist = {
        "time": [0.0],
        "burnup_percent_fima": [time_to_burnup_percent(0.0, p.fission_rate, p.lattice_parameter)],
        "c": [reconstruct_average(modes_c)],
        "mb": [reconstruct_average(modes_mb)],
        "md": [reconstruct_average(modes_md)],
        "Nb": [N_b],
        "Nd": [N_d],
        "Vb": [V_b],
        "Vd": [V_d],
        "Rb": [R_b],
        "Rd": [R_d],
        "nvb": [nvb],
        "nvd": [nvd],
        "generated": [generated],
        "retained": [retained],
        "q_gb": [q_gb],
        "qdot_gb": [0.0],
        "swelling_b": [N_b * V_b],
        "swelling_d": [N_d * V_d],
        "swelling_ig": [N_b * V_b + N_d * V_d],
        "p_b": [pressure_internal(p, p.mb0, nvb)],
        "p_d": [pressure_internal(p, p.md0, nvd)],
        "p_b_eq": [pressure_equilibrium(p, R_b)],
        "p_d_eq": [pressure_equilibrium(p, R_d)],
        "lambda_d": [coalescence_lambda(V_d, N_d)],
        "nu_b": [0.0],
        "phi_b": [0.0],
        "dnvb_dt": [0.0],
        "dnvd_dt": [0.0],
        "dVd_growth_dt": [0.0],
        "seed_vacancies_added_total": [0.0],
    }

    last_rates = {}
    n_steps = int(math.ceil(p.final_time / p.dt))

    for _ in range(n_steps):
        t_old = hist["time"][-1]
        dt = min(p.dt, p.final_time - t_old)
        if dt <= 0.0:
            break

        c_old = hist["c"][-1]
        mb_old = hist["mb"][-1]
        md_old = hist["md"][-1]

        Nb_old = N_b
        Nd_old = N_d
        Vd_old = V_d

        Dg, D_parts = xe_diffusivity_UN(p)
        Dv, Dv_parts = vacancy_diffusivity_UN(p)
        b_b, b_d = resolution_rates_UN(p, R_b, R_d)

        g_b_old, g_d, trapping_parts_old = trapping_rates_UN(p, Dg, R_b, Nb_old, R_d, Nd_old)

        # 1. Update N_b prima del gas solver.
        nu_b = nucleation_rate_bulk(p, Dg, c_old)
        phi_b = phi_bulk(mb_old, Nb_old)
        N_b = (Nb_old + dt * nu_b) / (1.0 + dt * b_b * phi_b)
        N_b = max(N_b, p.min_number_density)

        # 2. Seed opzionale per nuove bulk bubbles.
        dN_b_positive = max(N_b - Nb_old, 0.0)
        seed_vacancies_added_step = 0.0
        if dN_b_positive > 0.0 and p.bulk_seed_radius_nm > 0.0:
            V_seed = sphere_volume(p.bulk_seed_radius_nm * 1.0e-9)
            seed_vacancies_added_step = dN_b_positive * V_seed / omega_matrix(p)
            nvb += seed_vacancies_added_step
            seed_vacancies_added_total += seed_vacancies_added_step

        # 3. Correzione importante: ricalcola g_b con N_b aggiornato.
        g_b, _, trapping_parts_updated = trapping_rates_UN(p, Dg, R_b, N_b, R_d, Nd_old)
        trapping_parts = dict(trapping_parts_updated)
        trapping_parts["g_b_old_before_Nb_update"] = g_b_old
        trapping_parts["g_b_used_after_Nb_update"] = g_b
        trapping_parts["Nb_old"] = Nb_old
        trapping_parts["Nb_used_for_gb"] = N_b

        # 4. Gas solver.
        c_new, mb_new, md_new = sciantix_3x3_exchange_step(
            modes_c, modes_mb, modes_md,
            Dg, p.grain_radius, beta,
            g_b, g_d, b_b, b_d, dt,
        )

        dmb_dt = (mb_new - mb_old) / dt
        dmd_dt = (md_new - md_old) / dt

        # 5. Vacancy update.
        if p.update_bulk_vacancies:
            nvb, dnvb_dt = vacancy_concentration_implicit_step(p, Dv, R_b, N_b, mb_new, nvb, dt)
        else:
            dnvb_dt = 0.0

        nvd, dnvd_dt = vacancy_concentration_implicit_step(p, Dv, R_d, Nd_old, md_new, nvd, dt)

        # 6. Volume growth prima della coalescenza.
        if N_b > 0.0:
            V_b_growth = V_b + dt * (
                p.omega_fg / N_b * dmb_dt + omega_matrix(p) / N_b * dnvb_dt
            )
            V_b_growth = max(V_b_growth, p.min_volume)
        else:
            V_b_growth = V_b

        if Nd_old > 0.0:
            dVd_growth_dt = p.omega_fg / Nd_old * dmd_dt + omega_matrix(p) / Nd_old * dnvd_dt
            V_d_growth = V_d + dt * dVd_growth_dt
            V_d_growth = max(V_d_growth, p.min_volume)
        else:
            dVd_growth_dt = 0.0
            V_d_growth = V_d

        # 7. Coalescenza dislocation bubbles.
        lambda_d = coalescence_lambda(Vd_old, Nd_old)
        dVd_positive = max(V_d_growth - Vd_old, 0.0)
        if dVd_positive > 0.0 and Nd_old > 0.0:
            N_d = Nd_old / (1.0 + 4.0 * lambda_d * Nd_old * dVd_positive)
        else:
            N_d = Nd_old
        N_d = max(N_d, p.min_number_density)

        # 8. Ricostruzione volumi da gas + vacancies.
        if N_b > 0.0:
            V_b = (p.omega_fg * max(mb_new, 0.0) + omega_matrix(p) * nvb) / N_b
        else:
            V_b = 0.0

        if N_d > 0.0:
            V_d = (p.omega_fg * max(md_new, 0.0) + omega_matrix(p) * nvd) / N_d
        else:
            V_d = 0.0

        V_b = max(V_b, p.min_volume)
        V_d = max(V_d, p.min_volume)
        R_b = radius_from_volume(V_b)
        R_d = radius_from_volume(V_d)

        # 9. Bilancio gas a grain face.
        generated += beta * dt
        retained = c_new + mb_new + md_new
        q_gb_old = q_gb
        q_gb = max(initial_gas + generated - retained, 0.0)
        qdot_gb = max((q_gb - q_gb_old) / dt, 0.0)

        hist["time"].append(t_old + dt)
        hist["burnup_percent_fima"].append(time_to_burnup_percent(t_old + dt, p.fission_rate, p.lattice_parameter))
        hist["c"].append(c_new)
        hist["mb"].append(mb_new)
        hist["md"].append(md_new)
        hist["Nb"].append(N_b)
        hist["Nd"].append(N_d)
        hist["Vb"].append(V_b)
        hist["Vd"].append(V_d)
        hist["Rb"].append(R_b)
        hist["Rd"].append(R_d)
        hist["nvb"].append(nvb)
        hist["nvd"].append(nvd)
        hist["generated"].append(generated)
        hist["retained"].append(retained)
        hist["q_gb"].append(q_gb)
        hist["qdot_gb"].append(qdot_gb)
        hist["swelling_b"].append(N_b * V_b)
        hist["swelling_d"].append(N_d * V_d)
        hist["swelling_ig"].append(N_b * V_b + N_d * V_d)
        hist["p_b"].append(pressure_internal(p, mb_new, nvb))
        hist["p_d"].append(pressure_internal(p, md_new, nvd))
        hist["p_b_eq"].append(pressure_equilibrium(p, R_b))
        hist["p_d_eq"].append(pressure_equilibrium(p, R_d))
        hist["lambda_d"].append(lambda_d)
        hist["nu_b"].append(nu_b)
        hist["phi_b"].append(phi_b)
        hist["dnvb_dt"].append(dnvb_dt)
        hist["dnvd_dt"].append(dnvd_dt)
        hist["dVd_growth_dt"].append(dVd_growth_dt)
        hist["seed_vacancies_added_total"].append(seed_vacancies_added_total)

        last_rates = {
            "Dg": Dg,
            "Dv": Dv,
            "beta": beta,
            "g_b": g_b,
            "g_b_old_before_Nb_update": g_b_old,
            "g_d": g_d,
            "b_b": b_b,
            "b_d": b_d,
            "nu_b": nu_b,
            "phi_b": phi_b,
            "lambda_d": lambda_d,
            "dVd_growth_dt": dVd_growth_dt,
            "dnvb_dt": dnvb_dt,
            "dnvd_dt": dnvd_dt,
            "seed_vacancies_added_step": seed_vacancies_added_step,
            "seed_vacancies_added_total": seed_vacancies_added_total,
            "p_b": pressure_internal(p, mb_new, nvb),
            "p_d": pressure_internal(p, md_new, nvd),
            "p_b_eq": pressure_equilibrium(p, R_b),
            "p_d_eq": pressure_equilibrium(p, R_d),
            "Nb_old": Nb_old,
            "Nb_updated_before_gas": N_b,
            "Nd_old": Nd_old,
            "Nd_after_coalescence": N_d,
            **D_parts,
            **Dv_parts,
            **trapping_parts,
        }

    return hist, last_rates

# ============================================================
# WRAPPER E SCORE
# ============================================================

_RUN_CACHE = {}


def run_model_point(T, burnup, f_n, bulk_seed_radius_nm=BULK_SEED_RADIUS_NM, dt_h=DT_SCAN_H, n_modes=N_MODES_SCAN):
    key = (
        round(float(T), 6),
        round(float(burnup), 6),
        float(f_n),
        float(bulk_seed_radius_nm),
        float(dt_h),
        int(n_modes),
    )
    if key in _RUN_CACHE:
        return _RUN_CACHE[key]

    p = UNParameters(
        temperature=float(T),
        fission_rate=FISSION_RATE,
        grain_radius=GRAIN_RADIUS,
        target_burnup_percent_fima=float(burnup),
        dt=float(dt_h) * 3600.0,
        n_modes=int(n_modes),
        xe_yield=XE_YIELD,
        f_n=float(f_n),
        K_d=K_D_FIXED,
        R_b=0.0,
        N_b=0.0,
        R_d=0.0,
        N_d=None,
        c0=0.0,
        mb0=0.0,
        md0=0.0,
        vacancy_absorption_only=True,
        update_bulk_vacancies=True,
        bulk_seed_radius_nm=float(bulk_seed_radius_nm),
    )

    hist, rates = solve_UN_sciantix_intragranular_extended(p)

    generated = hist["generated"][-1]
    if generated > 0.0:
        matrix_gas_percent = 100.0 * hist["c"][-1] / generated
        bulk_gas_percent = 100.0 * hist["mb"][-1] / generated
        dislocation_gas_percent = 100.0 * hist["md"][-1] / generated
        qgb_gas_percent = 100.0 * hist["q_gb"][-1] / generated
    else:
        matrix_gas_percent = bulk_gas_percent = dislocation_gas_percent = qgb_gas_percent = 0.0

    p_b_eq = hist["p_b_eq"][-1]
    p_d_eq = hist["p_d_eq"][-1]

    row = {
        "T": float(T),
        "burnup": float(burnup),
        "f_n": float(f_n),
        "bulk_seed_radius_nm": float(bulk_seed_radius_nm),
        "swelling_b_percent": 100.0 * hist["swelling_b"][-1],
        "swelling_d_percent": 100.0 * hist["swelling_d"][-1],
        "swelling_ig_percent": 100.0 * hist["swelling_ig"][-1],
        "Nb": hist["Nb"][-1],
        "Nd": hist["Nd"][-1],
        "Rb_nm": hist["Rb"][-1] * 1.0e9,
        "Rd_nm": hist["Rd"][-1] * 1.0e9,
        "p_b": hist["p_b"][-1],
        "p_b_eq": p_b_eq,
        "p_d": hist["p_d"][-1],
        "p_d_eq": p_d_eq,
        "p_b_over_eq": hist["p_b"][-1] / p_b_eq if p_b_eq > 0.0 else math.nan,
        "p_d_over_eq": hist["p_d"][-1] / p_d_eq if p_d_eq > 0.0 else math.nan,
        "matrix_gas_percent": matrix_gas_percent,
        "bulk_gas_percent": bulk_gas_percent,
        "dislocation_gas_percent": dislocation_gas_percent,
        "qgb_gas_percent": qgb_gas_percent,
        "Dg": rates.get("Dg", math.nan),
        "Dv": rates.get("Dv", math.nan),
        "g_b": rates.get("g_b", math.nan),
        "g_d": rates.get("g_d", math.nan),
        "hist": hist,
        "rates": rates,
    }

    _RUN_CACHE[key] = row
    return row


def rmse(values):
    values = [v for v in values if math.isfinite(v)]
    if not values:
        return math.nan
    return math.sqrt(sum(v * v for v in values) / len(values))


def log10_rmse(pairs):
    errs = []
    for pred, exp in pairs:
        if pred > 0.0 and exp > 0.0 and math.isfinite(pred) and math.isfinite(exp):
            errs.append(math.log10(pred / exp))
    return rmse(errs)


def score_combo(f_n):
    swelling_errors = []
    for exp in EXP_SWELLING_T:
        out = run_model_point(exp["T"], exp["burnup"], f_n)
        swelling_errors.append(out["swelling_d_percent"] - exp["swelling"])

    score_swelling = rmse(swelling_errors)

    n_pairs = []
    for exp in EXP_ND_T_13:
        out = run_model_point(exp["T"], 1.3, f_n)
        n_pairs.append((out["Nd"], exp["N"]))

    r_pairs = []
    for exp in EXP_RD_T_13:
        out = run_model_point(exp["T"], 1.3, f_n)
        r_pairs.append((out["Rd_nm"], exp["R_nm"]))

    score_log_N = log10_rmse(n_pairs)
    score_log_R = log10_rmse(r_pairs)

    # Penalità leggera per pressioni troppo lontane da equilibrio a bassa T.
    p_low = run_model_point(984.0, 3.2, f_n)
    pressure_penalty = 0.0
    if p_low["p_b_over_eq"] > 0.0 and p_low["p_d_over_eq"] > 0.0:
        pressure_penalty = 0.15 * (
            abs(math.log10(max(p_low["p_b_over_eq"], 1.0e-30)))
            + abs(math.log10(max(p_low["p_d_over_eq"], 1.0e-30)))
        )

    total_score = score_swelling + score_log_N + score_log_R + pressure_penalty

    return {
        "f_n": f_n,
        "score_total": total_score,
        "score_swelling_rmse_percent": score_swelling,
        "score_log10_N": score_log_N,
        "score_log10_R": score_log_R,
        "pressure_penalty": pressure_penalty,
    }


def print_ranking(rows):
    rows_sorted = sorted(rows, key=lambda r: r["score_total"])
    print("\n" + "=" * 130)
    print("RANKING f_n ONLY, K_d FIXED")
    print("=" * 130)
    print(
        f"{'rank':>4s} | {'score':>12s} | {'RMSE_sw_%':>12s} | "
        f"{'logRMSE_N':>12s} | {'logRMSE_R':>12s} | {'p_penalty':>12s} | {'f_n':>12s}"
    )
    print("-" * 130)

    for i, r in enumerate(rows_sorted, start=1):
        print(
            f"{i:4d} | {r['score_total']:12.5e} | "
            f"{r['score_swelling_rmse_percent']:12.5e} | "
            f"{r['score_log10_N']:12.5e} | "
            f"{r['score_log10_R']:12.5e} | "
            f"{r['pressure_penalty']:12.5e} | "
            f"{r['f_n']:12.3e}"
        )

    return rows_sorted

# ============================================================
# OUTPUT E GRAFICI
# ============================================================

def print_key_table(title, rows):
    print("\n" + "=" * 170)
    print(title)
    print("=" * 170)
    print(
        f"{'T_K':>8s} | {'sw_b_%':>10s} | {'sw_d_%':>10s} | {'sw_ig_%':>10s} | "
        f"{'Rb_nm':>10s} | {'Rd_nm':>10s} | {'Nb':>12s} | {'Nd':>12s} | "
        f"{'pb/peq':>10s} | {'pd/peq':>10s} | "
        f"{'matrix_%':>10s} | {'bulkGas_%':>10s} | {'dislGas_%':>10s} | {'qgb_%':>10s}"
    )
    print("-" * 170)

    wanted = {900.0, 1000.0, 1200.0, 1400.0, 1500.0, 1600.0, 1700.0, 1800.0, 2000.0}
    for r in rows:
        if r["T"] in wanted:
            print(
                f"{r['T']:8.1f} | "
                f"{r['swelling_b_percent']:10.4e} | {r['swelling_d_percent']:10.4e} | {r['swelling_ig_percent']:10.4e} | "
                f"{r['Rb_nm']:10.4e} | {r['Rd_nm']:10.4e} | "
                f"{r['Nb']:12.5e} | {r['Nd']:12.5e} | "
                f"{r['p_b_over_eq']:10.4e} | {r['p_d_over_eq']:10.4e} | "
                f"{r['matrix_gas_percent']:10.4e} | {r['bulk_gas_percent']:10.4e} | "
                f"{r['dislocation_gas_percent']:10.4e} | {r['qgb_gas_percent']:10.4e}"
            )


def plot_exp_swelling_for_burnup(bu):
    series_names = sorted({p["series"] for p in EXP_SWELLING_T if abs(p["burnup"] - bu) < 1.0e-9})
    for series in series_names:
        pts = [p for p in EXP_SWELLING_T if abs(p["burnup"] - bu) < 1.0e-9 and p["series"] == series]
        marker = "x" if "119" in series else "^"
        plt.scatter([p["T"] for p in pts], [p["swelling"] for p in pts], marker=marker, s=55, label=f"Exp {series}")


def make_diagnostics(best_fn):
    T_PLOT = [float(T) for T in range(900, 2001, 50)]
    T_AMOUNT = [float(T) for T in range(900, 2601, 100)]
    BURNUPS = [1.1, 1.3, 3.2]

    results_by_burnup = {}
    for bu in BURNUPS:
        results_by_burnup[bu] = [run_model_point(T, bu, best_fn) for T in T_PLOT]
        print_key_table(f"DETAILED TABLE - f_n={best_fn:.3e}, burnup={bu:.1f}% FIMA", results_by_burnup[bu])

    gas_partition = {}
    for bu in [1.1, 3.2]:
        gas_partition[bu] = [run_model_point(T, bu, best_fn) for T in T_AMOUNT]

    if plt is None:
        print("\nmatplotlib non installato: grafici saltati.")
        return

    for bu in BURNUPS:
        rows = results_by_burnup[bu]
        Ts = [r["T"] for r in rows]
        plt.figure(figsize=(8, 5))
        plt.plot(Ts, [r["swelling_d_percent"] for r in rows], label="Model dislocation")
        plt.plot(Ts, [r["swelling_b_percent"] for r in rows], linestyle=":", label="Model bulk")
        plot_exp_swelling_for_burnup(bu)
        plt.xlabel("T [K]")
        plt.ylabel("Fission gas swelling [%]")
        plt.title(f"Swelling vs T at {bu:.1f}% FIMA; f_n={best_fn:.2e}, K_d={K_D_FIXED:.1e}")
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()

    burnup_grid = [0.2, 0.5, 0.8, 1.1, 1.3, 1.6, 2.0, 2.5, 3.2, 4.0, 5.0, 6.0]
    rows_bu = [run_model_point(1600.0, bu, best_fn) for bu in burnup_grid]
    plt.figure(figsize=(8, 5))
    plt.plot(burnup_grid, [r["swelling_d_percent"] for r in rows_bu], label="Model dislocation")
    plt.plot(burnup_grid, [r["swelling_b_percent"] for r in rows_bu], linestyle=":", label="Model bulk")
    plt.scatter([p["burnup"] for p in EXP_SWELLING_BURNUP_1600], [p["swelling"] for p in EXP_SWELLING_BURNUP_1600], marker="^", s=60, label="Exp approx")
    plt.xlabel("Burnup [% FIMA]")
    plt.ylabel("Fission gas swelling [%]")
    plt.title("Swelling vs burnup at 1600 K")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

    rows = results_by_burnup[1.3]
    Ts = [r["T"] for r in rows]
    plt.figure(figsize=(8, 5))
    plt.plot(Ts, [r["Nd"] for r in rows], label="Model dislocation N_d")
    plt.plot(Ts, [r["Nb"] for r in rows], linestyle=":", label="Model bulk N_b")
    plt.scatter([p["T"] for p in EXP_ND_T_13], [p["N"] for p in EXP_ND_T_13], marker="^", s=55, label="Exp large bubbles")
    plt.yscale("log")
    plt.xlabel("T [K]")
    plt.ylabel("Bubble concentration [m$^{-3}$]")
    plt.title("Bubble concentration at 1.3% FIMA")
    plt.grid(True, which="both", alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(8, 5))
    plt.plot(Ts, [r["Rd_nm"] for r in rows], label="Model dislocation R_d")
    plt.plot(Ts, [r["Rb_nm"] for r in rows], linestyle=":", label="Model bulk R_b")
    plt.scatter([p["T"] for p in EXP_RD_T_13], [p["R_nm"] for p in EXP_RD_T_13], marker="^", s=55, label="Exp large bubbles")
    plt.yscale("log")
    plt.xlabel("T [K]")
    plt.ylabel("Bubble radius [nm]")
    plt.title("Bubble radius at 1.3% FIMA")
    plt.grid(True, which="both", alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

    rows = results_by_burnup[3.2]
    Ts = [r["T"] for r in rows]
    plt.figure(figsize=(8, 5))
    plt.plot(Ts, [r["p_b"] for r in rows], label="Bulk bubble pressure")
    plt.plot(Ts, [r["p_b_eq"] for r in rows], linestyle=":", label="Bulk equilibrium pressure")
    plt.plot(Ts, [r["p_d"] for r in rows], label="Dislocation bubble pressure")
    plt.plot(Ts, [r["p_d_eq"] for r in rows], linestyle="--", label="Dislocation equilibrium pressure")
    plt.yscale("log")
    plt.xlabel("T [K]")
    plt.ylabel("Bubble pressure [Pa]")
    plt.title("Pressure diagnostic at 3.2% FIMA")
    plt.grid(True, which="both", alpha=0.3)
    plt.legend(fontsize=8)
    plt.tight_layout()
    plt.show()

    for bu, rows in gas_partition.items():
        Ts = [r["T"] for r in rows]
        plt.figure(figsize=(8, 5))
        plt.plot(Ts, [r["matrix_gas_percent"] for r in rows], label="Matrix")
        plt.plot(Ts, [r["bulk_gas_percent"] for r in rows], linestyle=":", label="Bulk bubbles")
        plt.plot(Ts, [r["dislocation_gas_percent"] for r in rows], linestyle="--", label="Dislocation bubbles")
        plt.plot(Ts, [r["qgb_gas_percent"] for r in rows], linestyle="-.", label="Gas to grain face q_gb")
        plt.xlabel("T [K]")
        plt.ylabel("Amount of generated gas [%]")
        plt.title(f"Gas partition diagnostic at {bu:.1f}% FIMA")
        plt.xlim(900, 2600)
        plt.ylim(0, 100)
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()

# ============================================================
# MAIN
# ============================================================

if RUN_SINGLE_TEST:
    test = run_model_point(1600.0, 1.3, 3.0e-9)
    print(test)

if RUN_CALIBRATION:
    print("\n" + "#" * 130)
    print("START CLEAN UN MODEL + f_n CALIBRATION")
    print("#" * 130)
    print(f"K_d fixed             = {K_D_FIXED:.6e} bub/m")
    print(f"rho_d default         = 3.0e13 m^-2")
    print(f"N_d0                  = {K_D_FIXED * 3.0e13:.6e} bub/m3")
    print(f"Xe yield              = {XE_YIELD:.6e}")
    print(f"dt scan               = {DT_SCAN_H:.2f} h")
    print(f"n_modes               = {N_MODES_SCAN:d}")
    print(f"bulk seed radius      = {BULK_SEED_RADIUS_NM:.3f} nm")
    print("Target principale: experimental large/P2 bubbles -> model dislocation bubbles.")

    ranking = []
    for f_n in FN_GRID:
        print(f"Running f_n = {f_n:.3e}")
        ranking.append(score_combo(f_n))

    ranking_sorted = print_ranking(ranking)
    best_fn = ranking_sorted[0]["f_n"]

    print("\n" + "=" * 130)
    print("SELECTED PARAMETER SET")
    print("=" * 130)
    print(f"best f_n              = {best_fn:.6e}")
    print(f"K_d                   = {K_D_FIXED:.6e} bub/m")
    print(f"bulk seed radius      = {BULK_SEED_RADIUS_NM:.3f} nm")
    print("=" * 130)

    make_diagnostics(best_fn)

    print("\n" + "#" * 130)
    print("END CLEAN UN MODEL + f_n CALIBRATION")
    print("#" * 130)
````

````python
import math
from dataclasses import dataclass
from typing import Optional

try:
    import matplotlib.pyplot as plt
except ModuleNotFoundError:
    plt = None

# ============================================================
# CONFIGURAZIONE GENERALE
# ============================================================

FISSION_RATE = 5.0e19
GRAIN_RADIUS = 5.0e-6
K_D_FIXED = 5.0e5
XE_YIELD = 0.24

DT_SCAN_H = 12.0
N_MODES_SCAN = 25

# 0.0 = nessun seed artificiale. Metti 2.0 solo per diagnostica pressione bulk.
BULK_SEED_RADIUS_NM = 0.0

# Modalità: FAST per il primo giro; FULL per una griglia più densa.
RUN_MODE = "FAST"

FN_GRID = [
    3.0e-10,
    5.0e-10,
    7.0e-10,
    1.0e-9,
    1.5e-9,
    2.0e-9,
    3.0e-9,
    5.0e-9,
    7.0e-9,
    1.0e-8,
    3.0e-8,
    1.0e-7,
    3.0e-7,
    1.0e-6,
]

if RUN_MODE.upper() == "FULL":
    FN_GRID_B = [3.0e-9, 1.0e-8, 3.0e-8, 1.0e-7, 3.0e-7, 1.0e-6]
    GD_SCALE_GRID = [0.5, 1.0, 2.0, 3.0, 5.0, 7.0, 10.0, 20.0]
    K_D_GRID = [3.0e5, 5.0e5, 8.0e5, 1.2e6, 2.0e6, 3.0e6]
else:
    FN_GRID_B = [3.0e-9, 1.0e-8, 3.0e-8, 1.0e-7, 3.0e-7, 1.0e-6]
    GD_SCALE_GRID = [1.0, 2.0, 3.0, 5.0, 10.0]
    K_D_GRID = [5.0e5, 1.2e6, 2.0e6]

BULK_SEED_GRID = [0.0, 1.0, 2.0]

# Se vuoi fare solo un test veloce, metti False e usa RUN_SINGLE_TEST=True.
RUN_CALIBRATION = True
RUN_PHASE_A_FN_ONLY = True
RUN_PHASE_B_GRID = True
RUN_PHASE_C_FIXED_FN_GD = True
RUN_PHASE_D_BULK_SEED = True
RUN_SINGLE_TEST = False

# ============================================================
# DATI SPERIMENTALI DIGITALIZZATI DAL MARKDOWN
# ============================================================

EXP_SWELLING_T = [
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1127.0, "swelling": 0.68},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1228.0, "swelling": 0.59},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1312.0, "swelling": 0.43},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1402.0, "swelling": 0.58},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1485.0, "swelling": 1.22},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1549.0, "swelling": 1.84},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1598.0, "swelling": 1.66},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1632.0, "swelling": 2.13},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1669.0, "swelling": 3.60},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1685.0, "swelling": 2.72},

    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 899.0,  "swelling": 0.63},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1154.0, "swelling": 1.17},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1228.0, "swelling": 1.08},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1325.0, "swelling": 1.28},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1435.0, "swelling": 1.32},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1514.0, "swelling": 2.10},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1570.0, "swelling": 2.72},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1608.0, "swelling": 2.91},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1635.0, "swelling": 3.28},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1656.0, "swelling": 3.75},

    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1044.0, "swelling": 1.11},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1220.0, "swelling": 1.22},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1377.0, "swelling": 1.33},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1534.0, "swelling": 2.83},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1595.0, "swelling": 3.53},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1639.0, "swelling": 3.86},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1661.0, "swelling": 2.45},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1709.0, "swelling": 2.93},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1724.0, "swelling": 3.15},

    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 984.0,  "swelling": 0.72},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1056.0, "swelling": 1.06},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1126.0, "swelling": 1.26},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1247.0, "swelling": 1.58},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1343.0, "swelling": 1.79},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1420.0, "swelling": 2.08},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1459.0, "swelling": 2.40},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1511.0, "swelling": 2.83},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1557.0, "swelling": 3.31},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1590.0, "swelling": 3.75},
]

EXP_SWELLING_BURNUP_1600 = [
    {"burnup": 1.12, "swelling": 1.64, "series": "approx 100 kW/m"},
    {"burnup": 1.11, "swelling": 2.90, "series": "approx 119 kW/m"},
    {"burnup": 1.31, "swelling": 3.51, "series": "measurement"},
    {"burnup": 3.18, "swelling": 3.72, "series": "measurement"},
]

EXP_ND_T_13 = [
    {"T": 1153.0, "N": 5.22e19},
    {"T": 1202.0, "N": 3.69e19},
    {"T": 1227.0, "N": 3.00e19},
    {"T": 1235.0, "N": 3.57e19},
    {"T": 1248.0, "N": 2.44e19},
    {"T": 1330.0, "N": 2.90e19},
    {"T": 1338.0, "N": 4.10e19},
    {"T": 1377.0, "N": 2.71e19},
    {"T": 1408.0, "N": 1.67e19},
    {"T": 1427.0, "N": 1.85e19},
    {"T": 1493.0, "N": 3.00e19},
    {"T": 1507.0, "N": 2.44e19},
    {"T": 1524.0, "N": 1.56e19},
    {"T": 1538.0, "N": 7.54e18},
    {"T": 1555.0, "N": 7.80e18},
    {"T": 1561.0, "N": 1.92e19},
    {"T": 1599.0, "N": 5.34e18},
    {"T": 1622.0, "N": 7.80e18},
    {"T": 1628.0, "N": 1.73e19},
    {"T": 1649.0, "N": 5.34e18},
    {"T": 1656.0, "N": 3.65e18},
    {"T": 1656.0, "N": 2.41e18},
    {"T": 1669.0, "N": 1.96e18},
    {"T": 1682.0, "N": 2.44e19},
    {"T": 1685.0, "N": 1.40e19},
    {"T": 1723.0, "N": 7.70e17},
    {"T": 1742.0, "N": 1.02e18},
    {"T": 1739.0, "N": 1.65e18},
]

EXP_RD_T_13 = [
    {"T": 1045.0, "R_nm": 54.06},
    {"T": 1219.0, "R_nm": 60.37},
    {"T": 1374.0, "R_nm": 69.58},
    {"T": 1535.0, "R_nm": 104.85},
    {"T": 1594.0, "R_nm": 120.83},
    {"T": 1641.0, "R_nm": 143.73},
    {"T": 1663.0, "R_nm": 122.76},
    {"T": 1710.0, "R_nm": 157.99},
    {"T": 1725.0, "R_nm": 173.67},
]

# ============================================================
# PARAMETRI E FUNZIONI BASE
# ============================================================

@dataclass
class UNParameters:
    temperature: float = 1600.0
    fission_rate: float = 5.0e19
    grain_radius: float = 5.0e-6
    final_time: float = 24.0 * 3600.0
    target_burnup_percent_fima: Optional[float] = None
    dt: float = 60.0
    n_modes: int = 40

    xe_yield: float = 0.24
    precursor_factor: float = 1.0

    D10: float = 1.56e-3
    Q1: float = 4.94
    A20: float = 1.21e-67
    B21: float = 25.87
    B22: float = -1.49
    B23: float = 0.0
    A30: float = 1.85e-39

    kB_eV: float = 8.617333262e-5
    kB_J: float = 1.380649e-23

    use_sciantix_D2_zero: bool = True

    D10_vU: float = 1.35e-2
    Q1_vU: float = 5.66
    B21_vU_refit: float = -0.62
    B22_vU_refit: float = -0.04
    A20_vU_fig4_refit: float = 4.6304523933553033e-29

    radius_in_lattice: float = 0.21e-9
    omega_fg: float = 8.5e-29
    lattice_parameter: float = 4.889e-10
    gamma_b: float = 1.11
    hydrostatic_stress: float = 0.0
    min_radius_for_pressure: float = 1.0e-15

    f_n: float = 1.0e-6

    rho_d: float = 3.0e13
    K_d: float = 5.0e5
    r_d: float = 3.46e-10
    Z_d: float = 5.0

    R_b: float = 0.0
    N_b: float = 0.0
    R_d: float = 0.0
    N_d: Optional[float] = None

    c0: float = 0.0
    mb0: float = 0.0
    md0: float = 0.0

    nvb0: Optional[float] = None
    nvd0: Optional[float] = None

    diffusivity_sf: float = 1.0
    trapping_sf: float = 1.0
    g_b_scale: float = 1.0
    g_d_scale: float = 1.0
    resolution_sf: float = 1.0

    vacancy_absorption_only: bool = True
    update_bulk_vacancies: bool = True
    min_number_density: float = 0.0
    min_volume: float = 0.0

    bulk_seed_radius_nm: float = 0.0

    def __post_init__(self):
        if self.N_d is None:
            self.N_d = self.K_d * self.rho_d

        if self.target_burnup_percent_fima is not None:
            self.final_time = burnup_percent_to_time(
                self.target_burnup_percent_fima,
                self.fission_rate,
                self.lattice_parameter,
            )


def omega_matrix(p):
    return p.lattice_parameter**3 / 4.0


def uranium_atom_density_from_lattice(lattice_parameter):
    return 4.0 / lattice_parameter**3


def burnup_percent_to_time(burnup_percent_fima, fission_rate, lattice_parameter):
    if fission_rate <= 0.0:
        raise ValueError("fission_rate must be positive to compute time from burnup.")

    burnup_fraction_fima = burnup_percent_fima / 100.0
    uranium_density = uranium_atom_density_from_lattice(lattice_parameter)
    return burnup_fraction_fima * uranium_density / fission_rate


def time_to_burnup_percent(time, fission_rate, lattice_parameter):
    uranium_density = uranium_atom_density_from_lattice(lattice_parameter)
    return 100.0 * fission_rate * time / uranium_density


def sphere_volume(R):
    if R <= 0.0:
        return 0.0
    return (4.0 / 3.0) * math.pi * R**3


def radius_from_volume(V):
    if V <= 0.0:
        return 0.0
    return (3.0 * V / (4.0 * math.pi)) ** (1.0 / 3.0)


def xe_diffusivity_UN(p):
    T = p.temperature
    F = p.fission_rate
    kBT = p.kB_eV * T

    D1 = p.D10 * math.exp(-p.Q1 / kBT)

    if p.use_sciantix_D2_zero:
        D2 = 0.0
    else:
        D2 = p.A20 * math.sqrt(F) * math.exp(
            -p.B21 / kBT
            -p.B22 / (kBT**2)
            -p.B23 / (kBT**3)
        )

    D3 = p.A30 * F
    Dg = (D1 + D2 + D3) * p.diffusivity_sf * p.precursor_factor
    return Dg, {"D1": D1, "D2": D2, "D3": D3, "Dg": Dg}


def vacancy_diffusivity_UN(p):
    T = p.temperature
    F = p.fission_rate
    kBT = p.kB_eV * T

    D1 = p.D10_vU * math.exp(-p.Q1_vU / kBT)
    D2 = math.sqrt(F) * p.A20_vU_fig4_refit * math.exp(
        p.B21_vU_refit / kBT
        + p.B22_vU_refit / (kBT**2)
    )

    Dv = D1 + D2
    return Dv, {"Dv1": D1, "Dv2": D2, "Dv": Dv}


def b0_resolution(R):
    R = max(R, 1.0e-15)
    return 1.0e-25 * (2.64 - 2.02 * math.exp(-2.61e-9 / R))


def resolution_rates_UN(p, R_b, R_d):
    Rb_eff = R_b + p.radius_in_lattice
    Rd_eff = R_d + p.radius_in_lattice

    b_b = p.fission_rate * b0_resolution(Rb_eff) * p.resolution_sf
    b_d = p.fission_rate * b0_resolution(Rd_eff) * p.resolution_sf
    return b_b, b_d


def trapping_rates_UN(p, Dg, R_b, N_b, R_d, N_d):
    Rb_eff = R_b + p.radius_in_lattice
    Rd_eff = R_d + p.radius_in_lattice

    if N_b <= 0.0:
        g_b = 0.0
    else:
        g_b = 4.0 * math.pi * Dg * Rb_eff * N_b

    Gamma_d = 1.0 / math.sqrt(math.pi * p.rho_d)
    den = math.log(Gamma_d / (p.Z_d * p.r_d)) - 3.0 / 5.0

    if den <= 0.0:
        raise ValueError(f"Invalid dislocation sink denominator: {den:g}")

    free_dislocation = max(p.rho_d - 2.0 * R_d * N_d, 0.0)
    term_bubbles = 4.0 * math.pi * Dg * Rd_eff * N_d
    term_dislocation = (2.0 * math.pi * Dg / den) * free_dislocation
    g_d = term_bubbles + term_dislocation

    g_b_scaled = p.trapping_sf * p.g_b_scale * g_b
    g_d_scaled = p.trapping_sf * p.g_d_scale * g_d

    return g_b_scaled, g_d_scaled, {
        "Gamma_d": Gamma_d,
        "den": den,
        "free_dislocation": free_dislocation,
        "term_bubbles": term_bubbles,
        "term_dislocation": term_dislocation,
        "g_b_unscaled": g_b,
        "g_d_unscaled": g_d,
        "g_b_scale": p.g_b_scale,
        "g_d_scale": p.g_d_scale,
    }


def beta_production(p):
    return p.xe_yield * p.fission_rate


def nucleation_rate_bulk(p, Dg, c):
    return 8.0 * math.pi * p.f_n * Dg * p.omega_fg ** (1.0 / 3.0) * max(c, 0.0) ** 2


def phi_bulk(mb, Nb):
    if Nb <= 0.0:
        return 0.0

    atoms_per_bubble = mb / Nb
    if atoms_per_bubble <= 1.0:
        return 0.0

    return 1.0 / (atoms_per_bubble - 1.0)


def coalescence_lambda(Vd, Nd):
    xi = Vd * Nd
    xi = max(0.0, min(xi, 0.999999))
    return (2.0 - xi) / (2.0 * (1.0 - xi) ** 3)


def pressure_internal(p, m_gas, n_vac):
    if m_gas <= 0.0:
        return 0.0
    if n_vac <= 0.0:
        return math.inf

    denominator = n_vac * omega_matrix(p)
    if denominator <= 0.0:
        return math.inf

    return p.kB_J * p.temperature * m_gas / denominator


def pressure_equilibrium(p, R):
    R_eff = max(R, p.min_radius_for_pressure)
    return 2.0 * p.gamma_b / R_eff - p.hydrostatic_stress


def gas_only_radius_for_population(p, m_gas, N):
    if m_gas <= 0.0 or N <= 0.0:
        return 0.0
    V_gas_per_bubble = p.omega_fg * m_gas / N
    return radius_from_volume(V_gas_per_bubble)


def radius_for_vacancy_update(p, R_old, N, m_gas):
    if R_old > 0.0:
        return R_old
    return gas_only_radius_for_population(p, m_gas, N)


def pressure_equilibrium_for_vacancy_update(p, R_update):
    if R_update <= 0.0:
        return math.inf
    return 2.0 * p.gamma_b / R_update - p.hydrostatic_stress


def wigner_seitz_delta(N):
    return (3.0 / (4.0 * math.pi * max(N, 1.0))) ** (1.0 / 3.0)


def zeta_geometry(R, N):
    delta = wigner_seitz_delta(N)
    psi = max(R / delta, 1.0e-12)
    den = -psi**6 + 5.0 * psi**2 - 9.0 * psi + 5.0

    if den <= 1.0e-30:
        den = 1.0e-30

    return max(10.0 * psi * (1.0 + psi**3) / den, 1.0e-30)


def vacancy_concentration_implicit_step(p, Dv, R, N, m_gas, n_old, dt):
    if N <= 0.0 or m_gas <= 0.0:
        return n_old, 0.0

    R_update = radius_for_vacancy_update(p, R, N, m_gas)
    if R_update <= 0.0:
        return n_old, 0.0

    p_eq = pressure_equilibrium_for_vacancy_update(p, R_update)
    p_int_old = pressure_internal(p, m_gas, n_old)

    if p.vacancy_absorption_only and p_int_old <= p_eq:
        return n_old, 0.0

    delta = wigner_seitz_delta(N)
    zeta = zeta_geometry(R_update, N)

    A = 2.0 * math.pi * Dv * delta * N / (p.kB_J * p.temperature * zeta)
    C = p.kB_J * p.temperature * m_gas / omega_matrix(p)

    B = n_old - dt * A * p_eq
    discriminant = B * B + 4.0 * dt * A * C

    if discriminant < 0.0:
        raise ValueError(f"Negative discriminant in vacancy implicit step: {discriminant:g}")

    sqrt_disc = math.sqrt(discriminant)

    if B >= 0.0:
        n_new = 0.5 * (B + sqrt_disc)
    else:
        denominator = sqrt_disc - B
        if denominator <= 0.0:
            n_new = 0.0
        else:
            n_new = (2.0 * dt * A * C) / denominator

    if p.vacancy_absorption_only:
        n_new = max(n_new, n_old)

    dn_dt = (n_new - n_old) / dt
    return n_new, dn_dt


def initialize_vacancy_concentration(p, N, R, m_gas):
    if N <= 0.0 or R <= 0.0:
        return 0.0

    total_bubble_volume = N * sphere_volume(R)
    gas_volume = p.omega_fg * m_gas
    vacancy_volume = max(total_bubble_volume - gas_volume, 0.0)
    return vacancy_volume / omega_matrix(p)


def initialize_modes_from_average(average, n_modes, n_iter=20):
    modes = [0.0 for _ in range(n_modes)]
    projection_coeff = -math.sqrt(8.0 / math.pi)
    remainder = average

    for _ in range(n_iter):
        reconstructed = 0.0
        for i in range(n_modes):
            n = i + 1
            n_coeff = (-1.0) ** n / n
            modes[i] += projection_coeff * n_coeff * remainder
            reconstructed += projection_coeff * n_coeff * modes[i] * 3.0 / (4.0 * math.pi)
        remainder = average - reconstructed

    return modes


def reconstruct_average(modes):
    projection_coeff = -2.0 * math.sqrt(2.0 / math.pi)
    average = 0.0

    for i, value in enumerate(modes):
        n = i + 1
        n_coeff = (-1.0) ** n / n
        average += projection_coeff * n_coeff * value / ((4.0 / 3.0) * math.pi)

    return average


def det3(A):
    return (
        A[0][0] * (A[1][1] * A[2][2] - A[1][2] * A[2][1])
        - A[0][1] * (A[1][0] * A[2][2] - A[1][2] * A[2][0])
        + A[0][2] * (A[1][0] * A[2][1] - A[1][1] * A[2][0])
    )


def solve3x3_cramer(A, b):
    detA = det3(A)
    if abs(detA) < 1.0e-300:
        raise ZeroDivisionError("Singular 3x3 system in spectral mode solve")

    Ax = [[b[i], A[i][1], A[i][2]] for i in range(3)]
    Ay = [[A[i][0], b[i], A[i][2]] for i in range(3)]
    Az = [[A[i][0], A[i][1], b[i]] for i in range(3)]

    return [det3(Ax) / detA, det3(Ay) / detA, det3(Az) / detA]


def sciantix_3x3_exchange_step(modes_c, modes_mb, modes_md, Dg, R, beta, g_b, g_d, b_b, b_d, dt):
    projection_coeff = -2.0 * math.sqrt(2.0 / math.pi)
    diffusion_rate_coeff = math.pi**2 * Dg / R**2

    for i in range(len(modes_c)):
        n = i + 1
        n_coeff = (-1.0) ** n / n

        diffusion_rate = diffusion_rate_coeff * n**2
        source_rate = projection_coeff * beta * n_coeff

        A = [
            [1.0 + (diffusion_rate + g_b + g_d) * dt, -b_b * dt, -b_d * dt],
            [-g_b * dt, 1.0 + b_b * dt, 0.0],
            [-g_d * dt, 0.0, 1.0 + b_d * dt],
        ]
        rhs = [modes_c[i] + source_rate * dt, modes_mb[i], modes_md[i]]
        modes_c[i], modes_mb[i], modes_md[i] = solve3x3_cramer(A, rhs)

    return reconstruct_average(modes_c), reconstruct_average(modes_mb), reconstruct_average(modes_md)

# ============================================================
# SOLVER PRINCIPALE UNICO
# ============================================================

def solve_UN_sciantix_intragranular_extended(p):
    modes_c = initialize_modes_from_average(p.c0, p.n_modes)
    modes_mb = initialize_modes_from_average(p.mb0, p.n_modes)
    modes_md = initialize_modes_from_average(p.md0, p.n_modes)

    R_b = p.R_b
    R_d = p.R_d
    N_b = p.N_b
    N_d = p.N_d

    V_b = sphere_volume(R_b)
    V_d = sphere_volume(R_d)

    if p.nvb0 is None:
        nvb = initialize_vacancy_concentration(p, N_b, R_b, p.mb0)
    else:
        nvb = p.nvb0

    if p.nvd0 is None:
        nvd = initialize_vacancy_concentration(p, N_d, R_d, p.md0)
    else:
        nvd = p.nvd0

    beta = beta_production(p)
    initial_gas = p.c0 + p.mb0 + p.md0
    generated = 0.0
    q_gb = 0.0
    retained = initial_gas
    seed_vacancies_added_total = 0.0

    hist = {
        "time": [0.0],
        "burnup_percent_fima": [time_to_burnup_percent(0.0, p.fission_rate, p.lattice_parameter)],
        "c": [reconstruct_average(modes_c)],
        "mb": [reconstruct_average(modes_mb)],
        "md": [reconstruct_average(modes_md)],
        "Nb": [N_b],
        "Nd": [N_d],
        "Vb": [V_b],
        "Vd": [V_d],
        "Rb": [R_b],
        "Rd": [R_d],
        "nvb": [nvb],
        "nvd": [nvd],
        "generated": [generated],
        "retained": [retained],
        "q_gb": [q_gb],
        "qdot_gb": [0.0],
        "swelling_b": [N_b * V_b],
        "swelling_d": [N_d * V_d],
        "swelling_ig": [N_b * V_b + N_d * V_d],
        "p_b": [pressure_internal(p, p.mb0, nvb)],
        "p_d": [pressure_internal(p, p.md0, nvd)],
        "p_b_eq": [pressure_equilibrium(p, R_b)],
        "p_d_eq": [pressure_equilibrium(p, R_d)],
        "lambda_d": [coalescence_lambda(V_d, N_d)],
        "nu_b": [0.0],
        "phi_b": [0.0],
        "dnvb_dt": [0.0],
        "dnvd_dt": [0.0],
        "dVd_growth_dt": [0.0],
        "seed_vacancies_added_total": [0.0],
    }

    last_rates = {}
    n_steps = int(math.ceil(p.final_time / p.dt))

    for _ in range(n_steps):
        t_old = hist["time"][-1]
        dt = min(p.dt, p.final_time - t_old)
        if dt <= 0.0:
            break

        c_old = hist["c"][-1]
        mb_old = hist["mb"][-1]
        md_old = hist["md"][-1]

        Nb_old = N_b
        Nd_old = N_d
        Vd_old = V_d

        Dg, D_parts = xe_diffusivity_UN(p)
        Dv, Dv_parts = vacancy_diffusivity_UN(p)
        b_b, b_d = resolution_rates_UN(p, R_b, R_d)

        g_b_old, g_d, trapping_parts_old = trapping_rates_UN(p, Dg, R_b, Nb_old, R_d, Nd_old)

        # 1. Update N_b prima del gas solver.
        nu_b = nucleation_rate_bulk(p, Dg, c_old)
        phi_b = phi_bulk(mb_old, Nb_old)
        N_b = (Nb_old + dt * nu_b) / (1.0 + dt * b_b * phi_b)
        N_b = max(N_b, p.min_number_density)

        # 2. Seed opzionale per nuove bulk bubbles.
        dN_b_positive = max(N_b - Nb_old, 0.0)
        seed_vacancies_added_step = 0.0
        if dN_b_positive > 0.0 and p.bulk_seed_radius_nm > 0.0:
            V_seed = sphere_volume(p.bulk_seed_radius_nm * 1.0e-9)
            seed_vacancies_added_step = dN_b_positive * V_seed / omega_matrix(p)
            nvb += seed_vacancies_added_step
            seed_vacancies_added_total += seed_vacancies_added_step

        # 3. Correzione importante: ricalcola g_b con N_b aggiornato.
        g_b, _, trapping_parts_updated = trapping_rates_UN(p, Dg, R_b, N_b, R_d, Nd_old)
        trapping_parts = dict(trapping_parts_updated)
        trapping_parts["g_b_old_before_Nb_update"] = g_b_old
        trapping_parts["g_b_used_after_Nb_update"] = g_b
        trapping_parts["Nb_old"] = Nb_old
        trapping_parts["Nb_used_for_gb"] = N_b

        # 4. Gas solver.
        c_new, mb_new, md_new = sciantix_3x3_exchange_step(
            modes_c, modes_mb, modes_md,
            Dg, p.grain_radius, beta,
            g_b, g_d, b_b, b_d, dt,
        )

        # Protezione contro piccoli valori negativi numerici.
        c_new = max(c_new, 0.0)
        mb_new = max(mb_new, 0.0)
        md_new = max(md_new, 0.0)

        dmb_dt = (mb_new - mb_old) / dt
        dmd_dt = (md_new - md_old) / dt

        # 5. Vacancy update.
        if p.update_bulk_vacancies:
            nvb, dnvb_dt = vacancy_concentration_implicit_step(p, Dv, R_b, N_b, mb_new, nvb, dt)
        else:
            dnvb_dt = 0.0

        nvd, dnvd_dt = vacancy_concentration_implicit_step(p, Dv, R_d, Nd_old, md_new, nvd, dt)

        # 6. Volume growth prima della coalescenza.
        if N_b > 0.0:
            V_b_growth = V_b + dt * (
                p.omega_fg / N_b * dmb_dt + omega_matrix(p) / N_b * dnvb_dt
            )
            V_b_growth = max(V_b_growth, p.min_volume)
        else:
            V_b_growth = V_b

        if Nd_old > 0.0:
            dVd_growth_dt = p.omega_fg / Nd_old * dmd_dt + omega_matrix(p) / Nd_old * dnvd_dt
            V_d_growth = V_d + dt * dVd_growth_dt
            V_d_growth = max(V_d_growth, p.min_volume)
        else:
            dVd_growth_dt = 0.0
            V_d_growth = V_d

        # 7. Coalescenza dislocation bubbles.
        lambda_d = coalescence_lambda(Vd_old, Nd_old)
        dVd_positive = max(V_d_growth - Vd_old, 0.0)
        if dVd_positive > 0.0 and Nd_old > 0.0:
            N_d = Nd_old / (1.0 + 4.0 * lambda_d * Nd_old * dVd_positive)
        else:
            N_d = Nd_old
        N_d = max(N_d, p.min_number_density)

        # 8. Ricostruzione volumi da gas + vacancies.
        if N_b > 0.0:
            V_b = (p.omega_fg * max(mb_new, 0.0) + omega_matrix(p) * nvb) / N_b
        else:
            V_b = 0.0

        if N_d > 0.0:
            V_d = (p.omega_fg * max(md_new, 0.0) + omega_matrix(p) * nvd) / N_d
        else:
            V_d = 0.0

        V_b = max(V_b, p.min_volume)
        V_d = max(V_d, p.min_volume)
        R_b = radius_from_volume(V_b)
        R_d = radius_from_volume(V_d)

        # 9. Bilancio gas a grain face.
        generated += beta * dt
        retained = c_new + mb_new + md_new
        q_gb_old = q_gb
        q_gb = max(initial_gas + generated - retained, 0.0)
        qdot_gb = max((q_gb - q_gb_old) / dt, 0.0)

        hist["time"].append(t_old + dt)
        hist["burnup_percent_fima"].append(time_to_burnup_percent(t_old + dt, p.fission_rate, p.lattice_parameter))
        hist["c"].append(c_new)
        hist["mb"].append(mb_new)
        hist["md"].append(md_new)
        hist["Nb"].append(N_b)
        hist["Nd"].append(N_d)
        hist["Vb"].append(V_b)
        hist["Vd"].append(V_d)
        hist["Rb"].append(R_b)
        hist["Rd"].append(R_d)
        hist["nvb"].append(nvb)
        hist["nvd"].append(nvd)
        hist["generated"].append(generated)
        hist["retained"].append(retained)
        hist["q_gb"].append(q_gb)
        hist["qdot_gb"].append(qdot_gb)
        hist["swelling_b"].append(N_b * V_b)
        hist["swelling_d"].append(N_d * V_d)
        hist["swelling_ig"].append(N_b * V_b + N_d * V_d)
        hist["p_b"].append(pressure_internal(p, mb_new, nvb))
        hist["p_d"].append(pressure_internal(p, md_new, nvd))
        hist["p_b_eq"].append(pressure_equilibrium(p, R_b))
        hist["p_d_eq"].append(pressure_equilibrium(p, R_d))
        hist["lambda_d"].append(lambda_d)
        hist["nu_b"].append(nu_b)
        hist["phi_b"].append(phi_b)
        hist["dnvb_dt"].append(dnvb_dt)
        hist["dnvd_dt"].append(dnvd_dt)
        hist["dVd_growth_dt"].append(dVd_growth_dt)
        hist["seed_vacancies_added_total"].append(seed_vacancies_added_total)

        last_rates = {
            "Dg": Dg,
            "Dv": Dv,
            "beta": beta,
            "g_b": g_b,
            "g_b_old_before_Nb_update": g_b_old,
            "g_d": g_d,
            "b_b": b_b,
            "b_d": b_d,
            "nu_b": nu_b,
            "phi_b": phi_b,
            "lambda_d": lambda_d,
            "dVd_growth_dt": dVd_growth_dt,
            "dnvb_dt": dnvb_dt,
            "dnvd_dt": dnvd_dt,
            "seed_vacancies_added_step": seed_vacancies_added_step,
            "seed_vacancies_added_total": seed_vacancies_added_total,
            "p_b": pressure_internal(p, mb_new, nvb),
            "p_d": pressure_internal(p, md_new, nvd),
            "p_b_eq": pressure_equilibrium(p, R_b),
            "p_d_eq": pressure_equilibrium(p, R_d),
            "Nb_old": Nb_old,
            "Nb_updated_before_gas": N_b,
            "Nd_old": Nd_old,
            "Nd_after_coalescence": N_d,
            **D_parts,
            **Dv_parts,
            **trapping_parts,
        }

    return hist, last_rates

# ============================================================
# WRAPPER, SCORE E TEST DIAGNOSTICI PHASE A/B/C/D
# ============================================================

_RUN_CACHE = {}


def run_model_point(
    T,
    burnup,
    f_n,
    K_d=K_D_FIXED,
    gd_scale=1.0,
    gb_scale=1.0,
    bulk_seed_radius_nm=BULK_SEED_RADIUS_NM,
    dt_h=DT_SCAN_H,
    n_modes=N_MODES_SCAN,
):
    key = (
        round(float(T), 6),
        round(float(burnup), 6),
        float(f_n),
        float(K_d),
        float(gd_scale),
        float(gb_scale),
        float(bulk_seed_radius_nm),
        float(dt_h),
        int(n_modes),
    )
    if key in _RUN_CACHE:
        return _RUN_CACHE[key]

    p = UNParameters(
        temperature=float(T),
        fission_rate=FISSION_RATE,
        grain_radius=GRAIN_RADIUS,
        target_burnup_percent_fima=float(burnup),
        dt=float(dt_h) * 3600.0,
        n_modes=int(n_modes),
        xe_yield=XE_YIELD,
        f_n=float(f_n),
        K_d=float(K_d),
        g_d_scale=float(gd_scale),
        g_b_scale=float(gb_scale),
        R_b=0.0,
        N_b=0.0,
        R_d=0.0,
        N_d=None,
        c0=0.0,
        mb0=0.0,
        md0=0.0,
        vacancy_absorption_only=True,
        update_bulk_vacancies=True,
        bulk_seed_radius_nm=float(bulk_seed_radius_nm),
    )

    hist, rates = solve_UN_sciantix_intragranular_extended(p)

    generated = hist["generated"][-1]
    if generated > 0.0:
        matrix_gas_percent = 100.0 * hist["c"][-1] / generated
        bulk_gas_percent = 100.0 * hist["mb"][-1] / generated
        dislocation_gas_percent = 100.0 * hist["md"][-1] / generated
        qgb_gas_percent = 100.0 * hist["q_gb"][-1] / generated
    else:
        matrix_gas_percent = bulk_gas_percent = dislocation_gas_percent = qgb_gas_percent = 0.0

    p_b_eq = hist["p_b_eq"][-1]
    p_d_eq = hist["p_d_eq"][-1]

    row = {
        "T": float(T),
        "burnup": float(burnup),
        "f_n": float(f_n),
        "K_d": float(K_d),
        "gd_scale": float(gd_scale),
        "gb_scale": float(gb_scale),
        "bulk_seed_radius_nm": float(bulk_seed_radius_nm),
        "swelling_b_percent": 100.0 * hist["swelling_b"][-1],
        "swelling_d_percent": 100.0 * hist["swelling_d"][-1],
        "swelling_ig_percent": 100.0 * hist["swelling_ig"][-1],
        "Nb": hist["Nb"][-1],
        "Nd": hist["Nd"][-1],
        "Rb_nm": hist["Rb"][-1] * 1.0e9,
        "Rd_nm": hist["Rd"][-1] * 1.0e9,
        "p_b": hist["p_b"][-1],
        "p_b_eq": p_b_eq,
        "p_d": hist["p_d"][-1],
        "p_d_eq": p_d_eq,
        "p_b_over_eq": hist["p_b"][-1] / p_b_eq if p_b_eq > 0.0 else math.nan,
        "p_d_over_eq": hist["p_d"][-1] / p_d_eq if p_d_eq > 0.0 else math.nan,
        "matrix_gas_percent": matrix_gas_percent,
        "bulk_gas_percent": bulk_gas_percent,
        "dislocation_gas_percent": dislocation_gas_percent,
        "qgb_gas_percent": qgb_gas_percent,
        "Dg": rates.get("Dg", math.nan),
        "Dv": rates.get("Dv", math.nan),
        "g_b": rates.get("g_b", math.nan),
        "g_d": rates.get("g_d", math.nan),
        "g_b_unscaled": rates.get("g_b_unscaled", math.nan),
        "g_d_unscaled": rates.get("g_d_unscaled", math.nan),
        "hist": hist,
        "rates": rates,
    }

    _RUN_CACHE[key] = row
    return row


def rmse(values):
    values = [v for v in values if math.isfinite(v)]
    if not values:
        return math.nan
    return math.sqrt(sum(v * v for v in values) / len(values))


def log10_rmse(pairs):
    errs = []
    for pred, exp in pairs:
        if pred > 0.0 and exp > 0.0 and math.isfinite(pred) and math.isfinite(exp):
            errs.append(math.log10(pred / exp))
    return rmse(errs)


EXP_SWELLING_SUBSET = [
    p for p in EXP_SWELLING_T
    if (
        (abs(p["burnup"] - 1.1) < 1e-9 and p["T"] in {1127.0, 1485.0, 1598.0, 1669.0})
        or (abs(p["burnup"] - 1.3) < 1e-9 and p["T"] in {1044.0, 1377.0, 1595.0, 1639.0, 1724.0})
        or (abs(p["burnup"] - 3.2) < 1e-9 and p["T"] in {984.0, 1247.0, 1511.0, 1590.0})
    )
]

EXP_ND_SUBSET = [p for p in EXP_ND_T_13 if p["T"] in {1153.0, 1330.0, 1493.0, 1599.0, 1669.0, 1723.0}]
EXP_RD_SUBSET = [p for p in EXP_RD_T_13 if p["T"] in {1045.0, 1374.0, 1535.0, 1594.0, 1641.0, 1725.0}]


def score_combo_general(f_n, K_d=K_D_FIXED, gd_scale=1.0, gb_scale=1.0, bulk_seed_radius_nm=BULK_SEED_RADIUS_NM, full=False):
    """Score diagnostico unico. full=False usa subset leggero per griglie larghe."""
    swelling_points = EXP_SWELLING_T if full else EXP_SWELLING_SUBSET
    nd_points = EXP_ND_T_13 if full else EXP_ND_SUBSET
    rd_points = EXP_RD_T_13 if full else EXP_RD_SUBSET

    swelling_errors = []
    for exp in swelling_points:
        out = run_model_point(exp["T"], exp["burnup"], f_n, K_d=K_d, gd_scale=gd_scale, gb_scale=gb_scale, bulk_seed_radius_nm=bulk_seed_radius_nm)
        swelling_errors.append(out["swelling_d_percent"] - exp["swelling"])
    score_swelling = rmse(swelling_errors)

    n_pairs = []
    for exp in nd_points:
        out = run_model_point(exp["T"], 1.3, f_n, K_d=K_d, gd_scale=gd_scale, gb_scale=gb_scale, bulk_seed_radius_nm=bulk_seed_radius_nm)
        n_pairs.append((out["Nd"], exp["N"]))
    score_log_N = log10_rmse(n_pairs)

    r_pairs = []
    for exp in rd_points:
        out = run_model_point(exp["T"], 1.3, f_n, K_d=K_d, gd_scale=gd_scale, gb_scale=gb_scale, bulk_seed_radius_nm=bulk_seed_radius_nm)
        r_pairs.append((out["Rd_nm"], exp["R_nm"]))
    score_log_R = log10_rmse(r_pairs)

    pressure_cases = [(984.0, 3.2), (1044.0, 1.3), (1200.0, 3.2)]
    pressure_terms = []
    for T, bu in pressure_cases:
        out = run_model_point(T, bu, f_n, K_d=K_d, gd_scale=gd_scale, gb_scale=gb_scale, bulk_seed_radius_nm=bulk_seed_radius_nm)
        for key in ["p_b_over_eq", "p_d_over_eq"]:
            val = out[key]
            if val > 0.0 and math.isfinite(val):
                pressure_terms.append(abs(math.log10(max(val, 1.0e-30))))
    pressure_penalty = rmse(pressure_terms) if pressure_terms else 0.0

    out_1600_32 = run_model_point(1600.0, 3.2, f_n, K_d=K_d, gd_scale=gd_scale, gb_scale=gb_scale, bulk_seed_radius_nm=bulk_seed_radius_nm)
    bulk_fraction = out_1600_32["bulk_gas_percent"] / 100.0
    disl_fraction = out_1600_32["dislocation_gas_percent"] / 100.0
    partition_penalty = 0.0
    if bulk_fraction > 0.75:
        partition_penalty += bulk_fraction - 0.75
    if disl_fraction < 0.05:
        partition_penalty += 0.05 - disl_fraction

    total_score = (
        score_swelling
        + 0.75 * score_log_N
        + 0.75 * score_log_R
        + 0.30 * pressure_penalty
        + 0.50 * partition_penalty
    )

    out_1600_13 = run_model_point(1600.0, 1.3, f_n, K_d=K_d, gd_scale=gd_scale, gb_scale=gb_scale, bulk_seed_radius_nm=bulk_seed_radius_nm)

    return {
        "f_n": f_n,
        "K_d": K_d,
        "gd_scale": gd_scale,
        "gb_scale": gb_scale,
        "bulk_seed_radius_nm": bulk_seed_radius_nm,
        "score_total": total_score,
        "score_swelling_rmse_percent": score_swelling,
        "score_log10_N": score_log_N,
        "score_log10_R": score_log_R,
        "pressure_penalty": pressure_penalty,
        "partition_penalty": partition_penalty,
        "sw_d_32_1600": out_1600_32["swelling_d_percent"],
        "sw_b_32_1600": out_1600_32["swelling_b_percent"],
        "pb_ratio_32_1600": out_1600_32["p_b_over_eq"],
        "pd_ratio_32_1600": out_1600_32["p_d_over_eq"],
        "bulk_gas_32_1600": out_1600_32["bulk_gas_percent"],
        "disl_gas_32_1600": out_1600_32["dislocation_gas_percent"],
        "Rd_13_1600": out_1600_13["Rd_nm"],
        "Nd_13_1600": out_1600_13["Nd"],
    }


def print_ranking(rows, title, n=25):
    rows_sorted = sorted(rows, key=lambda r: r["score_total"])
    print("\n" + "=" * 180)
    print(title)
    print("=" * 180)
    print(
        f"{'rank':>4s} | {'score':>11s} | {'swRMSE':>10s} | {'logN':>8s} | {'logR':>8s} | "
        f"{'pPen':>8s} | {'partPen':>8s} | {'f_n':>10s} | {'K_d':>10s} | {'gd':>6s} | {'seed':>6s} | "
        f"{'swD32@1600':>11s} | {'swB32@1600':>11s} | {'pb/peq':>9s} | {'pd/peq':>9s} | "
        f"{'bulkGas%':>9s} | {'dislGas%':>9s} | {'Rd13@1600':>10s} | {'Nd13@1600':>11s}"
    )
    print("-" * 180)
    for i, r in enumerate(rows_sorted[:n], start=1):
        print(
            f"{i:4d} | {r['score_total']:11.4e} | {r['score_swelling_rmse_percent']:10.4e} | "
            f"{r['score_log10_N']:8.3e} | {r['score_log10_R']:8.3e} | "
            f"{r['pressure_penalty']:8.3e} | {r['partition_penalty']:8.3e} | "
            f"{r['f_n']:10.2e} | {r['K_d']:10.2e} | {r['gd_scale']:6.2g} | {r['bulk_seed_radius_nm']:6.2f} | "
            f"{r['sw_d_32_1600']:11.4e} | {r['sw_b_32_1600']:11.4e} | "
            f"{r['pb_ratio_32_1600']:9.3e} | {r['pd_ratio_32_1600']:9.3e} | "
            f"{r['bulk_gas_32_1600']:9.3e} | {r['disl_gas_32_1600']:9.3e} | "
            f"{r['Rd_13_1600']:10.3e} | {r['Nd_13_1600']:11.3e}"
        )
    return rows_sorted


# ============================================================
# OUTPUT E GRAFICI DIAGNOSTICI PER UN SET SCELTO
# ============================================================

def print_key_table(title, rows):
    print("\n" + "=" * 180)
    print(title)
    print("=" * 180)
    print(
        f"{'T_K':>8s} | {'sw_b_%':>10s} | {'sw_d_%':>10s} | {'sw_ig_%':>10s} | "
        f"{'Rb_nm':>10s} | {'Rd_nm':>10s} | {'Nb':>12s} | {'Nd':>12s} | "
        f"{'pb/peq':>10s} | {'pd/peq':>10s} | "
        f"{'matrix_%':>10s} | {'bulkGas_%':>10s} | {'dislGas_%':>10s} | {'qgb_%':>10s}"
    )
    print("-" * 180)
    wanted = {900.0, 1000.0, 1200.0, 1400.0, 1500.0, 1600.0, 1700.0, 1800.0, 2000.0}
    for r in rows:
        if r["T"] in wanted:
            print(
                f"{r['T']:8.1f} | "
                f"{r['swelling_b_percent']:10.4e} | {r['swelling_d_percent']:10.4e} | {r['swelling_ig_percent']:10.4e} | "
                f"{r['Rb_nm']:10.4e} | {r['Rd_nm']:10.4e} | "
                f"{r['Nb']:12.5e} | {r['Nd']:12.5e} | "
                f"{r['p_b_over_eq']:10.4e} | {r['p_d_over_eq']:10.4e} | "
                f"{r['matrix_gas_percent']:10.4e} | {r['bulk_gas_percent']:10.4e} | "
                f"{r['dislocation_gas_percent']:10.4e} | {r['qgb_gas_percent']:10.4e}"
            )


def plot_exp_swelling_for_burnup(bu):
    series_names = sorted({p["series"] for p in EXP_SWELLING_T if abs(p["burnup"] - bu) < 1.0e-9})
    for series in series_names:
        pts = [p for p in EXP_SWELLING_T if abs(p["burnup"] - bu) < 1.0e-9 and p["series"] == series]
        marker = "x" if "119" in series else "^"
        plt.scatter([p["T"] for p in pts], [p["swelling"] for p in pts], marker=marker, s=55, label=f"Exp {series}")


def make_diagnostics(label, f_n, K_d, gd_scale, bulk_seed_radius_nm=0.0):
    T_PLOT = [float(T) for T in range(900, 2001, 50)]
    T_AMOUNT = [float(T) for T in range(900, 2601, 100)]
    BURNUPS = [1.1, 1.3, 3.2]

    print("\n" + "#" * 150)
    print(f"DIAGNOSTICS: {label}")
    print(f"f_n={f_n:.3e}, K_d={K_d:.3e}, gd_scale={gd_scale:.3g}, bulk_seed={bulk_seed_radius_nm:.2f} nm")
    print("#" * 150)

    results_by_burnup = {}
    for bu in BURNUPS:
        rows = [run_model_point(T, bu, f_n, K_d=K_d, gd_scale=gd_scale, bulk_seed_radius_nm=bulk_seed_radius_nm) for T in T_PLOT]
        results_by_burnup[bu] = rows
        print_key_table(f"DETAILED TABLE - {label}, burnup={bu:.1f}% FIMA", rows)

    gas_partition = {
        bu: [run_model_point(T, bu, f_n, K_d=K_d, gd_scale=gd_scale, bulk_seed_radius_nm=bulk_seed_radius_nm) for T in T_AMOUNT]
        for bu in [1.1, 3.2]
    }

    if plt is None:
        print("\nmatplotlib non installato: grafici saltati.")
        return

    for bu in BURNUPS:
        rows = results_by_burnup[bu]
        Ts = [r["T"] for r in rows]
        plt.figure(figsize=(8, 5))
        plt.plot(Ts, [r["swelling_d_percent"] for r in rows], label="Model dislocation")
        plt.plot(Ts, [r["swelling_b_percent"] for r in rows], linestyle=":", label="Model bulk")
        plot_exp_swelling_for_burnup(bu)
        plt.xlabel("T [K]")
        plt.ylabel("Fission gas swelling [%]")
        plt.title(f"Swelling vs T {bu:.1f}% FIMA - {label}")
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()

    rows = results_by_burnup[1.3]
    Ts = [r["T"] for r in rows]
    plt.figure(figsize=(8, 5))
    plt.plot(Ts, [r["Nd"] for r in rows], label="Model dislocation N_d")
    plt.plot(Ts, [r["Nb"] for r in rows], linestyle=":", label="Model bulk N_b")
    plt.scatter([p["T"] for p in EXP_ND_T_13], [p["N"] for p in EXP_ND_T_13], marker="^", s=55, label="Exp large bubbles")
    plt.yscale("log")
    plt.xlabel("T [K]")
    plt.ylabel("Bubble concentration [m$^{-3}$]")
    plt.title(f"Bubble concentration 1.3% FIMA - {label}")
    plt.grid(True, which="both", alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(8, 5))
    plt.plot(Ts, [r["Rd_nm"] for r in rows], label="Model dislocation R_d")
    plt.plot(Ts, [r["Rb_nm"] for r in rows], linestyle=":", label="Model bulk R_b")
    plt.scatter([p["T"] for p in EXP_RD_T_13], [p["R_nm"] for p in EXP_RD_T_13], marker="^", s=55, label="Exp large bubbles")
    plt.yscale("log")
    plt.xlabel("T [K]")
    plt.ylabel("Bubble radius [nm]")
    plt.title(f"Bubble radius 1.3% FIMA - {label}")
    plt.grid(True, which="both", alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

    rows = results_by_burnup[3.2]
    Ts = [r["T"] for r in rows]
    plt.figure(figsize=(8, 5))
    plt.plot(Ts, [r["p_b"] for r in rows], label="Bulk pressure")
    plt.plot(Ts, [r["p_b_eq"] for r in rows], linestyle=":", label="Bulk equilibrium")
    plt.plot(Ts, [r["p_d"] for r in rows], label="Dislocation pressure")
    plt.plot(Ts, [r["p_d_eq"] for r in rows], linestyle="--", label="Dislocation equilibrium")
    plt.yscale("log")
    plt.xlabel("T [K]")
    plt.ylabel("Bubble pressure [Pa]")
    plt.title(f"Pressure 3.2% FIMA - {label}")
    plt.grid(True, which="both", alpha=0.3)
    plt.legend(fontsize=8)
    plt.tight_layout()
    plt.show()

    for bu, rows in gas_partition.items():
        Ts = [r["T"] for r in rows]
        plt.figure(figsize=(8, 5))
        plt.plot(Ts, [r["matrix_gas_percent"] for r in rows], label="Matrix")
        plt.plot(Ts, [r["bulk_gas_percent"] for r in rows], linestyle=":", label="Bulk bubbles")
        plt.plot(Ts, [r["dislocation_gas_percent"] for r in rows], linestyle="--", label="Dislocation bubbles")
        plt.plot(Ts, [r["qgb_gas_percent"] for r in rows], linestyle="-.", label="Gas to grain face q_gb")
        plt.xlabel("T [K]")
        plt.ylabel("Amount of generated gas [%]")
        plt.title(f"Gas partition {bu:.1f}% FIMA - {label}")
        plt.xlim(900, 2600)
        plt.ylim(0, 100)
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()


# ============================================================
# MAIN
# ============================================================

if RUN_SINGLE_TEST:
    test = run_model_point(1600.0, 1.3, 3.0e-9)
    print(test)

if RUN_CALIBRATION:
    print("\n" + "#" * 150)
    print("START UN MODEL SENSITIVITY SUITE: f_n, g_d_scale, K_d, bulk seed")
    print("#" * 150)
    print(f"RUN_MODE              = {RUN_MODE}")
    print(f"dt scan               = {DT_SCAN_H:.2f} h")
    print(f"n_modes               = {N_MODES_SCAN:d}")
    print(f"Xe yield              = {XE_YIELD:.6e}")
    print(f"baseline K_d          = {K_D_FIXED:.6e} bub/m")
    print(f"baseline seed         = {BULK_SEED_RADIUS_NM:.3f} nm")
    print("Target principale: experimental large/P2 bubbles -> model dislocation bubbles.")

    phase_a_sorted = []
    if RUN_PHASE_A_FN_ONLY:
        print("\nPHASE A: f_n only, K_d fixed, gd_scale=1, seed baseline")
        phase_a = []
        for f_n in FN_GRID:
            print(f"  Running Phase A f_n={f_n:.3e}")
            phase_a.append(score_combo_general(f_n, K_d=K_D_FIXED, gd_scale=1.0, bulk_seed_radius_nm=BULK_SEED_RADIUS_NM, full=True))
        phase_a_sorted = print_ranking(phase_a, "PHASE A - f_n ONLY, FULL DIGITALIZED SCORE", n=len(phase_a))

    phase_b_sorted = []
    if RUN_PHASE_B_GRID:
        print("\nPHASE B: grid f_n x gd_scale x K_d, no bulk seed, subset score")
        phase_b = []
        total = len(FN_GRID_B) * len(GD_SCALE_GRID) * len(K_D_GRID)
        count = 0
        for f_n in FN_GRID_B:
            for K_d in K_D_GRID:
                for gd in GD_SCALE_GRID:
                    count += 1
                    print(f"  Phase B {count}/{total}: f_n={f_n:.2e}, K_d={K_d:.2e}, gd={gd:g}")
                    phase_b.append(score_combo_general(f_n, K_d=K_d, gd_scale=gd, bulk_seed_radius_nm=0.0, full=False))
        phase_b_sorted = print_ranking(phase_b, "PHASE B - f_n x gd_scale x K_d, NO SEED, SUBSET SCORE", n=40)

        print("\nPHASE B2: re-score top Phase B with full digitalized score")
        top = phase_b_sorted[:12]
        phase_b2 = []
        for r in top:
            print(f"  Full re-score: f_n={r['f_n']:.2e}, K_d={r['K_d']:.2e}, gd={r['gd_scale']:g}")
            phase_b2.append(score_combo_general(r["f_n"], K_d=r["K_d"], gd_scale=r["gd_scale"], bulk_seed_radius_nm=0.0, full=True))
        phase_b_sorted = print_ranking(phase_b2, "PHASE B2 - TOP GRID FULL DIGITALIZED SCORE", n=len(phase_b2))

    phase_c_sorted = []
    if RUN_PHASE_C_FIXED_FN_GD:
        print("\nPHASE C: fixed f_n values, vary gd_scale and K_d, full score")
        fixed_fns = []
        if phase_a_sorted:
            fixed_fns.append(phase_a_sorted[0]["f_n"])
        for val in [3.0e-9, 1.0e-8, 1.0e-7, 1.0e-6]:
            if val not in fixed_fns:
                fixed_fns.append(val)
        phase_c = []
        for f_n in fixed_fns:
            for K_d in K_D_GRID:
                for gd in GD_SCALE_GRID:
                    print(f"  Phase C: f_n={f_n:.2e}, K_d={K_d:.2e}, gd={gd:g}")
                    phase_c.append(score_combo_general(f_n, K_d=K_d, gd_scale=gd, bulk_seed_radius_nm=0.0, full=True))
        phase_c_sorted = print_ranking(phase_c, "PHASE C - FIXED f_n OPTIONS, FULL DIGITALIZED SCORE", n=40)

    phase_d_sorted = []
    if RUN_PHASE_D_BULK_SEED:
        print("\nPHASE D: bulk seed sensitivity on best candidates")
        candidates = []
        seen = set()
        for source in [phase_b_sorted[:5], phase_c_sorted[:5], phase_a_sorted[:3]]:
            for r in source:
                key = (r["f_n"], r["K_d"], r["gd_scale"])
                if key not in seen:
                    seen.add(key)
                    candidates.append({"f_n": r["f_n"], "K_d": r["K_d"], "gd_scale": r["gd_scale"]})
        phase_d = []
        for c in candidates[:10]:
            for seed in BULK_SEED_GRID:
                print(f"  Phase D: f_n={c['f_n']:.2e}, K_d={c['K_d']:.2e}, gd={c['gd_scale']:g}, seed={seed:.1f} nm")
                phase_d.append(score_combo_general(c["f_n"], K_d=c["K_d"], gd_scale=c["gd_scale"], bulk_seed_radius_nm=seed, full=True))
        phase_d_sorted = print_ranking(phase_d, "PHASE D - BULK SEED SENSITIVITY, FULL DIGITALIZED SCORE", n=40)

    all_ranked = []
    for arr in [phase_a_sorted[:3], phase_b_sorted[:5], phase_c_sorted[:5], phase_d_sorted[:5]]:
        all_ranked.extend(arr)
    all_ranked = sorted(all_ranked, key=lambda r: r["score_total"])

    if all_ranked:
        best = all_ranked[0]
        make_diagnostics("BEST_OVERALL", best["f_n"], best["K_d"], best["gd_scale"], best["bulk_seed_radius_nm"])

    if phase_a_sorted:
        base = phase_a_sorted[0]
        make_diagnostics("BEST_PHASE_A_FN_ONLY", base["f_n"], base["K_d"], base["gd_scale"], base["bulk_seed_radius_nm"])

    print("\n" + "#" * 150)
    print("END UN MODEL SENSITIVITY SUITE")
    print("#" * 150)
    print("Mandami il notebook/output completo: controlleremo soprattutto PHASE B2, PHASE C e PHASE D.")
````

````python
import math
from dataclasses import dataclass
from typing import Optional

try:
    import matplotlib.pyplot as plt
except ModuleNotFoundError:
    plt = None

# ============================================================
# CONFIGURAZIONE GENERALE
# ============================================================

FISSION_RATE = 5.0e19
GRAIN_RADIUS = 6.0e-6
K_D_FIXED = 5.0e5
XE_YIELD = 0.24

DT_SCAN_H = 12.0
N_MODES_SCAN = 25

# 0.0 = nessun seed artificiale. Metti 2.0 solo per diagnostica pressione bulk.
BULK_SEED_RADIUS_NM = 0.0

# Modalità: FAST per il primo giro; FULL per una griglia più densa.
RUN_MODE = "FAST"

FN_GRID = [
    3.0e-10,
    5.0e-10,
    7.0e-10,
    1.0e-9,
    1.5e-9,
    2.0e-9,
    3.0e-9,
    5.0e-9,
    7.0e-9,
    1.0e-8,
    3.0e-8,
    1.0e-7,
    3.0e-7,
    1.0e-6,
]

if RUN_MODE.upper() == "FULL":
    FN_GRID_B = [3.0e-9, 1.0e-8, 3.0e-8, 1.0e-7, 3.0e-7, 1.0e-6]
    GD_SCALE_GRID = [0.5, 1.0, 2.0, 3.0, 5.0, 7.0, 10.0, 20.0]
    K_D_GRID = [3.0e5, 5.0e5, 8.0e5, 1.2e6, 2.0e6, 3.0e6]
else:
    FN_GRID_B = [3.0e-9, 1.0e-8, 3.0e-8, 1.0e-7, 3.0e-7, 1.0e-6]
    GD_SCALE_GRID = [1.0, 2.0, 3.0, 5.0, 10.0]
    K_D_GRID = [5.0e5, 1.2e6, 2.0e6]

BULK_SEED_GRID = [0.0, 1.0, 2.0]

# Se vuoi fare solo un test veloce, metti False e usa RUN_SINGLE_TEST=True.
RUN_CALIBRATION = True
RUN_PHASE_A_FN_ONLY = True
RUN_PHASE_B_GRID = True
RUN_PHASE_C_FIXED_FN_GD = True
RUN_PHASE_D_BULK_SEED = True
RUN_SINGLE_TEST = False

# ============================================================
# DATI SPERIMENTALI DIGITALIZZATI DAL MARKDOWN
# ============================================================

EXP_SWELLING_T = [
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1127.0, "swelling": 0.68},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1228.0, "swelling": 0.59},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1312.0, "swelling": 0.43},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1402.0, "swelling": 0.58},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1485.0, "swelling": 1.22},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1549.0, "swelling": 1.84},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1598.0, "swelling": 1.66},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1632.0, "swelling": 2.13},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1669.0, "swelling": 3.60},
    {"figure": "Fig3a", "burnup": 1.1, "series": "100 kW/m", "T": 1685.0, "swelling": 2.72},

    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 899.0,  "swelling": 0.63},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1154.0, "swelling": 1.17},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1228.0, "swelling": 1.08},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1325.0, "swelling": 1.28},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1435.0, "swelling": 1.32},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1514.0, "swelling": 2.10},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1570.0, "swelling": 2.72},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1608.0, "swelling": 2.91},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1635.0, "swelling": 3.28},
    {"figure": "Fig3a", "burnup": 1.1, "series": "119 kW/m", "T": 1656.0, "swelling": 3.75},

    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1044.0, "swelling": 1.11},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1220.0, "swelling": 1.22},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1377.0, "swelling": 1.33},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1534.0, "swelling": 2.83},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1595.0, "swelling": 3.53},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1639.0, "swelling": 3.86},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1661.0, "swelling": 2.45},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1709.0, "swelling": 2.93},
    {"figure": "Fig3b", "burnup": 1.3, "series": "measurement", "T": 1724.0, "swelling": 3.15},

    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 984.0,  "swelling": 0.72},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1056.0, "swelling": 1.06},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1126.0, "swelling": 1.26},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1247.0, "swelling": 1.58},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1343.0, "swelling": 1.79},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1420.0, "swelling": 2.08},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1459.0, "swelling": 2.40},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1511.0, "swelling": 2.83},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1557.0, "swelling": 3.31},
    {"figure": "Fig3c", "burnup": 3.2, "series": "measurement", "T": 1590.0, "swelling": 3.75},
]

EXP_SWELLING_BURNUP_1600 = [
    {"burnup": 1.12, "swelling": 1.64, "series": "approx 100 kW/m"},
    {"burnup": 1.11, "swelling": 2.90, "series": "approx 119 kW/m"},
    {"burnup": 1.31, "swelling": 3.51, "series": "measurement"},
    {"burnup": 3.18, "swelling": 3.72, "series": "measurement"},
]

EXP_ND_T_13 = [
    {"T": 1153.0, "N": 5.22e19},
    {"T": 1202.0, "N": 3.69e19},
    {"T": 1227.0, "N": 3.00e19},
    {"T": 1235.0, "N": 3.57e19},
    {"T": 1248.0, "N": 2.44e19},
    {"T": 1330.0, "N": 2.90e19},
    {"T": 1338.0, "N": 4.10e19},
    {"T": 1377.0, "N": 2.71e19},
    {"T": 1408.0, "N": 1.67e19},
    {"T": 1427.0, "N": 1.85e19},
    {"T": 1493.0, "N": 3.00e19},
    {"T": 1507.0, "N": 2.44e19},
    {"T": 1524.0, "N": 1.56e19},
    {"T": 1538.0, "N": 7.54e18},
    {"T": 1555.0, "N": 7.80e18},
    {"T": 1561.0, "N": 1.92e19},
    {"T": 1599.0, "N": 5.34e18},
    {"T": 1622.0, "N": 7.80e18},
    {"T": 1628.0, "N": 1.73e19},
    {"T": 1649.0, "N": 5.34e18},
    {"T": 1656.0, "N": 3.65e18},
    {"T": 1656.0, "N": 2.41e18},
    {"T": 1669.0, "N": 1.96e18},
    {"T": 1682.0, "N": 2.44e19},
    {"T": 1685.0, "N": 1.40e19},
    {"T": 1723.0, "N": 7.70e17},
    {"T": 1742.0, "N": 1.02e18},
    {"T": 1739.0, "N": 1.65e18},
]

EXP_RD_T_13 = [
    {"T": 1045.0, "R_nm": 54.06},
    {"T": 1219.0, "R_nm": 60.37},
    {"T": 1374.0, "R_nm": 69.58},
    {"T": 1535.0, "R_nm": 104.85},
    {"T": 1594.0, "R_nm": 120.83},
    {"T": 1641.0, "R_nm": 143.73},
    {"T": 1663.0, "R_nm": 122.76},
    {"T": 1710.0, "R_nm": 157.99},
    {"T": 1725.0, "R_nm": 173.67},
]

# ============================================================
# PARAMETRI E FUNZIONI BASE
# ============================================================

@dataclass
class UNParameters:
    temperature: float = 1600.0
    fission_rate: float = 5.0e19
    grain_radius: float = 5.0e-6
    final_time: float = 24.0 * 3600.0
    target_burnup_percent_fima: Optional[float] = None
    dt: float = 60.0
    n_modes: int = 40

    xe_yield: float = 0.24
    precursor_factor: float = 1.0

    D10: float = 1.56e-3
    Q1: float = 4.94
    A20: float = 1.21e-67
    B21: float = 25.87
    B22: float = -1.49
    B23: float = 0.0
    A30: float = 1.85e-39

    kB_eV: float = 8.617333262e-5
    kB_J: float = 1.380649e-23

    use_sciantix_D2_zero: bool = True

    D10_vU: float = 1.35e-2
    Q1_vU: float = 5.66
    B21_vU_refit: float = -0.62
    B22_vU_refit: float = -0.04
    A20_vU_fig4_refit: float = 4.6304523933553033e-29

    radius_in_lattice: float = 0.21e-9
    omega_fg: float = 8.5e-29
    lattice_parameter: float = 4.889e-10
    gamma_b: float = 1.11
    hydrostatic_stress: float = 0.0
    min_radius_for_pressure: float = 1.0e-15

    f_n: float = 1.0e-6

    rho_d: float = 3.0e13
    K_d: float = 5.0e5
    r_d: float = 3.46e-10
    Z_d: float = 5.0

    R_b: float = 0.0
    N_b: float = 0.0
    R_d: float = 0.0
    N_d: Optional[float] = None

    c0: float = 0.0
    mb0: float = 0.0
    md0: float = 0.0

    nvb0: Optional[float] = None
    nvd0: Optional[float] = None

    diffusivity_sf: float = 1.0
    trapping_sf: float = 1.0
    g_b_scale: float = 1.0
    g_d_scale: float = 1.0
    resolution_sf: float = 1.0

    vacancy_absorption_only: bool = True
    update_bulk_vacancies: bool = True
    min_number_density: float = 0.0
    min_volume: float = 0.0

    bulk_seed_radius_nm: float = 0.0

    def __post_init__(self):
        if self.N_d is None:
            self.N_d = self.K_d * self.rho_d

        if self.target_burnup_percent_fima is not None:
            self.final_time = burnup_percent_to_time(
                self.target_burnup_percent_fima,
                self.fission_rate,
                self.lattice_parameter,
            )


def omega_matrix(p):
    return p.lattice_parameter**3 / 4.0


def uranium_atom_density_from_lattice(lattice_parameter):
    return 4.0 / lattice_parameter**3


def burnup_percent_to_time(burnup_percent_fima, fission_rate, lattice_parameter):
    if fission_rate <= 0.0:
        raise ValueError("fission_rate must be positive to compute time from burnup.")

    burnup_fraction_fima = burnup_percent_fima / 100.0
    uranium_density = uranium_atom_density_from_lattice(lattice_parameter)
    return burnup_fraction_fima * uranium_density / fission_rate


def time_to_burnup_percent(time, fission_rate, lattice_parameter):
    uranium_density = uranium_atom_density_from_lattice(lattice_parameter)
    return 100.0 * fission_rate * time / uranium_density


def sphere_volume(R):
    if R <= 0.0:
        return 0.0
    return (4.0 / 3.0) * math.pi * R**3


def radius_from_volume(V):
    if V <= 0.0:
        return 0.0
    return (3.0 * V / (4.0 * math.pi)) ** (1.0 / 3.0)


def xe_diffusivity_UN(p):
    T = p.temperature
    F = p.fission_rate
    kBT = p.kB_eV * T

    D1 = p.D10 * math.exp(-p.Q1 / kBT)

    if p.use_sciantix_D2_zero:
        D2 = 0.0
    else:
        D2 = p.A20 * math.sqrt(F) * math.exp(
            -p.B21 / kBT
            -p.B22 / (kBT**2)
            -p.B23 / (kBT**3)
        )

    D3 = p.A30 * F
    Dg = (D1 + D2 + D3) * p.diffusivity_sf * p.precursor_factor
    return Dg, {"D1": D1, "D2": D2, "D3": D3, "Dg": Dg}


def vacancy_diffusivity_UN(p):
    T = p.temperature
    F = p.fission_rate
    kBT = p.kB_eV * T

    D1 = p.D10_vU * math.exp(-p.Q1_vU / kBT)
    D2 = math.sqrt(F) * p.A20_vU_fig4_refit * math.exp(
        p.B21_vU_refit / kBT
        + p.B22_vU_refit / (kBT**2)
    )

    Dv = D1 + D2
    return Dv, {"Dv1": D1, "Dv2": D2, "Dv": Dv}


def b0_resolution(R):
    R = max(R, 1.0e-15)
    return 1.0e-25 * (2.64 - 2.02 * math.exp(-2.61e-9 / R))


def resolution_rates_UN(p, R_b, R_d):
    Rb_eff = R_b + p.radius_in_lattice
    Rd_eff = R_d + p.radius_in_lattice

    b_b = p.fission_rate * b0_resolution(Rb_eff) * p.resolution_sf
    b_d = p.fission_rate * b0_resolution(Rd_eff) * p.resolution_sf
    return b_b, b_d


def trapping_rates_UN(p, Dg, R_b, N_b, R_d, N_d):
    Rb_eff = R_b + p.radius_in_lattice
    Rd_eff = R_d + p.radius_in_lattice

    if N_b <= 0.0:
        g_b = 0.0
    else:
        g_b = 4.0 * math.pi * Dg * Rb_eff * N_b

    Gamma_d = 1.0 / math.sqrt(math.pi * p.rho_d)
    den = math.log(Gamma_d / (p.Z_d * p.r_d)) - 3.0 / 5.0

    if den <= 0.0:
        raise ValueError(f"Invalid dislocation sink denominator: {den:g}")

    free_dislocation = max(p.rho_d - 2.0 * R_d * N_d, 0.0)
    term_bubbles = 4.0 * math.pi * Dg * Rd_eff * N_d
    term_dislocation = (2.0 * math.pi * Dg / den) * free_dislocation
    g_d = term_bubbles + term_dislocation

    g_b_scaled = p.trapping_sf * p.g_b_scale * g_b
    g_d_scaled = p.trapping_sf * p.g_d_scale * g_d

    return g_b_scaled, g_d_scaled, {
        "Gamma_d": Gamma_d,
        "den": den,
        "free_dislocation": free_dislocation,
        "term_bubbles": term_bubbles,
        "term_dislocation": term_dislocation,
        "g_b_unscaled": g_b,
        "g_d_unscaled": g_d,
        "g_b_scale": p.g_b_scale,
        "g_d_scale": p.g_d_scale,
    }


def beta_production(p):
    return p.xe_yield * p.fission_rate


def nucleation_rate_bulk(p, Dg, c):
    return 8.0 * math.pi * p.f_n * Dg * p.omega_fg ** (1.0 / 3.0) * max(c, 0.0) ** 2


def phi_bulk(mb, Nb):
    if Nb <= 0.0:
        return 0.0

    atoms_per_bubble = mb / Nb
    if atoms_per_bubble <= 1.0:
        return 0.0

    return 1.0 / (atoms_per_bubble - 1.0)


def coalescence_lambda(Vd, Nd):
    xi = Vd * Nd
    xi = max(0.0, min(xi, 0.999999))
    return (2.0 - xi) / (2.0 * (1.0 - xi) ** 3)


def pressure_internal(p, m_gas, n_vac):
    if m_gas <= 0.0:
        return 0.0
    if n_vac <= 0.0:
        return math.inf

    denominator = n_vac * omega_matrix(p)
    if denominator <= 0.0:
        return math.inf

    return p.kB_J * p.temperature * m_gas / denominator


def pressure_equilibrium(p, R):
    R_eff = max(R, p.min_radius_for_pressure)
    return 2.0 * p.gamma_b / R_eff - p.hydrostatic_stress


def gas_only_radius_for_population(p, m_gas, N):
    if m_gas <= 0.0 or N <= 0.0:
        return 0.0
    V_gas_per_bubble = p.omega_fg * m_gas / N
    return radius_from_volume(V_gas_per_bubble)


def radius_for_vacancy_update(p, R_old, N, m_gas):
    if R_old > 0.0:
        return R_old
    return gas_only_radius_for_population(p, m_gas, N)


def pressure_equilibrium_for_vacancy_update(p, R_update):
    if R_update <= 0.0:
        return math.inf
    return 2.0 * p.gamma_b / R_update - p.hydrostatic_stress


def wigner_seitz_delta(N):
    return (3.0 / (4.0 * math.pi * max(N, 1.0))) ** (1.0 / 3.0)


def zeta_geometry(R, N):
    delta = wigner_seitz_delta(N)
    psi = max(R / delta, 1.0e-12)
    den = -psi**6 + 5.0 * psi**2 - 9.0 * psi + 5.0

    if den <= 1.0e-30:
        den = 1.0e-30

    return max(10.0 * psi * (1.0 + psi**3) / den, 1.0e-30)


def vacancy_concentration_implicit_step(p, Dv, R, N, m_gas, n_old, dt):
    if N <= 0.0 or m_gas <= 0.0:
        return n_old, 0.0

    R_update = radius_for_vacancy_update(p, R, N, m_gas)
    if R_update <= 0.0:
        return n_old, 0.0

    p_eq = pressure_equilibrium_for_vacancy_update(p, R_update)
    p_int_old = pressure_internal(p, m_gas, n_old)

    if p.vacancy_absorption_only and p_int_old <= p_eq:
        return n_old, 0.0

    delta = wigner_seitz_delta(N)
    zeta = zeta_geometry(R_update, N)

    A = 2.0 * math.pi * Dv * delta * N / (p.kB_J * p.temperature * zeta)
    C = p.kB_J * p.temperature * m_gas / omega_matrix(p)

    B = n_old - dt * A * p_eq
    discriminant = B * B + 4.0 * dt * A * C

    if discriminant < 0.0:
        raise ValueError(f"Negative discriminant in vacancy implicit step: {discriminant:g}")

    sqrt_disc = math.sqrt(discriminant)

    if B >= 0.0:
        n_new = 0.5 * (B + sqrt_disc)
    else:
        denominator = sqrt_disc - B
        if denominator <= 0.0:
            n_new = 0.0
        else:
            n_new = (2.0 * dt * A * C) / denominator

    if p.vacancy_absorption_only:
        n_new = max(n_new, n_old)

    dn_dt = (n_new - n_old) / dt
    return n_new, dn_dt


def initialize_vacancy_concentration(p, N, R, m_gas):
    if N <= 0.0 or R <= 0.0:
        return 0.0

    total_bubble_volume = N * sphere_volume(R)
    gas_volume = p.omega_fg * m_gas
    vacancy_volume = max(total_bubble_volume - gas_volume, 0.0)
    return vacancy_volume / omega_matrix(p)


def initialize_modes_from_average(average, n_modes, n_iter=20):
    modes = [0.0 for _ in range(n_modes)]
    projection_coeff = -math.sqrt(8.0 / math.pi)
    remainder = average

    for _ in range(n_iter):
        reconstructed = 0.0
        for i in range(n_modes):
            n = i + 1
            n_coeff = (-1.0) ** n / n
            modes[i] += projection_coeff * n_coeff * remainder
            reconstructed += projection_coeff * n_coeff * modes[i] * 3.0 / (4.0 * math.pi)
        remainder = average - reconstructed

    return modes


def reconstruct_average(modes):
    projection_coeff = -2.0 * math.sqrt(2.0 / math.pi)
    average = 0.0

    for i, value in enumerate(modes):
        n = i + 1
        n_coeff = (-1.0) ** n / n
        average += projection_coeff * n_coeff * value / ((4.0 / 3.0) * math.pi)

    return average


def det3(A):
    return (
        A[0][0] * (A[1][1] * A[2][2] - A[1][2] * A[2][1])
        - A[0][1] * (A[1][0] * A[2][2] - A[1][2] * A[2][0])
        + A[0][2] * (A[1][0] * A[2][1] - A[1][1] * A[2][0])
    )


def solve3x3_cramer(A, b):
    detA = det3(A)
    if abs(detA) < 1.0e-300:
        raise ZeroDivisionError("Singular 3x3 system in spectral mode solve")

    Ax = [[b[i], A[i][1], A[i][2]] for i in range(3)]
    Ay = [[A[i][0], b[i], A[i][2]] for i in range(3)]
    Az = [[A[i][0], A[i][1], b[i]] for i in range(3)]

    return [det3(Ax) / detA, det3(Ay) / detA, det3(Az) / detA]


def sciantix_3x3_exchange_step(modes_c, modes_mb, modes_md, Dg, R, beta, g_b, g_d, b_b, b_d, dt):
    projection_coeff = -2.0 * math.sqrt(2.0 / math.pi)
    diffusion_rate_coeff = math.pi**2 * Dg / R**2

    for i in range(len(modes_c)):
        n = i + 1
        n_coeff = (-1.0) ** n / n

        diffusion_rate = diffusion_rate_coeff * n**2
        source_rate = projection_coeff * beta * n_coeff

        A = [
            [1.0 + (diffusion_rate + g_b + g_d) * dt, -b_b * dt, -b_d * dt],
            [-g_b * dt, 1.0 + b_b * dt, 0.0],
            [-g_d * dt, 0.0, 1.0 + b_d * dt],
        ]
        rhs = [modes_c[i] + source_rate * dt, modes_mb[i], modes_md[i]]
        modes_c[i], modes_mb[i], modes_md[i] = solve3x3_cramer(A, rhs)

    return reconstruct_average(modes_c), reconstruct_average(modes_mb), reconstruct_average(modes_md)

# ============================================================
# SOLVER PRINCIPALE UNICO
# ============================================================

def solve_UN_sciantix_intragranular_extended(p):
    modes_c = initialize_modes_from_average(p.c0, p.n_modes)
    modes_mb = initialize_modes_from_average(p.mb0, p.n_modes)
    modes_md = initialize_modes_from_average(p.md0, p.n_modes)

    R_b = p.R_b
    R_d = p.R_d
    N_b = p.N_b
    N_d = p.N_d

    V_b = sphere_volume(R_b)
    V_d = sphere_volume(R_d)

    if p.nvb0 is None:
        nvb = initialize_vacancy_concentration(p, N_b, R_b, p.mb0)
    else:
        nvb = p.nvb0

    if p.nvd0 is None:
        nvd = initialize_vacancy_concentration(p, N_d, R_d, p.md0)
    else:
        nvd = p.nvd0

    beta = beta_production(p)
    initial_gas = p.c0 + p.mb0 + p.md0
    generated = 0.0
    q_gb = 0.0
    retained = initial_gas
    seed_vacancies_added_total = 0.0

    hist = {
        "time": [0.0],
        "burnup_percent_fima": [time_to_burnup_percent(0.0, p.fission_rate, p.lattice_parameter)],
        "c": [reconstruct_average(modes_c)],
        "mb": [reconstruct_average(modes_mb)],
        "md": [reconstruct_average(modes_md)],
        "Nb": [N_b],
        "Nd": [N_d],
        "Vb": [V_b],
        "Vd": [V_d],
        "Rb": [R_b],
        "Rd": [R_d],
        "nvb": [nvb],
        "nvd": [nvd],
        "generated": [generated],
        "retained": [retained],
        "q_gb": [q_gb],
        "qdot_gb": [0.0],
        "swelling_b": [N_b * V_b],
        "swelling_d": [N_d * V_d],
        "swelling_ig": [N_b * V_b + N_d * V_d],
        "p_b": [pressure_internal(p, p.mb0, nvb)],
        "p_d": [pressure_internal(p, p.md0, nvd)],
        "p_b_eq": [pressure_equilibrium(p, R_b)],
        "p_d_eq": [pressure_equilibrium(p, R_d)],
        "lambda_d": [coalescence_lambda(V_d, N_d)],
        "nu_b": [0.0],
        "phi_b": [0.0],
        "dnvb_dt": [0.0],
        "dnvd_dt": [0.0],
        "dVd_growth_dt": [0.0],
        "seed_vacancies_added_total": [0.0],
    }

    last_rates = {}
    n_steps = int(math.ceil(p.final_time / p.dt))

    for _ in range(n_steps):
        t_old = hist["time"][-1]
        dt = min(p.dt, p.final_time - t_old)
        if dt <= 0.0:
            break

        c_old = hist["c"][-1]
        mb_old = hist["mb"][-1]
        md_old = hist["md"][-1]

        Nb_old = N_b
        Nd_old = N_d
        Vd_old = V_d

        Dg, D_parts = xe_diffusivity_UN(p)
        Dv, Dv_parts = vacancy_diffusivity_UN(p)
        b_b, b_d = resolution_rates_UN(p, R_b, R_d)

        g_b_old, g_d, trapping_parts_old = trapping_rates_UN(p, Dg, R_b, Nb_old, R_d, Nd_old)

        # 1. Update N_b prima del gas solver.
        nu_b = nucleation_rate_bulk(p, Dg, c_old)
        phi_b = phi_bulk(mb_old, Nb_old)
        N_b = (Nb_old + dt * nu_b) / (1.0 + dt * b_b * phi_b)
        N_b = max(N_b, p.min_number_density)

        # 2. Seed opzionale per nuove bulk bubbles.
        dN_b_positive = max(N_b - Nb_old, 0.0)
        seed_vacancies_added_step = 0.0
        if dN_b_positive > 0.0 and p.bulk_seed_radius_nm > 0.0:
            V_seed = sphere_volume(p.bulk_seed_radius_nm * 1.0e-9)
            seed_vacancies_added_step = dN_b_positive * V_seed / omega_matrix(p)
            nvb += seed_vacancies_added_step
            seed_vacancies_added_total += seed_vacancies_added_step

        # 3. Coefficienti old-step per il gas solver.
        # Versione allineata a Codex / operator-splitting SCIANTIX-like:
        # N_b viene aggiornato per la microstruttura, ma g_b nel gas solver
        # resta quello calcolato con Nb_old all'inizio dello step.
        g_b = g_b_old
        trapping_parts = dict(trapping_parts_old)
        trapping_parts["g_b_used_mode"] = "old_step"
        trapping_parts["g_b_old_before_Nb_update"] = g_b_old
        trapping_parts["g_b_used_in_gas_solver"] = g_b
        trapping_parts["Nb_old"] = Nb_old
        trapping_parts["Nb_updated_for_microstructure"] = N_b

        # 4. Gas solver.
        c_new, mb_new, md_new = sciantix_3x3_exchange_step(
            modes_c, modes_mb, modes_md,
            Dg, p.grain_radius, beta,
            g_b, g_d, b_b, b_d, dt,
        )

        # Protezione contro piccoli valori negativi numerici.
        c_new = max(c_new, 0.0)
        mb_new = max(mb_new, 0.0)
        md_new = max(md_new, 0.0)

        dmb_dt = (mb_new - mb_old) / dt
        dmd_dt = (md_new - md_old) / dt

        # 5. Vacancy update.
        if p.update_bulk_vacancies:
            nvb, dnvb_dt = vacancy_concentration_implicit_step(p, Dv, R_b, N_b, mb_new, nvb, dt)
        else:
            dnvb_dt = 0.0

        nvd, dnvd_dt = vacancy_concentration_implicit_step(p, Dv, R_d, Nd_old, md_new, nvd, dt)

        # 6. Volume growth prima della coalescenza.
        if N_b > 0.0:
            V_b_growth = V_b + dt * (
                p.omega_fg / N_b * dmb_dt + omega_matrix(p) / N_b * dnvb_dt
            )
            V_b_growth = max(V_b_growth, p.min_volume)
        else:
            V_b_growth = V_b

        if Nd_old > 0.0:
            dVd_growth_dt = p.omega_fg / Nd_old * dmd_dt + omega_matrix(p) / Nd_old * dnvd_dt
            V_d_growth = V_d + dt * dVd_growth_dt
            V_d_growth = max(V_d_growth, p.min_volume)
        else:
            dVd_growth_dt = 0.0
            V_d_growth = V_d

        # 7. Coalescenza dislocation bubbles.
        lambda_d = coalescence_lambda(Vd_old, Nd_old)
        dVd_positive = max(V_d_growth - Vd_old, 0.0)
        if dVd_positive > 0.0 and Nd_old > 0.0:
            N_d = Nd_old / (1.0 + 4.0 * lambda_d * Nd_old * dVd_positive)
        else:
            N_d = Nd_old
        N_d = max(N_d, p.min_number_density)

        # 8. Ricostruzione volumi da gas + vacancies.
        if N_b > 0.0:
            V_b = (p.omega_fg * max(mb_new, 0.0) + omega_matrix(p) * nvb) / N_b
        else:
            V_b = 0.0

        if N_d > 0.0:
            V_d = (p.omega_fg * max(md_new, 0.0) + omega_matrix(p) * nvd) / N_d
        else:
            V_d = 0.0

        V_b = max(V_b, p.min_volume)
        V_d = max(V_d, p.min_volume)
        R_b = radius_from_volume(V_b)
        R_d = radius_from_volume(V_d)

        # 9. Bilancio gas a grain face.
        generated += beta * dt
        retained = c_new + mb_new + md_new
        q_gb_old = q_gb
        q_gb = max(initial_gas + generated - retained, 0.0)
        qdot_gb = max((q_gb - q_gb_old) / dt, 0.0)

        hist["time"].append(t_old + dt)
        hist["burnup_percent_fima"].append(time_to_burnup_percent(t_old + dt, p.fission_rate, p.lattice_parameter))
        hist["c"].append(c_new)
        hist["mb"].append(mb_new)
        hist["md"].append(md_new)
        hist["Nb"].append(N_b)
        hist["Nd"].append(N_d)
        hist["Vb"].append(V_b)
        hist["Vd"].append(V_d)
        hist["Rb"].append(R_b)
        hist["Rd"].append(R_d)
        hist["nvb"].append(nvb)
        hist["nvd"].append(nvd)
        hist["generated"].append(generated)
        hist["retained"].append(retained)
        hist["q_gb"].append(q_gb)
        hist["qdot_gb"].append(qdot_gb)
        hist["swelling_b"].append(N_b * V_b)
        hist["swelling_d"].append(N_d * V_d)
        hist["swelling_ig"].append(N_b * V_b + N_d * V_d)
        hist["p_b"].append(pressure_internal(p, mb_new, nvb))
        hist["p_d"].append(pressure_internal(p, md_new, nvd))
        hist["p_b_eq"].append(pressure_equilibrium(p, R_b))
        hist["p_d_eq"].append(pressure_equilibrium(p, R_d))
        hist["lambda_d"].append(lambda_d)
        hist["nu_b"].append(nu_b)
        hist["phi_b"].append(phi_b)
        hist["dnvb_dt"].append(dnvb_dt)
        hist["dnvd_dt"].append(dnvd_dt)
        hist["dVd_growth_dt"].append(dVd_growth_dt)
        hist["seed_vacancies_added_total"].append(seed_vacancies_added_total)

        last_rates = {
            "Dg": Dg,
            "Dv": Dv,
            "beta": beta,
            "g_b": g_b,
            "g_b_old_before_Nb_update": g_b_old,
            "g_d": g_d,
            "b_b": b_b,
            "b_d": b_d,
            "nu_b": nu_b,
            "phi_b": phi_b,
            "lambda_d": lambda_d,
            "dVd_growth_dt": dVd_growth_dt,
            "dnvb_dt": dnvb_dt,
            "dnvd_dt": dnvd_dt,
            "seed_vacancies_added_step": seed_vacancies_added_step,
            "seed_vacancies_added_total": seed_vacancies_added_total,
            "p_b": pressure_internal(p, mb_new, nvb),
            "p_d": pressure_internal(p, md_new, nvd),
            "p_b_eq": pressure_equilibrium(p, R_b),
            "p_d_eq": pressure_equilibrium(p, R_d),
            "Nb_old": Nb_old,
            "Nb_updated_before_gas": N_b,
            "Nd_old": Nd_old,
            "Nd_after_coalescence": N_d,
            **D_parts,
            **Dv_parts,
            **trapping_parts,
        }

    return hist, last_rates

# ============================================================
# WRAPPER, SCORE E TEST DIAGNOSTICI PHASE A/B/C/D
# ============================================================

_RUN_CACHE = {}


def run_model_point(
    T,
    burnup,
    f_n,
    K_d=K_D_FIXED,
    gd_scale=1.0,
    gb_scale=1.0,
    bulk_seed_radius_nm=BULK_SEED_RADIUS_NM,
    dt_h=DT_SCAN_H,
    n_modes=N_MODES_SCAN,
):
    key = (
        round(float(T), 6),
        round(float(burnup), 6),
        float(f_n),
        float(K_d),
        float(gd_scale),
        float(gb_scale),
        float(bulk_seed_radius_nm),
        float(dt_h),
        int(n_modes),
    )
    if key in _RUN_CACHE:
        return _RUN_CACHE[key]

    p = UNParameters(
        temperature=float(T),
        fission_rate=FISSION_RATE,
        grain_radius=GRAIN_RADIUS,
        target_burnup_percent_fima=float(burnup),
        dt=float(dt_h) * 3600.0,
        n_modes=int(n_modes),
        xe_yield=XE_YIELD,
        f_n=float(f_n),
        K_d=float(K_d),
        g_d_scale=float(gd_scale),
        g_b_scale=float(gb_scale),
        R_b=0.0,
        N_b=0.0,
        R_d=0.0,
        N_d=None,
        c0=0.0,
        mb0=0.0,
        md0=0.0,
        vacancy_absorption_only=True,
        update_bulk_vacancies=True,
        bulk_seed_radius_nm=float(bulk_seed_radius_nm),
    )

    hist, rates = solve_UN_sciantix_intragranular_extended(p)

    generated = hist["generated"][-1]
    if generated > 0.0:
        matrix_gas_percent = 100.0 * hist["c"][-1] / generated
        bulk_gas_percent = 100.0 * hist["mb"][-1] / generated
        dislocation_gas_percent = 100.0 * hist["md"][-1] / generated
        qgb_gas_percent = 100.0 * hist["q_gb"][-1] / generated
    else:
        matrix_gas_percent = bulk_gas_percent = dislocation_gas_percent = qgb_gas_percent = 0.0

    p_b_eq = hist["p_b_eq"][-1]
    p_d_eq = hist["p_d_eq"][-1]

    row = {
        "T": float(T),
        "burnup": float(burnup),
        "f_n": float(f_n),
        "K_d": float(K_d),
        "gd_scale": float(gd_scale),
        "gb_scale": float(gb_scale),
        "bulk_seed_radius_nm": float(bulk_seed_radius_nm),
        "swelling_b_percent": 100.0 * hist["swelling_b"][-1],
        "swelling_d_percent": 100.0 * hist["swelling_d"][-1],
        "swelling_ig_percent": 100.0 * hist["swelling_ig"][-1],
        "Nb": hist["Nb"][-1],
        "Nd": hist["Nd"][-1],
        "Rb_nm": hist["Rb"][-1] * 1.0e9,
        "Rd_nm": hist["Rd"][-1] * 1.0e9,
        "p_b": hist["p_b"][-1],
        "p_b_eq": p_b_eq,
        "p_d": hist["p_d"][-1],
        "p_d_eq": p_d_eq,
        "p_b_over_eq": hist["p_b"][-1] / p_b_eq if p_b_eq > 0.0 else math.nan,
        "p_d_over_eq": hist["p_d"][-1] / p_d_eq if p_d_eq > 0.0 else math.nan,
        "matrix_gas_percent": matrix_gas_percent,
        "bulk_gas_percent": bulk_gas_percent,
        "dislocation_gas_percent": dislocation_gas_percent,
        "qgb_gas_percent": qgb_gas_percent,
        "Dg": rates.get("Dg", math.nan),
        "Dv": rates.get("Dv", math.nan),
        "g_b": rates.get("g_b", math.nan),
        "g_d": rates.get("g_d", math.nan),
        "g_b_unscaled": rates.get("g_b_unscaled", math.nan),
        "g_d_unscaled": rates.get("g_d_unscaled", math.nan),
        "hist": hist,
        "rates": rates,
    }

    _RUN_CACHE[key] = row
    return row


def rmse(values):
    values = [v for v in values if math.isfinite(v)]
    if not values:
        return math.nan
    return math.sqrt(sum(v * v for v in values) / len(values))


def log10_rmse(pairs):
    errs = []
    for pred, exp in pairs:
        if pred > 0.0 and exp > 0.0 and math.isfinite(pred) and math.isfinite(exp):
            errs.append(math.log10(pred / exp))
    return rmse(errs)


EXP_SWELLING_SUBSET = [
    p for p in EXP_SWELLING_T
    if (
        (abs(p["burnup"] - 1.1) < 1e-9 and p["T"] in {1127.0, 1485.0, 1598.0, 1669.0})
        or (abs(p["burnup"] - 1.3) < 1e-9 and p["T"] in {1044.0, 1377.0, 1595.0, 1639.0, 1724.0})
        or (abs(p["burnup"] - 3.2) < 1e-9 and p["T"] in {984.0, 1247.0, 1511.0, 1590.0})
    )
]

EXP_ND_SUBSET = [p for p in EXP_ND_T_13 if p["T"] in {1153.0, 1330.0, 1493.0, 1599.0, 1669.0, 1723.0}]
EXP_RD_SUBSET = [p for p in EXP_RD_T_13 if p["T"] in {1045.0, 1374.0, 1535.0, 1594.0, 1641.0, 1725.0}]


def score_combo_general(f_n, K_d=K_D_FIXED, gd_scale=1.0, gb_scale=1.0, bulk_seed_radius_nm=BULK_SEED_RADIUS_NM, full=False):
    """Score diagnostico unico. full=False usa subset leggero per griglie larghe."""
    swelling_points = EXP_SWELLING_T if full else EXP_SWELLING_SUBSET
    nd_points = EXP_ND_T_13 if full else EXP_ND_SUBSET
    rd_points = EXP_RD_T_13 if full else EXP_RD_SUBSET

    swelling_errors = []
    for exp in swelling_points:
        out = run_model_point(exp["T"], exp["burnup"], f_n, K_d=K_d, gd_scale=gd_scale, gb_scale=gb_scale, bulk_seed_radius_nm=bulk_seed_radius_nm)
        swelling_errors.append(out["swelling_d_percent"] - exp["swelling"])
    score_swelling = rmse(swelling_errors)

    n_pairs = []
    for exp in nd_points:
        out = run_model_point(exp["T"], 1.3, f_n, K_d=K_d, gd_scale=gd_scale, gb_scale=gb_scale, bulk_seed_radius_nm=bulk_seed_radius_nm)
        n_pairs.append((out["Nd"], exp["N"]))
    score_log_N = log10_rmse(n_pairs)

    r_pairs = []
    for exp in rd_points:
        out = run_model_point(exp["T"], 1.3, f_n, K_d=K_d, gd_scale=gd_scale, gb_scale=gb_scale, bulk_seed_radius_nm=bulk_seed_radius_nm)
        r_pairs.append((out["Rd_nm"], exp["R_nm"]))
    score_log_R = log10_rmse(r_pairs)

    pressure_cases = [(984.0, 3.2), (1044.0, 1.3), (1200.0, 3.2)]
    pressure_terms = []
    for T, bu in pressure_cases:
        out = run_model_point(T, bu, f_n, K_d=K_d, gd_scale=gd_scale, gb_scale=gb_scale, bulk_seed_radius_nm=bulk_seed_radius_nm)
        for key in ["p_b_over_eq", "p_d_over_eq"]:
            val = out[key]
            if val > 0.0 and math.isfinite(val):
                pressure_terms.append(abs(math.log10(max(val, 1.0e-30))))
    pressure_penalty = rmse(pressure_terms) if pressure_terms else 0.0

    out_1600_32 = run_model_point(1600.0, 3.2, f_n, K_d=K_d, gd_scale=gd_scale, gb_scale=gb_scale, bulk_seed_radius_nm=bulk_seed_radius_nm)
    bulk_fraction = out_1600_32["bulk_gas_percent"] / 100.0
    disl_fraction = out_1600_32["dislocation_gas_percent"] / 100.0
    partition_penalty = 0.0
    if bulk_fraction > 0.75:
        partition_penalty += bulk_fraction - 0.75
    if disl_fraction < 0.05:
        partition_penalty += 0.05 - disl_fraction

    total_score = (
        score_swelling
        + 0.75 * score_log_N
        + 0.75 * score_log_R
        + 0.30 * pressure_penalty
        + 0.50 * partition_penalty
    )

    out_1600_13 = run_model_point(1600.0, 1.3, f_n, K_d=K_d, gd_scale=gd_scale, gb_scale=gb_scale, bulk_seed_radius_nm=bulk_seed_radius_nm)

    return {
        "f_n": f_n,
        "K_d": K_d,
        "gd_scale": gd_scale,
        "gb_scale": gb_scale,
        "bulk_seed_radius_nm": bulk_seed_radius_nm,
        "score_total": total_score,
        "score_swelling_rmse_percent": score_swelling,
        "score_log10_N": score_log_N,
        "score_log10_R": score_log_R,
        "pressure_penalty": pressure_penalty,
        "partition_penalty": partition_penalty,
        "sw_d_32_1600": out_1600_32["swelling_d_percent"],
        "sw_b_32_1600": out_1600_32["swelling_b_percent"],
        "pb_ratio_32_1600": out_1600_32["p_b_over_eq"],
        "pd_ratio_32_1600": out_1600_32["p_d_over_eq"],
        "bulk_gas_32_1600": out_1600_32["bulk_gas_percent"],
        "disl_gas_32_1600": out_1600_32["dislocation_gas_percent"],
        "Rd_13_1600": out_1600_13["Rd_nm"],
        "Nd_13_1600": out_1600_13["Nd"],
    }


def print_ranking(rows, title, n=25):
    rows_sorted = sorted(rows, key=lambda r: r["score_total"])
    print("\n" + "=" * 180)
    print(title)
    print("=" * 180)
    print(
        f"{'rank':>4s} | {'score':>11s} | {'swRMSE':>10s} | {'logN':>8s} | {'logR':>8s} | "
        f"{'pPen':>8s} | {'partPen':>8s} | {'f_n':>10s} | {'K_d':>10s} | {'gd':>6s} | {'seed':>6s} | "
        f"{'swD32@1600':>11s} | {'swB32@1600':>11s} | {'pb/peq':>9s} | {'pd/peq':>9s} | "
        f"{'bulkGas%':>9s} | {'dislGas%':>9s} | {'Rd13@1600':>10s} | {'Nd13@1600':>11s}"
    )
    print("-" * 180)
    for i, r in enumerate(rows_sorted[:n], start=1):
        print(
            f"{i:4d} | {r['score_total']:11.4e} | {r['score_swelling_rmse_percent']:10.4e} | "
            f"{r['score_log10_N']:8.3e} | {r['score_log10_R']:8.3e} | "
            f"{r['pressure_penalty']:8.3e} | {r['partition_penalty']:8.3e} | "
            f"{r['f_n']:10.2e} | {r['K_d']:10.2e} | {r['gd_scale']:6.2g} | {r['bulk_seed_radius_nm']:6.2f} | "
            f"{r['sw_d_32_1600']:11.4e} | {r['sw_b_32_1600']:11.4e} | "
            f"{r['pb_ratio_32_1600']:9.3e} | {r['pd_ratio_32_1600']:9.3e} | "
            f"{r['bulk_gas_32_1600']:9.3e} | {r['disl_gas_32_1600']:9.3e} | "
            f"{r['Rd_13_1600']:10.3e} | {r['Nd_13_1600']:11.3e}"
        )
    return rows_sorted


# ============================================================
# OUTPUT E GRAFICI DIAGNOSTICI PER UN SET SCELTO
# ============================================================

def print_key_table(title, rows):
    print("\n" + "=" * 180)
    print(title)
    print("=" * 180)
    print(
        f"{'T_K':>8s} | {'sw_b_%':>10s} | {'sw_d_%':>10s} | {'sw_ig_%':>10s} | "
        f"{'Rb_nm':>10s} | {'Rd_nm':>10s} | {'Nb':>12s} | {'Nd':>12s} | "
        f"{'pb/peq':>10s} | {'pd/peq':>10s} | "
        f"{'matrix_%':>10s} | {'bulkGas_%':>10s} | {'dislGas_%':>10s} | {'qgb_%':>10s}"
    )
    print("-" * 180)
    wanted = {900.0, 1000.0, 1200.0, 1400.0, 1500.0, 1600.0, 1700.0, 1800.0, 2000.0}
    for r in rows:
        if r["T"] in wanted:
            print(
                f"{r['T']:8.1f} | "
                f"{r['swelling_b_percent']:10.4e} | {r['swelling_d_percent']:10.4e} | {r['swelling_ig_percent']:10.4e} | "
                f"{r['Rb_nm']:10.4e} | {r['Rd_nm']:10.4e} | "
                f"{r['Nb']:12.5e} | {r['Nd']:12.5e} | "
                f"{r['p_b_over_eq']:10.4e} | {r['p_d_over_eq']:10.4e} | "
                f"{r['matrix_gas_percent']:10.4e} | {r['bulk_gas_percent']:10.4e} | "
                f"{r['dislocation_gas_percent']:10.4e} | {r['qgb_gas_percent']:10.4e}"
            )


def plot_exp_swelling_for_burnup(bu):
    series_names = sorted({p["series"] for p in EXP_SWELLING_T if abs(p["burnup"] - bu) < 1.0e-9})
    for series in series_names:
        pts = [p for p in EXP_SWELLING_T if abs(p["burnup"] - bu) < 1.0e-9 and p["series"] == series]
        marker = "x" if "119" in series else "^"
        plt.scatter([p["T"] for p in pts], [p["swelling"] for p in pts], marker=marker, s=55, label=f"Exp {series}")


def make_diagnostics(label, f_n, K_d, gd_scale, bulk_seed_radius_nm=0.0):
    T_PLOT = [float(T) for T in range(900, 2001, 50)]
    T_AMOUNT = [float(T) for T in range(900, 2601, 100)]
    BURNUPS = [1.1, 1.3, 3.2]

    print("\n" + "#" * 150)
    print(f"DIAGNOSTICS: {label}")
    print(f"f_n={f_n:.3e}, K_d={K_d:.3e}, gd_scale={gd_scale:.3g}, bulk_seed={bulk_seed_radius_nm:.2f} nm")
    print("#" * 150)

    results_by_burnup = {}
    for bu in BURNUPS:
        rows = [run_model_point(T, bu, f_n, K_d=K_d, gd_scale=gd_scale, bulk_seed_radius_nm=bulk_seed_radius_nm) for T in T_PLOT]
        results_by_burnup[bu] = rows
        print_key_table(f"DETAILED TABLE - {label}, burnup={bu:.1f}% FIMA", rows)

    gas_partition = {
        bu: [run_model_point(T, bu, f_n, K_d=K_d, gd_scale=gd_scale, bulk_seed_radius_nm=bulk_seed_radius_nm) for T in T_AMOUNT]
        for bu in [1.1, 3.2]
    }

    if plt is None:
        print("\nmatplotlib non installato: grafici saltati.")
        return

    for bu in BURNUPS:
        rows = results_by_burnup[bu]
        Ts = [r["T"] for r in rows]
        plt.figure(figsize=(8, 5))
        plt.plot(Ts, [r["swelling_d_percent"] for r in rows], label="Model dislocation")
        plt.plot(Ts, [r["swelling_b_percent"] for r in rows], linestyle=":", label="Model bulk")
        plot_exp_swelling_for_burnup(bu)
        plt.xlabel("T [K]")
        plt.ylabel("Fission gas swelling [%]")
        plt.title(f"Swelling vs T {bu:.1f}% FIMA - {label}")
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()

    rows = results_by_burnup[1.3]
    Ts = [r["T"] for r in rows]
    plt.figure(figsize=(8, 5))
    plt.plot(Ts, [r["Nd"] for r in rows], label="Model dislocation N_d")
    plt.plot(Ts, [r["Nb"] for r in rows], linestyle=":", label="Model bulk N_b")
    plt.scatter([p["T"] for p in EXP_ND_T_13], [p["N"] for p in EXP_ND_T_13], marker="^", s=55, label="Exp large bubbles")
    plt.yscale("log")
    plt.xlabel("T [K]")
    plt.ylabel("Bubble concentration [m$^{-3}$]")
    plt.title(f"Bubble concentration 1.3% FIMA - {label}")
    plt.grid(True, which="both", alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(8, 5))
    plt.plot(Ts, [r["Rd_nm"] for r in rows], label="Model dislocation R_d")
    plt.plot(Ts, [r["Rb_nm"] for r in rows], linestyle=":", label="Model bulk R_b")
    plt.scatter([p["T"] for p in EXP_RD_T_13], [p["R_nm"] for p in EXP_RD_T_13], marker="^", s=55, label="Exp large bubbles")
    plt.yscale("log")
    plt.xlabel("T [K]")
    plt.ylabel("Bubble radius [nm]")
    plt.title(f"Bubble radius 1.3% FIMA - {label}")
    plt.grid(True, which="both", alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

    rows = results_by_burnup[3.2]
    Ts = [r["T"] for r in rows]
    plt.figure(figsize=(8, 5))
    plt.plot(Ts, [r["p_b"] for r in rows], label="Bulk pressure")
    plt.plot(Ts, [r["p_b_eq"] for r in rows], linestyle=":", label="Bulk equilibrium")
    plt.plot(Ts, [r["p_d"] for r in rows], label="Dislocation pressure")
    plt.plot(Ts, [r["p_d_eq"] for r in rows], linestyle="--", label="Dislocation equilibrium")
    plt.yscale("log")
    plt.xlabel("T [K]")
    plt.ylabel("Bubble pressure [Pa]")
    plt.title(f"Pressure 3.2% FIMA - {label}")
    plt.grid(True, which="both", alpha=0.3)
    plt.legend(fontsize=8)
    plt.tight_layout()
    plt.show()

    for bu, rows in gas_partition.items():
        Ts = [r["T"] for r in rows]
        plt.figure(figsize=(8, 5))
        plt.plot(Ts, [r["matrix_gas_percent"] for r in rows], label="Matrix")
        plt.plot(Ts, [r["bulk_gas_percent"] for r in rows], linestyle=":", label="Bulk bubbles")
        plt.plot(Ts, [r["dislocation_gas_percent"] for r in rows], linestyle="--", label="Dislocation bubbles")
        plt.plot(Ts, [r["qgb_gas_percent"] for r in rows], linestyle="-.", label="Gas to grain face q_gb")
        plt.xlabel("T [K]")
        plt.ylabel("Amount of generated gas [%]")
        plt.title(f"Gas partition {bu:.1f}% FIMA - {label}")
        plt.xlim(900, 2600)
        plt.ylim(0, 100)
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.tight_layout()
        plt.show()


# ============================================================
# MAIN
# ============================================================

if RUN_SINGLE_TEST:
    test = run_model_point(1600.0, 1.3, 3.0e-9)
    print(test)

if RUN_CALIBRATION:
    print("\n" + "#" * 150)
    print("START UN MODEL SENSITIVITY SUITE: f_n, g_d_scale, K_d, bulk seed")
    print("#" * 150)
    print(f"RUN_MODE              = {RUN_MODE}")
    print(f"dt scan               = {DT_SCAN_H:.2f} h")
    print(f"n_modes               = {N_MODES_SCAN:d}")
    print(f"Xe yield              = {XE_YIELD:.6e}")
    print(f"baseline K_d          = {K_D_FIXED:.6e} bub/m")
    print(f"baseline seed         = {BULK_SEED_RADIUS_NM:.3f} nm")
    print("Target principale: experimental large/P2 bubbles -> model dislocation bubbles.")

    phase_a_sorted = []
    if RUN_PHASE_A_FN_ONLY:
        print("\nPHASE A: f_n only, K_d fixed, gd_scale=1, seed baseline")
        phase_a = []
        for f_n in FN_GRID:
            print(f"  Running Phase A f_n={f_n:.3e}")
            phase_a.append(score_combo_general(f_n, K_d=K_D_FIXED, gd_scale=1.0, bulk_seed_radius_nm=BULK_SEED_RADIUS_NM, full=True))
        phase_a_sorted = print_ranking(phase_a, "PHASE A - f_n ONLY, FULL DIGITALIZED SCORE", n=len(phase_a))

    phase_b_sorted = []
    if RUN_PHASE_B_GRID:
        print("\nPHASE B: grid f_n x gd_scale x K_d, no bulk seed, subset score")
        phase_b = []
        total = len(FN_GRID_B) * len(GD_SCALE_GRID) * len(K_D_GRID)
        count = 0
        for f_n in FN_GRID_B:
            for K_d in K_D_GRID:
                for gd in GD_SCALE_GRID:
                    count += 1
                    print(f"  Phase B {count}/{total}: f_n={f_n:.2e}, K_d={K_d:.2e}, gd={gd:g}")
                    phase_b.append(score_combo_general(f_n, K_d=K_d, gd_scale=gd, bulk_seed_radius_nm=0.0, full=False))
        phase_b_sorted = print_ranking(phase_b, "PHASE B - f_n x gd_scale x K_d, NO SEED, SUBSET SCORE", n=40)

        print("\nPHASE B2: re-score top Phase B with full digitalized score")
        top = phase_b_sorted[:12]
        phase_b2 = []
        for r in top:
            print(f"  Full re-score: f_n={r['f_n']:.2e}, K_d={r['K_d']:.2e}, gd={r['gd_scale']:g}")
            phase_b2.append(score_combo_general(r["f_n"], K_d=r["K_d"], gd_scale=r["gd_scale"], bulk_seed_radius_nm=0.0, full=True))
        phase_b_sorted = print_ranking(phase_b2, "PHASE B2 - TOP GRID FULL DIGITALIZED SCORE", n=len(phase_b2))

    phase_c_sorted = []
    if RUN_PHASE_C_FIXED_FN_GD:
        print("\nPHASE C: fixed f_n values, vary gd_scale and K_d, full score")
        fixed_fns = []
        if phase_a_sorted:
            fixed_fns.append(phase_a_sorted[0]["f_n"])
        for val in [3.0e-9, 1.0e-8, 1.0e-7, 1.0e-6]:
            if val not in fixed_fns:
                fixed_fns.append(val)
        phase_c = []
        for f_n in fixed_fns:
            for K_d in K_D_GRID:
                for gd in GD_SCALE_GRID:
                    print(f"  Phase C: f_n={f_n:.2e}, K_d={K_d:.2e}, gd={gd:g}")
                    phase_c.append(score_combo_general(f_n, K_d=K_d, gd_scale=gd, bulk_seed_radius_nm=0.0, full=True))
        phase_c_sorted = print_ranking(phase_c, "PHASE C - FIXED f_n OPTIONS, FULL DIGITALIZED SCORE", n=40)

    phase_d_sorted = []
    if RUN_PHASE_D_BULK_SEED:
        print("\nPHASE D: bulk seed sensitivity on best candidates")
        candidates = []
        seen = set()
        for source in [phase_b_sorted[:5], phase_c_sorted[:5], phase_a_sorted[:3]]:
            for r in source:
                key = (r["f_n"], r["K_d"], r["gd_scale"])
                if key not in seen:
                    seen.add(key)
                    candidates.append({"f_n": r["f_n"], "K_d": r["K_d"], "gd_scale": r["gd_scale"]})
        phase_d = []
        for c in candidates[:10]:
            for seed in BULK_SEED_GRID:
                print(f"  Phase D: f_n={c['f_n']:.2e}, K_d={c['K_d']:.2e}, gd={c['gd_scale']:g}, seed={seed:.1f} nm")
                phase_d.append(score_combo_general(c["f_n"], K_d=c["K_d"], gd_scale=c["gd_scale"], bulk_seed_radius_nm=seed, full=True))
        phase_d_sorted = print_ranking(phase_d, "PHASE D - BULK SEED SENSITIVITY, FULL DIGITALIZED SCORE", n=40)

    all_ranked = []
    for arr in [phase_a_sorted[:3], phase_b_sorted[:5], phase_c_sorted[:5], phase_d_sorted[:5]]:
        all_ranked.extend(arr)
    all_ranked = sorted(all_ranked, key=lambda r: r["score_total"])

    if all_ranked:
        best = all_ranked[0]
        make_diagnostics("BEST_OVERALL", best["f_n"], best["K_d"], best["gd_scale"], best["bulk_seed_radius_nm"])

    if phase_a_sorted:
        base = phase_a_sorted[0]
        make_diagnostics("BEST_PHASE_A_FN_ONLY", base["f_n"], base["K_d"], base["gd_scale"], base["bulk_seed_radius_nm"])

    print("\n" + "#" * 150)
    print("END UN MODEL SENSITIVITY SUITE")
    print("#" * 150)
    print("Mandami il notebook/output completo: controlleremo soprattutto PHASE B2, PHASE C e PHASE D.")
````
